# Portfolio Variance $w^\top \Sigma w$

The two-asset variance formula from module 03, written once for any number of assets. One
matrix product replaces a growing pile of covariance terms, and makes the limits of
diversification visible.

## 1. Introduction

With two assets the portfolio variance has three terms. With 3 assets it has 9, with 20 it has
400, with the 500 stocks of a broad index 250,000. Writing them out stops being an option.

Linear algebra packs all of them into one expression:

$$
\sigma_p^2 = w^\top \Sigma\, w
$$

After this notebook you can:

- build $\Sigma$ from volatilities and a correlation matrix;
- compute expected return, variance, and the covariance between two portfolios;
- read the variance as a sum of contributions and see which pairs dominate;
- explain why diversification stops at a floor set by the average correlation;
- compute tracking error as $(w-b)^\top \Sigma (w-b)$.

## 2. Theory

### Notation

$n$ assets with random returns collected in a column vector $R = (R_1, \dots, R_n)^\top$.

- Expected returns: $\mu = E[R]$, an $n$-vector.
- Covariance matrix: $\Sigma = E\big[(R-\mu)(R-\mu)^\top\big]$, an $n \times n$ matrix with
  $\Sigma_{ii} = \sigma_i^2$ and $\Sigma_{ij} = \rho_{ij}\,\sigma_i\sigma_j$. It is symmetric.
- Weights: column vector $w$ with $\mathbf{1}^\top w = 1$ for a fully invested portfolio.

### $\Sigma$ from volatilities and correlations

With $D = \operatorname{diag}(\sigma_1, \dots, \sigma_n)$ and correlation matrix $C$ ($C_{ii} = 1$, $C_{ij} = \rho_{ij}$):

$$
\Sigma = D\,C\,D
$$

This is how risk models are usually specified: volatilities and correlations are easier to
reason about than raw covariances.

### Portfolio return, mean and variance

The portfolio return is the dot product $R_p = w^\top R = \sum_i w_i R_i$. Then

$$
E[R_p] = w^\top \mu
$$

$$
\operatorname{Var}(R_p) = E\big[(w^\top (R-\mu))^2\big]
= E\big[w^\top (R-\mu)(R-\mu)^\top w\big]
= w^\top\, E\big[(R-\mu)(R-\mu)^\top\big]\, w
= w^\top \Sigma\, w
$$

The second step writes the square of a scalar as the scalar times its own transpose;
the third pulls the constant $w$ out of the expectation.

### The double sum behind the matrix

$$
w^\top \Sigma\, w = \sum_{i=1}^n \sum_{j=1}^n w_i w_j \Sigma_{ij}
= \underbrace{\sum_i w_i^2 \sigma_i^2}_{n \text{ variance terms}}
+ \underbrace{\sum_{i \ne j} w_i w_j \rho_{ij}\sigma_i\sigma_j}_{n(n-1) \text{ covariance terms}}
$$

For $n = 2$, $w = (w_1, w_2)$, this is exactly 03/01:
$w_1^2\sigma_1^2 + w_2^2\sigma_2^2 + 2w_1w_2\rho\,\sigma_1\sigma_2$.

### Covariance between two portfolios

For weight vectors $w$ and $v$ on the same assets:

$$
\operatorname{Cov}(w^\top R,\ v^\top R) = w^\top \Sigma\, v
$$

**Tracking error** is the volatility of the active return $(w - b)^\top R$ against a
benchmark $b$:

$$
TE^2 = (w-b)^\top \Sigma\, (w-b)
$$

the $n$-asset version of 03/01's $\sigma_F^2 + \sigma_B^2 - 2\rho\sigma_F\sigma_B$.

### The limit of diversification

Take $n$ assets with equal volatility $\sigma$, equal pairwise correlation $\rho$, and equal
weights $w_i = 1/n$. Of the $n^2$ terms, $n$ are variances and $n(n-1)$ covariances:

$$
\sigma_p^2 = n \cdot \frac{1}{n^2}\sigma^2 + n(n-1)\cdot\frac{1}{n^2}\rho\sigma^2
= \frac{\sigma^2}{n} + \Big(1 - \frac{1}{n}\Big)\rho\,\sigma^2
\;\xrightarrow{\;n\to\infty\;}\; \rho\,\sigma^2
$$

The variance terms vanish as $1/n$; the covariance terms do not. What remains is
**systematic risk**, $\sigma\sqrt{\rho}$, which no amount of diversification removes.

### Units

$\Sigma$ inherits the periodicity of the returns. With independent periods, variances and
covariances scale linearly with time: $\Sigma_{\text{annual}} = \text{periods\_per\_year} \cdot \Sigma_{\text{period}}$.

## 3. Worked Examples

Three asset classes for a CHF investor, annual figures:

| | $\sigma$ | Swiss equities | Global equities | Swiss bonds |
|---|---|---|---|---|
| Swiss equities | 16% | 1 | 0.8 | 0.1 |
| Global equities (in CHF) | 17% | 0.8 | 1 | 0.0 |
| Swiss bonds | 5% | 0.1 | 0.0 | 1 |

### Example 1: building $\Sigma$

$\Sigma_{ij} = \rho_{ij}\sigma_i\sigma_j$:

$$
\Sigma = \begin{pmatrix}
0.0256 & 0.02176 & 0.0008 \\
0.02176 & 0.0289 & 0 \\
0.0008 & 0 & 0.0025
\end{pmatrix}
$$

### Example 2: portfolio variance, term by term

$w = (0.4,\ 0.3,\ 0.3)^\top$.

| Term | Value |
|---|---|
| $w_1^2\Sigma_{11} = 0.16 \cdot 0.0256$ | 0.004096 |
| $w_2^2\Sigma_{22} = 0.09 \cdot 0.0289$ | 0.002601 |
| $w_3^2\Sigma_{33} = 0.09 \cdot 0.0025$ | 0.000225 |
| $2w_1w_2\Sigma_{12} = 2 \cdot 0.12 \cdot 0.02176$ | 0.005222 |
| $2w_1w_3\Sigma_{13} = 2 \cdot 0.12 \cdot 0.0008$ | 0.000192 |
| $2w_2w_3\Sigma_{23}$ | 0 |
| **$w^\top\Sigma w$** | **0.012336** |

$\sigma_p = \sqrt{0.012336} = 11.1\%$. The single largest term is the **covariance** between the
two equity blocks (42% of the total), not any asset's own variance.

### Example 3: tracking error

Benchmark $b = (0.3,\ 0.4,\ 0.3)^\top$, so the active weights are $a = w - b = (0.1,\ -0.1,\ 0)^\top$:

$$
TE^2 = 0.01 \cdot 0.0256 + 0.01 \cdot 0.0289 - 2 \cdot 0.01 \cdot 0.02176 = 0.0001098,
\qquad TE = 1.05\%
$$

A 10-percentage-point tilt from global to Swiss equities costs only about 1% tracking error,
because the two are 0.8 correlated: it is a difference of correlated returns (03/01).

### Example 4: how many stocks is enough?

Equal-weighted stocks, each $\sigma = 30\%$, pairwise $\rho = 0.3$:

| $n$ | 1 | 10 | 30 | $\infty$ |
|---|---|---|---|---|
| $\sigma_p$ | 30.0% | 18.2% | 17.1% | $30\% \cdot \sqrt{0.3} = 16.4\%$ |

Going from 1 to 10 stocks removes most of what diversification can remove. From 30 to
infinity gains less than one percentage point.

## 4. Finance Application

### Risk models are $\Sigma$

Every portfolio risk system (vendor risk models, a bank's internal model, a pension fund's ALM
tool) is at its core a covariance matrix for the investable universe. Portfolio volatility,
tracking error, VaR (module 05) and risk contributions are all computed from $w$ and $\Sigma$.

### Which positions drive risk

The matrix of terms $w_i w_j \Sigma_{ij}$ shows where the variance comes from. In balanced
portfolios the equity–equity covariance block usually dominates, even when the equity weights
look moderate. Module 05/04 turns this into a per-asset risk contribution.

### Systematic vs idiosyncratic risk

The $\rho\sigma^2$ floor is why the market rewards (in theory) only systematic risk: the
idiosyncratic part can be diversified away almost for free, so no one should be paid for
holding it. This is the starting point of CAPM (04/05).

### Tracking error budgets

Active managers get a tracking-error budget (e.g. 2% against the SMI). Whether a set of active
bets fits the budget is a single quadratic form, $(w-b)^\top\Sigma(w-b)$. Bets between highly
correlated assets are "cheap" in TE; bets into uncorrelated assets are expensive.

### The usual misreadings

- "Portfolio volatility is the weighted average of the asset volatilities." That is the
  $\rho = 1$ case only (03/04); with any lower correlation it overstates risk.
- "Holding many stocks makes a portfolio safe." It removes idiosyncratic risk only. Fifty
  equities are still an equity portfolio with roughly market risk.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
```

### Building $\Sigma = D C D$

```python
assets = ["Swiss equities", "Global equities", "Swiss bonds"]
vols = np.array([0.16, 0.17, 0.05])          # annual, decimals
corr = np.array([
    [1.0, 0.8, 0.1],
    [0.8, 1.0, 0.0],
    [0.1, 0.0, 1.0],
])
mu = np.array([0.065, 0.07, 0.02])           # annual expected returns, decimals

D = np.diag(vols)
Sigma = D @ corr @ D

pd.DataFrame(Sigma, index=assets, columns=assets).round(5)
```

### Expected return and variance of a portfolio

```python
w = np.array([0.4, 0.3, 0.3])

port_mu = w @ mu
port_var = w @ Sigma @ w
print(f"expected return: {port_mu:.2%}")
print(f"variance:        {port_var:.6f}")
print(f"volatility:      {np.sqrt(port_var):.2%}")

# The same number from the explicit double sum
double_sum = sum(w[i] * w[j] * Sigma[i, j] for i in range(3) for j in range(3))
print(f"double sum:      {double_sum:.6f}")
```

### Where the variance comes from

The matrix $w_i w_j \Sigma_{ij}$ (element-wise product of $w w^\top$ and $\Sigma$) sums to
$w^\top\Sigma w$. Row sums give each asset's share once its covariances are included.

```python
contrib = np.outer(w, w) * Sigma
contrib_share = pd.DataFrame(contrib / port_var, index=assets, columns=assets)

print(f"sum of all terms equals w'Σw: {np.isclose(contrib.sum(), port_var)}")
print("\nShare of total variance per term:")
print(contrib_share.round(3))
print("\nRow sums (asset's own variance + its covariances):")
print(contrib_share.sum(axis=1).round(3))
```

### Check by simulation

Draw monthly returns from a multivariate normal with $\Sigma/12$, form the portfolio return
`r_p = r @ w`, and compare its annualized variance with $w^\top\Sigma w$.

```python
periods_per_year = 12
n_months = 200_000

r = rng.multivariate_normal(mu / periods_per_year, Sigma / periods_per_year, size=n_months)
r_p = r @ w

print(f"annualized variance (simulated): {r_p.var(ddof=1) * periods_per_year:.6f}")
print(f"w' Sigma w (theory):             {port_var:.6f}")
```

### Covariance between two portfolios, and tracking error (Example 3)

```python
b = np.array([0.3, 0.4, 0.3])   # benchmark weights
active = w - b

cov_wb = w @ Sigma @ b
var_b = b @ Sigma @ b
tracking_error = np.sqrt(active @ Sigma @ active)

print(f"Cov(portfolio, benchmark): {cov_wb:.6f}")
print(f"correlation:               {cov_wb / np.sqrt(port_var * var_b):.4f}")
print(f"tracking error:            {tracking_error:.2%}")

# Same number via the 03/01 formula for a difference
te_0301 = np.sqrt(port_var + var_b - 2 * cov_wb)
print(f"via Var(w) + Var(b) - 2 Cov: {te_0301:.2%}")

# Simulated active returns
print(f"simulated (annualized):      {(r @ active).std(ddof=1) * np.sqrt(periods_per_year):.2%}")
```

### The diversification floor (Example 4)

```python
def equal_weight_vol(n, sigma, rho):
    # Build the equicorrelated covariance matrix and compute w' Sigma w with w = 1/n
    corr_n = np.full((n, n), rho)
    np.fill_diagonal(corr_n, 1.0)
    Sigma_n = sigma**2 * corr_n
    w_n = np.full(n, 1 / n)
    return np.sqrt(w_n @ Sigma_n @ w_n)


sigma_stock, rho_stock = 0.30, 0.3
for n in [1, 2, 5, 10, 30, 100, 500]:
    closed_form = np.sqrt(sigma_stock**2 / n + (1 - 1 / n) * rho_stock * sigma_stock**2)
    print(f"n = {n:3d}: vol = {equal_weight_vol(n, sigma_stock, rho_stock):.2%}   (closed form {closed_form:.2%})")
print(f"floor sigma * sqrt(rho) = {sigma_stock * np.sqrt(rho_stock):.2%}")
```

### Annualizing $\Sigma$ from daily returns

```python
trading_days = 252
daily_r = rng.multivariate_normal(np.zeros(3), Sigma / trading_days, size=10 * trading_days)

Sigma_daily_hat = np.cov(daily_r, rowvar=False)
Sigma_annual_hat = Sigma_daily_hat * trading_days
vols_hat = np.sqrt(np.diag(Sigma_annual_hat))

pd.DataFrame({"true vol": vols, "estimated vol (10y daily)": vols_hat}, index=assets).round(4)
```

```python
n_grid = np.arange(1, 101)

plt.figure(figsize=(8, 5))
for rho_value in [0.0, 0.1, 0.3, 0.6]:
    vols_n = np.sqrt(sigma_stock**2 / n_grid + (1 - 1 / n_grid) * rho_value * sigma_stock**2)
    line, = plt.plot(n_grid, vols_n, label=f"ρ = {rho_value}")
    plt.axhline(sigma_stock * np.sqrt(rho_value), color=line.get_color(), linestyle=":", alpha=0.7)

plt.title("Equal-weighted portfolio of stocks with 30% volatility (dotted: floor σ√ρ)")
plt.xlabel("number of stocks n")
plt.ylabel("portfolio volatility (annual)")
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

With three assets the set of possible portfolios is no longer a curve. Random long-only weight
vectors fill a region; its upper-left edge is the efficient frontier that 04/03 and 04/04 compute.

```python
n_portfolios = 5_000
random_w = rng.dirichlet(np.ones(3), size=n_portfolios)   # long-only, weights sum to 1

random_mu = random_w @ mu
random_vol = np.sqrt(np.einsum("ij,jk,ik->i", random_w, Sigma, random_w))   # w' Sigma w per row

plt.figure(figsize=(8, 5.5))
sc = plt.scatter(random_vol, random_mu, c=random_w[:, 2], cmap="viridis", s=6, alpha=0.6)
plt.colorbar(sc, label="weight in Swiss bonds")
plt.scatter(vols, mu, color="red", zorder=3)
for name, x, y in zip(assets, vols, mu):
    plt.annotate(name, (x, y), textcoords="offset points", xytext=(6, -4))
plt.scatter(np.sqrt(port_var), port_mu, marker="*", s=200, color="black", zorder=4, label="w = (0.4, 0.3, 0.3)")
plt.xlim(0.04, 0.215)
plt.title("5,000 random long-only portfolios of three assets")
plt.xlabel("annual volatility")
plt.ylabel("annual expected return")
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.1%}"))
plt.grid(True, alpha=0.3)
plt.legend(loc="lower right")
plt.show()
```

## 8. Solutions / Checks

```python
# 1.
w1 = np.array([0.2, 0.2, 0.6])
terms = np.outer(w1, w1) * Sigma
print(f"1. w'Σw = {w1 @ Sigma @ w1:.6f} (sum of terms {terms.sum():.6f}), vol = {np.sqrt(w1 @ Sigma @ w1):.2%}")

# 2.
print(f"2. symmetric: {np.allclose(Sigma, Sigma.T)}")

# 3.
for label, tilt in [("global -> Swiss equities", np.array([0.1, -0.1, 0.0])),
                    ("Swiss bonds -> global equities", np.array([0.0, 0.1, -0.1]))]:
    print(f"3. {label:32s}: TE = {np.sqrt(tilt @ Sigma @ tilt):.2%}")

# 4.
n, sigma_avg, rho_avg = 40, 0.25, 0.4
vol_40 = np.sqrt(sigma_avg**2 / n + (1 - 1 / n) * rho_avg * sigma_avg**2)
print(f"4. vol of 40 stocks = {vol_40:.2%}, floor = {sigma_avg * np.sqrt(rho_avg):.2%}")

# 5.
bad_corr = np.array([
    [1.0, 0.9, 0.9],
    [0.9, 1.0, -0.9],
    [0.9, -0.9, 1.0],
])
w_bad = np.array([1.0, -1.0, -1.0])
print(f"5. w'Cw with w = {w_bad}: {w_bad @ bad_corr @ w_bad:+.2f}")
```

**1.** Terms: $0.04 \cdot 0.0256 + 0.04 \cdot 0.0289 + 0.36 \cdot 0.0025 + 2 \cdot 0.04 \cdot 0.02176 + 2 \cdot 0.12 \cdot 0.0008 + 0$
$= 0.001024 + 0.001156 + 0.0009 + 0.0017408 + 0.000192 = 0.0050128$, volatility $\approx 7.08\%$.

**2.** $\Sigma^\top = (DCD)^\top = D^\top C^\top D^\top = DCD$, since $D$ is diagonal and $C$ symmetric.

**3.** The equity-to-equity tilt is a difference of two 0.8-correlated returns (about 1% TE).
The bonds-to-equities tilt adds equity risk against an almost uncorrelated asset (about 1.8% TE).
The first uses half of a 2% budget, the second almost all of it.

**4.** About 16% volatility, against a floor of $25\% \cdot \sqrt{0.4} \approx 15.8\%$. The
fund is as diversified as 40 stocks with that correlation can be, but it still carries
nearly the full systematic risk of Swiss equities. "Diversified" refers to stock-specific risk
only.

**5.** With unit variances, $w = (1,-1,0)$ gives $2 - 2\Sigma_{12} \ge 0$ and $w = (1,1,0)$ gives
$2 + 2\Sigma_{12} \ge 0$, so $-1 \le \Sigma_{12} \le 1$. For the "matrix" with
$\rho_{23} = -0.9$, $w = (1,-1,-1)$ gives
$3 - 2(0.9) - 2(0.9) + 2(-0.9) = -2.4 < 0$: a negative variance. Each correlation is
individually valid, but together they are impossible. Assets 2 and 3 cannot both be highly
correlated with asset 1 and strongly negatively correlated with each other.

## 9. Conclusion

- $E[R_p] = w^\top\mu$ and $\operatorname{Var}(R_p) = w^\top\Sigma w$: one dot product, one quadratic form, for any $n$.
- $\Sigma = DCD$: risk models are specified as volatilities plus correlations.
- $w^\top\Sigma w = \sum_{i,j} w_iw_j\Sigma_{ij}$; the covariance terms outnumber the variance terms $n-1$ to 1 and usually dominate.
- $\operatorname{Cov}(w^\top R, v^\top R) = w^\top\Sigma v$; tracking error is $\sqrt{(w-b)^\top\Sigma(w-b)}$.
- Equal-weight diversification removes the $\sigma^2/n$ term; the floor $\rho\sigma^2$ is systematic risk.
