# Global Minimum-Variance Portfolio with Lagrange Multipliers

The least risky fully invested portfolio of $n$ assets, in closed form. The tool is the Lagrange
multiplier: minimize a function subject to a constraint. The multiplier turns out to be the
portfolio's own variance.

## 1. Introduction

In 03/04 the minimum-variance portfolio of two assets came from a one-variable problem: write
$w_2 = 1 - w_1$, differentiate, set to zero. With $n$ assets, eliminating one weight leaves an
$(n-1)$-variable problem with messy algebra. Lagrange multipliers keep all $n$ weights and
handle the constraint $\sum_i w_i = 1$ separately.

The result is the **global minimum-variance (GMV) portfolio**:

$$
w_{\text{gmv}} = \frac{\Sigma^{-1}\mathbf{1}}{\mathbf{1}^\top\Sigma^{-1}\mathbf{1}}
$$

It is the leftmost point of the efficient frontier, and the only point that needs **no expected
returns**. After this notebook you can:

- use a Lagrange multiplier to solve an equality-constrained minimization;
- derive and compute $w_{\text{gmv}}$ and its variance;
- read the first-order condition: every asset has the same covariance with the GMV portfolio;
- explain why the GMV portfolio often shorts assets, and what a long-only constraint changes.

## 2. Theory

### Lagrange multipliers in one picture

Minimize $f(x, y)$ subject to $g(x, y) = c$. Walk along the constraint curve. At the constrained
minimum you cannot lower $f$ by moving along the curve, so the gradient of $f$ has no component
along the curve: it is perpendicular to the curve, like $\nabla g$. Hence the two gradients are
parallel:

$$
\nabla f = \lambda\,\nabla g
$$

for some number $\lambda$, the **Lagrange multiplier**. Geometrically: the level curve of $f$
through the optimum **touches** the constraint curve (see the first chart in section 6).

### The recipe

Form the Lagrangian $\mathcal{L}(x, \lambda) = f(x) - \lambda\,(g(x) - c)$ and set **all**
partial derivatives to zero:

$$
\nabla_x \mathcal{L} = \nabla f - \lambda\nabla g = 0, \qquad
\frac{\partial\mathcal{L}}{\partial\lambda} = -(g(x) - c) = 0
$$

The first is the parallel-gradient condition, the second gives back the constraint.
$n$ variables plus one multiplier, $n + 1$ equations.

### Applied to the GMV problem

$$
\min_w\ \tfrac12\,w^\top\Sigma w \quad\text{subject to}\quad \mathbf{1}^\top w = 1
$$

(The $\tfrac12$ only tidies the derivative.) With the gradient rules
$\nabla_w(\tfrac12 w^\top\Sigma w) = \Sigma w$ for symmetric $\Sigma$, and $\nabla_w(\mathbf{1}^\top w) = \mathbf{1}$:

$$
\mathcal{L}(w, \lambda) = \tfrac12\,w^\top\Sigma w - \lambda\,(\mathbf{1}^\top w - 1)
$$

$$
\nabla_w\mathcal{L} = \Sigma w - \lambda\mathbf{1} = 0
\quad\Longrightarrow\quad
w = \lambda\,\Sigma^{-1}\mathbf{1}
$$

Plug into the constraint: $\lambda\,\mathbf{1}^\top\Sigma^{-1}\mathbf{1} = 1$, so

$$
\lambda = \frac{1}{\mathbf{1}^\top\Sigma^{-1}\mathbf{1}},
\qquad
\boxed{\,w_{\text{gmv}} = \frac{\Sigma^{-1}\mathbf{1}}{\mathbf{1}^\top\Sigma^{-1}\mathbf{1}}\,}
$$

### Its variance, and what $\lambda$ means

$$
\sigma_{\text{gmv}}^2 = w_{\text{gmv}}^\top\Sigma\,w_{\text{gmv}}
= \lambda^2\,\mathbf{1}^\top\Sigma^{-1}\Sigma\,\Sigma^{-1}\mathbf{1}
= \lambda^2\,\mathbf{1}^\top\Sigma^{-1}\mathbf{1}
= \lambda
$$

The multiplier **is** the minimum variance. More generally, a multiplier is a **shadow price**:
the rate at which the optimal objective changes when the constraint is relaxed. With budget
$\mathbf{1}^\top w = c$ the same steps give $\sigma^2_{\min}(c) = c^2\lambda$, so
$\frac{d}{dc}\big[\tfrac12\sigma^2_{\min}\big] = c\lambda = \lambda$ at $c = 1$.

### Is it a minimum?

If $\Sigma$ is positive definite, $\tfrac12 w^\top\Sigma w$ is strictly convex (a bowl), and its
restriction to the flat set $\mathbf{1}^\top w = 1$ is still a bowl. The stationary point is the
unique global minimum. If $\Sigma$ is singular (04/02), $\Sigma^{-1}$ does not exist and a
riskless combination makes the problem degenerate.

### The first-order condition, in finance language

$\Sigma w_{\text{gmv}} = \lambda\mathbf{1}$ says, row by row:

$$
\operatorname{Cov}(R_i,\ R_{\text{gmv}}) = (\Sigma w_{\text{gmv}})_i = \lambda = \sigma_{\text{gmv}}^2
\quad\text{for every asset } i
$$

**Every asset has the same covariance with the GMV portfolio.** If one asset had a lower
covariance, moving a little weight into it (out of an asset with higher covariance) would
lower the portfolio variance, so the portfolio would not be the minimum.

### Special cases

- **Uncorrelated assets** ($\Sigma$ diagonal): $(\Sigma^{-1}\mathbf{1})_i = 1/\sigma_i^2$, so
  $w_i \propto 1/\sigma_i^2$. **Inverse-variance weighting.**
- **Equal volatilities and equal correlations**: by symmetry, equal weights $1/n$.
- **Two assets**: the formula reduces to 03/04's
  $w_1 = (\sigma_2^2 - \rho\sigma_1\sigma_2)/(\sigma_1^2 + \sigma_2^2 - 2\rho\sigma_1\sigma_2)$.

### Inequality constraints

"No short positions" ($w_i \ge 0$) is an **inequality** constraint. Lagrange handles equalities;
the extension to inequalities is the Karush–Kuhn–Tucker (KKT) conditions, solved numerically
by a quadratic-programming solver. There is no closed form, but the idea is the same: binding
constraints get multipliers, slack ones get zero.

## 3. Worked Examples

### Example 1: two uncorrelated assets

$\sigma_1 = 10\%$, $\sigma_2 = 20\%$, $\rho = 0$. Inverse-variance weights:

$$
\frac{1}{\sigma_1^2} = 100,\quad \frac{1}{\sigma_2^2} = 25
\quad\Longrightarrow\quad
w = \Big(\tfrac{100}{125},\ \tfrac{25}{125}\Big) = (80\%,\ 20\%)
$$

$$
\sigma_{\text{gmv}}^2 = \lambda = \frac{1}{125} = 0.008, \qquad \sigma_{\text{gmv}} = 8.94\%
$$

Lower than the 10% of the safer asset alone. Check the first-order condition:
$\operatorname{Cov}(R_1, R_{\text{gmv}}) = 0.8 \cdot 0.01 = 0.008$ and
$\operatorname{Cov}(R_2, R_{\text{gmv}}) = 0.2 \cdot 0.04 = 0.008$. Equal, and equal to $\lambda$.

### Example 2: the two-asset frontier again

Swiss equities ($\sigma = 16\%$) and Swiss bonds ($\sigma = 5\%$), $\rho = 0.1$. The Lagrange
formula gives 6.4% in equities, the same as 03/04 (checked in Python below).

### Example 3: three asset classes

The 04/01 universe: Swiss equities 16%, global equities 17%, Swiss bonds 5%; correlations
0.8 (equities), 0.1 (Swiss equities–bonds), 0.0 (global equities–bonds). The result
(computed in Python below):

| | Swiss equities | Global equities | Swiss bonds |
|---|---|---|---|
| $w_{\text{gmv}}$ | −1.9% | 9.4% | 92.5% |

$\sigma_{\text{gmv}} = 4.79\%$, below the 5% of bonds alone.

The GMV portfolio **shorts** Swiss equities. Both equity blocks are risky and highly
correlated, but global equities are uncorrelated with bonds while Swiss equities are slightly
positively correlated with them. The optimizer keeps the global block as the diversifier and
uses a small short in Swiss equities to hedge part of the global equity risk.

## 4. Finance Application

### Minimum-volatility strategies

Minimum-volatility equity indices and funds are, at their core, constrained GMV portfolios:
long-only, with caps on single stocks, sectors and countries. They exist because the GMV
portfolio needs **only** $\Sigma$, and covariances are estimated far better than expected
returns (module 06). Empirically, low-volatility portfolios have historically delivered
returns comparable to the market with less risk (the "low-volatility anomaly").

### Reading the weights

- The GMV loads on assets that are **low-volatility and weakly correlated with the rest**,
  not simply on the lowest-volatility asset.
- It shorts assets that are highly correlated with something it already holds, to hedge.
- Its weights can be extreme and unstable: $\Sigma^{-1}$ amplifies the directions with the
  smallest eigenvalues, which are the worst-estimated ones (04/02). Practitioners add
  constraints or shrink $\Sigma$ (module 08).

### Why constraints change the answer

With a long-only constraint the first-order condition holds only for assets **in** the
portfolio: they all share the same covariance with it. Excluded assets have a **higher**
covariance with the portfolio; adding them would increase variance, and shorting them is not
allowed.

### The usual misreadings

- "The minimum-variance portfolio is the lowest-volatility asset." Only if that asset is
  uncorrelated enough; in Example 3 it is not.
- "Minimum variance means safe." It means least **variance** given $\Sigma$. Estimation error,
  regime changes and fat tails are not in the formula.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp
from scipy.optimize import minimize

rng = np.random.default_rng(42)
```

### Lagrange in SymPy: two assets

Solve the full system $\nabla_w\mathcal{L} = 0$, $\partial\mathcal{L}/\partial\lambda = 0$ and compare
with the 03/04 formula.

```python
w1, w2, lam = sp.symbols("w1 w2 lambda", real=True)
s1, s2 = sp.symbols("sigma_1 sigma_2", positive=True)
rho = sp.symbols("rho", real=True)

Sigma_sym = sp.Matrix([[s1**2, rho * s1 * s2], [rho * s1 * s2, s2**2]])
w_vec = sp.Matrix([w1, w2])

L = sp.Rational(1, 2) * (w_vec.T * Sigma_sym * w_vec)[0] - lam * (w1 + w2 - 1)
solution = sp.solve([sp.diff(L, w1), sp.diff(L, w2), sp.diff(L, lam)], [w1, w2, lam], dict=True)[0]

w1_0304 = (s2**2 - rho * s1 * s2) / (s1**2 + s2**2 - 2 * rho * s1 * s2)
print("w1 (Lagrange)      =", sp.simplify(solution[w1]))
print("matches 03/04:      ", sp.simplify(solution[w1] - w1_0304) == 0)

var_min = sp.simplify((w_vec.T * Sigma_sym * w_vec)[0].subs(solution))
print("lambda equals min variance:", sp.simplify(var_min - solution[lam]) == 0)
```

### The GMV formula in NumPy

Compute $\Sigma^{-1}\mathbf{1}$ by **solving** $\Sigma x = \mathbf{1}$ rather than forming the inverse:
faster and numerically more stable.

```python
def gmv_weights(Sigma):
    ones = np.ones(Sigma.shape[0])
    x = np.linalg.solve(Sigma, ones)   # x = Sigma^{-1} 1
    return x / x.sum()


def gmv_variance(Sigma):
    ones = np.ones(Sigma.shape[0])
    return 1.0 / (ones @ np.linalg.solve(Sigma, ones))   # = lambda


assets = ["Swiss equities", "Global equities", "Swiss bonds"]
vols = np.array([0.16, 0.17, 0.05])                       # annual, decimals
corr = np.array([[1.0, 0.8, 0.1], [0.8, 1.0, 0.0], [0.1, 0.0, 1.0]])
mu = np.array([0.065, 0.07, 0.02])                        # annual expected returns
Sigma = np.diag(vols) @ corr @ np.diag(vols)

w_gmv = gmv_weights(Sigma)
lam_value = gmv_variance(Sigma)

print(pd.Series(w_gmv, index=assets).map("{:+.2%}".format).to_string())
print(f"\nsum of weights:     {w_gmv.sum():.6f}")
print(f"lambda:             {lam_value:.6f}")
print(f"w' Sigma w:         {w_gmv @ Sigma @ w_gmv:.6f}")
print(f"GMV volatility:     {np.sqrt(lam_value):.2%}")
print(f"GMV expected return {w_gmv @ mu:.2%}  (not used in the optimization)")
```

### Check the first-order condition: equal covariances with the GMV portfolio

```python
cov_with_gmv = Sigma @ w_gmv
print("Cov(R_i, R_gmv):", np.round(cov_with_gmv, 6))
print("all equal to lambda:", np.allclose(cov_with_gmv, lam_value))

# Compare: covariances with an equally weighted portfolio are not equal
w_eq = np.full(3, 1 / 3)
print("Cov(R_i, R_equal):", np.round(Sigma @ w_eq, 6))
```

### Check by brute force and with a numerical optimizer

```python
# Brute force: grid over (w1, w2), w3 = 1 - w1 - w2
grid = np.linspace(-0.3, 1.0, 1301)
W1, W2 = np.meshgrid(grid, grid)
W3 = 1 - W1 - W2
W = np.stack([W1, W2, W3], axis=-1)
var_grid = np.einsum("...i,ij,...j->...", W, Sigma, W)
idx = np.unravel_index(np.argmin(var_grid), var_grid.shape)
print("grid search:   ", np.round(W[idx], 3), f"vol {np.sqrt(var_grid[idx]):.4%}")

# Numerical optimizer with the equality constraint
budget = {"type": "eq", "fun": lambda w: w.sum() - 1}
res = minimize(lambda w: w @ Sigma @ w, x0=np.full(3, 1 / 3), constraints=[budget], method="SLSQP")
print("SLSQP:         ", np.round(res.x, 3), f"vol {np.sqrt(res.fun):.4%}")
print("closed form:   ", np.round(w_gmv, 3), f"vol {np.sqrt(lam_value):.4%}")
```

### The multiplier as a shadow price

Solve the problem with budget $c$ instead of 1 and differentiate $\tfrac12\sigma^2_{\min}(c)$
numerically at $c = 1$.

```python
def half_min_variance(c, Sigma):
    w = c * gmv_weights(Sigma)   # the optimum scales linearly with the budget
    return 0.5 * w @ Sigma @ w


h = 1e-6
shadow_price = (half_min_variance(1 + h, Sigma) - half_min_variance(1 - h, Sigma)) / (2 * h)
print(f"d/dc [1/2 sigma^2_min] at c = 1: {shadow_price:.6f}")
print(f"lambda:                          {lam_value:.6f}")
```

### Long-only GMV on a larger universe

Ten stocks from a one-factor model (market beta between 0.5 and 1.5, market volatility 16%,
idiosyncratic volatility between 8% and 30%). Compare the unconstrained closed form with a
long-only solution from SLSQP.

```python
n = 10
betas = rng.uniform(0.5, 1.5, size=n)
idio_vols = rng.uniform(0.08, 0.30, size=n)
market_vol = 0.16

Sigma_10 = market_vol**2 * np.outer(betas, betas) + np.diag(idio_vols**2)   # annual
stock_vols = np.sqrt(np.diag(Sigma_10))

w_unc = gmv_weights(Sigma_10)

bounds = [(0.0, 1.0)] * n
res_lo = minimize(lambda w: w @ Sigma_10 @ w, x0=np.full(n, 1 / n),
                  constraints=[budget], bounds=bounds, method="SLSQP",
                  options={"ftol": 1e-14, "maxiter": 500})
w_lo = np.where(res_lo.x < 1e-6, 0.0, res_lo.x)

comparison = pd.DataFrame({
    "beta": betas, "vol": stock_vols,
    "w unconstrained": w_unc, "w long-only": w_lo,
}, index=[f"stock {i + 1}" for i in range(n)])
print(comparison.round(3))
print(f"\nGMV vol unconstrained: {np.sqrt(w_unc @ Sigma_10 @ w_unc):.2%}, "
      f"gross exposure {np.abs(w_unc).sum():.2f}")
print(f"GMV vol long-only:     {np.sqrt(w_lo @ Sigma_10 @ w_lo):.2%}, "
      f"holdings {np.sum(w_lo > 0)} of {n}")
print(f"lowest single-stock vol: {stock_vols.min():.2%}")

# KKT check: held stocks share one covariance with the portfolio, excluded ones are higher
cov_lo = Sigma_10 @ w_lo
print(f"\nCov with long-only GMV, held stocks:     {np.round(cov_lo[w_lo > 0], 5)}")
print(f"Cov with long-only GMV, excluded stocks: {np.round(cov_lo[w_lo == 0], 5)}")
```

The unconstrained GMV shorts the stocks with **high beta and low idiosyncratic volatility**:
they are nearly pure market bets, so a short in them hedges the market exposure of everything
else. Long-only, those stocks are simply dropped, and the KKT pattern is visible: held stocks
share one covariance with the portfolio, excluded stocks have a higher one.

### In-sample vs true risk

Estimate $\Sigma$ from 60 monthly returns of these ten stocks, compute the GMV from the
estimate, and compare the risk the estimate **promises** with the risk the portfolio **has**
under the true $\Sigma$.

```python
periods_per_year = 12
n_trials = 1_000
promised, actual = [], []
for _ in range(n_trials):
    r = rng.multivariate_normal(np.zeros(n), Sigma_10 / periods_per_year, size=60)
    Sigma_hat = np.cov(r, rowvar=False) * periods_per_year
    w_hat = gmv_weights(Sigma_hat)
    promised.append(np.sqrt(w_hat @ Sigma_hat @ w_hat))
    actual.append(np.sqrt(w_hat @ Sigma_10 @ w_hat))

promised, actual = np.array(promised), np.array(actual)
print(f"true GMV volatility:             {np.sqrt(gmv_variance(Sigma_10)):.2%}")
print(f"average promised (in-sample):    {promised.mean():.2%}")
print(f"average actual (true Sigma):     {actual.mean():.2%}")
```

The estimated GMV portfolio looks safer than the true optimum and is riskier in reality: the
optimizer chases directions where the sample happens to show low variance. With 10 assets and
60 observations the effect is modest; module 08 shows how it grows with $n/T$.

## 6. Visualizations

```python
# Lagrange picture for two assets: level curves of variance and the budget line w1 + w2 = 1
s_a, s_b, r_ab = 0.16, 0.05, 0.1
Sigma_2 = np.array([[s_a**2, r_ab * s_a * s_b], [r_ab * s_a * s_b, s_b**2]])
w_star = gmv_weights(Sigma_2)
var_star = w_star @ Sigma_2 @ w_star

g1, g2 = np.meshgrid(np.linspace(-0.4, 1.2, 400), np.linspace(-0.2, 1.6, 400))
V = Sigma_2[0, 0] * g1**2 + 2 * Sigma_2[0, 1] * g1 * g2 + Sigma_2[1, 1] * g2**2

fig, ax = plt.subplots(figsize=(7, 7))
cs = ax.contour(g1, g2, V, levels=var_star * np.array([2.0, 4.0, 8.0]), colors="grey", linewidths=1)
ax.clabel(cs, fmt=lambda v: f"σ={np.sqrt(v):.1%}", fontsize=8)
cs_star = ax.contour(g1, g2, V, levels=[var_star], colors="tab:red", linewidths=2)
ax.clabel(cs_star, fmt=lambda v: f"σ={np.sqrt(v):.1%}", fontsize=8)
line_w1 = np.linspace(-0.4, 1.2, 2)
ax.plot(line_w1, 1 - line_w1, color="black", label="budget w1 + w2 = 1")
ax.scatter(*w_star, color="tab:red", zorder=3, label=f"GMV ({w_star[0]:.1%}, {w_star[1]:.1%})")

# Gradients at the optimum: grad of 1/2 w'Σw is Σw; grad of the constraint is (1, 1)
grad_f = Sigma_2 @ w_star
scale = 0.25 / np.linalg.norm(grad_f)
ax.annotate("", xy=w_star + scale * grad_f, xytext=w_star,
            arrowprops=dict(arrowstyle="->", color="tab:red", lw=2))
ax.annotate("", xy=w_star + 0.25 * np.array([1, 1]) / np.sqrt(2), xytext=w_star,
            arrowprops=dict(arrowstyle="->", color="black", lw=1, linestyle="--"))
ax.text(*(w_star + np.array([0.06, 0.2])), "∇f = Σw  ∥  ∇g = (1, 1)", color="tab:red")

ax.set_title("Tangency: the lowest variance ellipse touching the budget line")
ax.set_xlabel("w1 (Swiss equities)")
ax.set_ylabel("w2 (Swiss bonds)")
ax.set_aspect("equal")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper right")
plt.show()
```

## 8. Solutions / Checks

```python
# 1.
Sigma_u = np.diag([0.10, 0.15, 0.20]) ** 2
inv_var = 1 / np.diag(Sigma_u)
print("1. by hand:  ", np.round(inv_var / inv_var.sum(), 4), f"vol {np.sqrt(1 / inv_var.sum()):.2%}")
print("   formula:  ", np.round(gmv_weights(Sigma_u), 4), f"vol {np.sqrt(gmv_variance(Sigma_u)):.2%}")

# 2. Lagrange by SymPy
x, y, l, c = sp.symbols("x y lambda c", real=True)
L2 = x**2 + 2 * y**2 - l * (x + y - c)
sol2 = sp.solve([sp.diff(L2, x), sp.diff(L2, y), sp.diff(L2, l)], [x, y, l], dict=True)[0]
f_min = sp.simplify((x**2 + 2 * y**2).subs(sol2))
print(f"2. x = {sol2[x]}, y = {sol2[y]}, lambda = {sol2[l]}, f_min(c) = {f_min}")
print(f"   at c = 3: lambda = {sol2[l].subs(c, 3)}, f_min = {f_min.subs(c, 3)}; "
      f"f_min(3.1) - f_min(3) = {float(f_min.subs(c, 3.1) - f_min.subs(c, 3)):.3f}")

# 3.
corr_3 = corr.copy()
corr_3[0, 2] = corr_3[2, 0] = 0.3
Sigma_3 = np.diag(vols) @ corr_3 @ np.diag(vols)
print("3. rho(CH eq, bonds) = 0.1:", np.round(w_gmv, 4))
print("   rho(CH eq, bonds) = 0.3:", np.round(gmv_weights(Sigma_3), 4))

# 5. numerical check with a random fully invested portfolio
w_any = rng.normal(size=3)
w_any /= w_any.sum()
lhs = w_any @ Sigma @ w_any
rhs = lam_value + (w_any - w_gmv) @ Sigma @ (w_any - w_gmv)
print(f"5. Cov(w, gmv) = {w_any @ Sigma @ w_gmv:.6f} vs lambda {lam_value:.6f};  "
      f"sigma_w^2 = {lhs:.6f} vs lambda + Var(diff) = {rhs:.6f}")
```

**1.** $1/\sigma^2 = 100,\ 44.4,\ 25$, sum $169.4$: weights $59.0\%,\ 26.2\%,\ 14.8\%$,
volatility $\sqrt{1/169.4} = 7.68\%$.

**2.** $2x = \lambda$, $4y = \lambda$, $x + y = c$ give $x = 2c/3$, $y = c/3$, $\lambda = 4c/3$ and
$f_{\min}(c) = 2c^2/3$. At $c = 3$: $\lambda = 4$, $f_{\min} = 6$. Relaxing to 3.1 raises the
minimum by about $\lambda \cdot 0.1 = 0.4$ (exactly 0.407): the multiplier is the shadow price.

**3.** Swiss equities now co-move more with bonds, so they are a worse diversifier for the
dominant bond position. The optimizer shorts much more of them (about −20%) and offsets the
equity short with more global equities (about +22%). The Swiss equity short now hedges part of
the bond risk as well as part of the global equity risk.

**4.** The first-order condition says each holding must have the same covariance with the
portfolio. A low-volatility utility with low correlation to the rest has a low covariance with
the portfolio at small weight, so the optimizer adds it until its covariance catches up. The
risk $\Sigma$ does not see: concentration in one name and one sector (regulation, interest-rate
sensitivity, a single bad event), and the possibility that the estimated low correlation is
noise. This is why minimum-volatility products cap single names and sectors.

**5.** $\operatorname{Cov}(R_w, R_{\text{gmv}}) = w^\top\Sigma w_{\text{gmv}} = w^\top(\lambda\mathbf{1}) = \lambda$,
using $\Sigma w_{\text{gmv}} = \lambda\mathbf{1}$ and $\mathbf{1}^\top w = 1$. Then

$$
\operatorname{Var}(R_w - R_{\text{gmv}}) = \sigma_w^2 + \lambda - 2\lambda = \sigma_w^2 - \sigma_{\text{gmv}}^2
$$

Every fully invested portfolio is the GMV portfolio plus a zero-cost long-short portfolio that
is **uncorrelated** with it, and the variances add. Any deviation from the GMV adds risk; none
can remove it.

## 9. Conclusion

- Lagrange multipliers: at a constrained optimum $\nabla f = \lambda\nabla g$; solve together with the constraint.
- GMV: $w_{\text{gmv}} = \Sigma^{-1}\mathbf{1}/(\mathbf{1}^\top\Sigma^{-1}\mathbf{1})$, variance $\sigma^2_{\text{gmv}} = 1/(\mathbf{1}^\top\Sigma^{-1}\mathbf{1}) = \lambda$.
- The multiplier is a shadow price: how fast the optimal objective moves when the constraint is relaxed.
- First-order condition: every asset has covariance $\sigma^2_{\text{gmv}}$ with the GMV portfolio.
- Uncorrelated assets give inverse-variance weights; correlated ones can produce shorts.
- No expected returns needed, but $\Sigma^{-1}$ amplifies estimation error; long-only and caps are standard in practice.
