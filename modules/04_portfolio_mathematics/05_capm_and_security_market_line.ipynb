# CAPM and the Security Market Line

04/04 ended with an algebraic fact about the tangency portfolio: every asset's excess return is
its beta times the tangency portfolio's excess return. The Capital Asset Pricing Model adds one
economic argument, that in equilibrium the tangency portfolio **is** the market, and turns the
fact into a theory of expected returns.

## 1. Introduction

Modules 03 and 04 took expected returns $\mu$ as given and asked how to build portfolios.
CAPM reverses the question: if everyone builds portfolios the 04/04 way, what must expected
returns be?

The answer is the **security market line (SML)**:

$$
E[R_i] - r_f = \beta_i\,\big(E[R_M] - r_f\big)
$$

Expected excess return depends on **beta** alone, not on volatility. After this notebook you can:

- derive the SML from 04/04 plus the equilibrium argument;
- split an asset's variance into a priced (systematic) and an unpriced (idiosyncratic) part;
- tell the CML from the SML, and compute alpha;
- back out the expected returns that a set of market weights implies (reverse optimization);
- see why testing CAPM with real data is hard.

## 2. Theory

### From tangency portfolio to market portfolio

Assumptions: all investors are mean-variance optimizers, agree on $\mu$ and $\Sigma$, can lend and
borrow at $r_f$, and face no taxes or transaction costs.

- By 04/04, every investor holds the **same** risky portfolio $w_{\text{tan}}$, plus cash.
- Adding up all investors, the aggregate risky holdings are proportional to $w_{\text{tan}}$.
- Aggregate holdings must equal what exists: the **market portfolio** $w_M$, each asset at its
  market-capitalization weight.

Hence $w_{\text{tan}} = w_M$: in equilibrium the tangency portfolio is the market.

### The security market line

The first-order condition from 04/04, $\Sigma w_{\text{tan}} \propto \mu - r_f\mathbf{1}$, with
$w_{\text{tan}} = w_M$:

$$
\mu_i - r_f = \beta_i\,(\mu_M - r_f),
\qquad
\beta_i = \frac{\operatorname{Cov}(R_i, R_M)}{\sigma_M^2} = \frac{(\Sigma w_M)_i}{w_M^\top\Sigma w_M}
$$

- $\beta_M = 1$ and $\beta_{r_f} = 0$: the line passes through $(0, r_f)$ and $(1, \mu_M)$.
- Beta is linear in weights: a portfolio's beta is $\beta_p = w^\top\beta$, so every portfolio
  lies on the SML too.

### Systematic and idiosyncratic risk

Regress an asset's excess return on the market's:

$$
R_i - r_f = \alpha_i + \beta_i\,(R_M - r_f) + \varepsilon_i,
\qquad \operatorname{Cov}(\varepsilon_i, R_M) = 0
$$

Taking variances (the cross term vanishes):

$$
\sigma_i^2 = \underbrace{\beta_i^2\,\sigma_M^2}_{\text{systematic}} + \underbrace{\sigma_{\varepsilon,i}^2}_{\text{idiosyncratic}}
$$

The idiosyncratic part diversifies away (04/01: the $\sigma^2/n$ term), so no one needs to be
paid for bearing it. Only the systematic part, measured by $\beta$, earns a premium. CAPM says
$\alpha_i = 0$ for every asset.

### CML vs SML

| | Capital market line (04/04) | Security market line |
|---|---|---|
| Axes | volatility $\sigma$, expected return | beta $\beta$, expected return |
| Who is on it | efficient portfolios only (cash + market) | **every** asset and portfolio |
| Slope | market Sharpe ratio $(\mu_M - r_f)/\sigma_M$ | market risk premium $\mu_M - r_f$ |

A single stock lies **below** the CML (it carries idiosyncratic risk that earns nothing) but
**on** the SML. With $\rho_{iM}$ the correlation with the market,
$\beta_i = \rho_{iM}\sigma_i/\sigma_M$, so its Sharpe ratio is $\rho_{iM}$ times the market's.

### Alpha

Given an estimate of an asset's or fund's expected return, **Jensen's alpha** is its distance
above the SML:

$$
\alpha_i = (\mu_i - r_f) - \beta_i\,(\mu_M - r_f)
$$

Positive alpha means more expected return than its systematic risk justifies. Under CAPM it is
zero in equilibrium; in practice it is what active managers claim and what is hard to prove
(module 06/03).

### Reverse optimization: implied expected returns

Run 04/04 backwards. If the market portfolio is the tangency portfolio, then

$$
\Pi = \mu - r_f\mathbf{1} = \delta\,\Sigma\,w_M,
\qquad
\delta = \frac{\mu_M - r_f}{\sigma_M^2}
$$

$\delta$ is the market's price of variance (risk aversion). Given $\Sigma$, market weights and a
single number for the market risk premium, this gives a full vector of **equilibrium expected
returns**, and the tangency portfolio built from them is exactly $w_M$. Black–Litterman
(08/04) starts from this $\Pi$.

## 3. Worked Examples

Throughout: $r_f = 0.5\%$, market risk premium $\mu_M - r_f = 5\%$, $\sigma_M = 16\%$ (annual).

### Example 1: expected returns from beta

| Stock | $\beta$ | CAPM expected return |
|---|---|---|
| Utility | 0.6 | $0.5\% + 0.6 \cdot 5\% = 3.5\%$ |
| Tech | 1.4 | $0.5\% + 1.4 \cdot 5\% = 7.5\%$ |
| Portfolio 50/50 | $0.5 \cdot 0.6 + 0.5 \cdot 1.4 = 1.0$ | $5.5\%$ |

### Example 2: which risk is paid?

A stock with $\sigma = 30\%$ and $\beta = 1.2$:

$$
\text{systematic: } 1.2^2 \cdot 0.16^2 = 0.0369 \;(\sigma = 19.2\%),
\qquad
\text{idiosyncratic: } 0.09 - 0.0369 = 0.0531 \;(\sigma = 23.1\%)
$$

59% of its variance is idiosyncratic and earns nothing. Its CAPM expected return is
$0.5\% + 1.2 \cdot 5\% = 6.5\%$, the same as any other asset with $\beta = 1.2$, whatever its volatility.

### Example 3: alpha of a fund

A fund with $\beta = 1.1$ is expected to return 8%. CAPM requires
$0.5\% + 1.1 \cdot 5\% = 6.0\%$, so $\alpha = 2.0\%$.

### Example 4: implied returns for the three asset classes

The 04/01 universe with market weights 25% Swiss equities, 45% global equities, 30% Swiss
bonds, and a market premium of 4%. The market portfolio has $\sigma_M = 11.3\%$, so
$\delta = 0.04/0.113^2 = 3.15$ and (computed in Python below):

| | Swiss equities | Global equities | Swiss bonds |
|---|---|---|---|
| $\beta$ vs this market | 1.29 | 1.45 | 0.07 |
| implied excess return $\Pi$ | 5.18% | 5.81% | 0.30% |
| 04/04 assumption | 6.0% | 6.5% | 1.5% |

Bonds barely co-move with this equity-heavy market, so equilibrium assigns them almost no
premium. The 04/04 assumptions are not consistent with these market weights, which is why the
04/04 tangency portfolio held 70% bonds.

## 4. Finance Application

### Cost of equity

Corporate finance uses CAPM to set the discount rate for equity cash flows:
$r_E = r_f + \beta\,\text{(market risk premium)}$. Valuations, hurdle rates and regulated
utility tariffs depend on it.

### Performance evaluation

Fund returns are judged against the SML, not against raw returns. A fund that beat the market
by taking more beta did not add value; alpha measures the part beta does not explain. Related
measures: the Treynor ratio $(\mu_p - r_f)/\beta_p$ (return per unit of systematic risk) and the
information ratio $\alpha/\sigma_\varepsilon$ (alpha per unit of idiosyncratic risk).

### Equilibrium returns as optimizer input

04/04 showed tangency weights swing wildly with $\mu$. Reverse optimization gives a $\mu$ that
is consistent with $\Sigma$ and with observed market weights, so the optimizer reproduces the
market instead of producing extreme positions. Views are then added on top (Black–Litterman,
08/04).

### What the data say

Empirically the SML is **flatter** than CAPM predicts: low-beta stocks earned more and
high-beta stocks less than the model implies (the low-beta anomaly, related to the
low-volatility anomaly in 04/03). Other characteristics (size, value, momentum) explain average
returns that beta does not, which led to multi-factor models (module 07/03). CAPM remains the
reference point: the model every other model is compared with.

### Why testing CAPM is hard

- Expected returns are not observed; average returns are a very noisy proxy (module 06/01).
- The "market portfolio" includes all risky assets (bonds, real estate, private firms, human
  capital). Any index is only a proxy, so a rejection may reject the proxy, not the model
  (Roll's critique).

### The usual misreading

"Riskier stocks have higher expected returns." Only systematic risk is rewarded. A
high-volatility stock with a low beta should, under CAPM, have a **low** expected return.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
rf = 0.005   # CHF risk-free rate, annual
```

### A CAPM world by construction

Thirty stocks in a one-factor model. Choose market-cap weights, build $\Sigma$, and set
expected returns by reverse optimization, so that CAPM holds exactly. Then check the
properties derived above.

```python
n = 30
true_betas_factor = rng.uniform(0.4, 1.6, size=n)     # exposure to a common factor
idio_vols = rng.uniform(0.10, 0.35, size=n)
factor_vol = 0.16

Sigma = factor_vol**2 * np.outer(true_betas_factor, true_betas_factor) + np.diag(idio_vols**2)
caps = rng.lognormal(mean=0.0, sigma=1.0, size=n)     # market capitalizations
w_m = caps / caps.sum()

var_m = w_m @ Sigma @ w_m
market_premium = 0.05
delta = market_premium / var_m

Pi = delta * Sigma @ w_m                              # implied excess returns
mu = rf + Pi
betas = Sigma @ w_m / var_m                           # betas vs the market portfolio

print(f"market volatility:    {np.sqrt(var_m):.2%}")
print(f"risk aversion delta:  {delta:.2f}")
print(f"market beta w_m'β:    {w_m @ betas:.4f}")
print(f"market excess return: {w_m @ Pi:.2%}")

# Tangency portfolio of (mu, Sigma) is the market portfolio
x = np.linalg.solve(Sigma, mu - rf)
w_tan = x / x.sum()
print(f"max |w_tan - w_m|:    {np.abs(w_tan - w_m).max():.2e}")

# SML holds exactly
print(f"max |Π - β·premium|:  {np.abs(Pi - betas * market_premium).max():.2e}")
```

### Systematic vs idiosyncratic variance

```python
total_var = np.diag(Sigma)
systematic_var = betas**2 * var_m
idio_var = total_var - systematic_var

decomp = pd.DataFrame({
    "vol": np.sqrt(total_var),
    "beta": betas,
    "systematic share": systematic_var / total_var,
    "expected return": mu,
    "Sharpe": (mu - rf) / np.sqrt(total_var),
}, index=[f"stock {i + 1}" for i in range(n)])

print(decomp.sort_values("vol").round(3).head(8))
print(f"\nmarket Sharpe ratio: {market_premium / np.sqrt(var_m):.3f}")
print(f"highest stock Sharpe: {decomp['Sharpe'].max():.3f}")

# Sharpe of each stock = correlation with market x market Sharpe
corr_with_m = (Sigma @ w_m) / (np.sqrt(total_var) * np.sqrt(var_m))
print(f"Sharpe_i = rho_iM * Sharpe_M holds: {np.allclose(decomp['Sharpe'], corr_with_m * market_premium / np.sqrt(var_m))}")
```

### Example 2 and 3 by formula

```python
premium, sigma_m = 0.05, 0.16

def capm_expected_return(beta, rf=rf, premium=premium):
    return rf + beta * premium


stock_vol, stock_beta = 0.30, 1.2
sys_var = stock_beta**2 * sigma_m**2
print(f"Example 2: systematic vol {np.sqrt(sys_var):.1%}, idiosyncratic vol {np.sqrt(stock_vol**2 - sys_var):.1%}, "
      f"idiosyncratic share {1 - sys_var / stock_vol**2:.0%}, CAPM return {capm_expected_return(stock_beta):.1%}")

fund_mu, fund_beta = 0.08, 1.1
print(f"Example 3: CAPM return {capm_expected_return(fund_beta):.1%}, alpha {fund_mu - capm_expected_return(fund_beta):.1%}")
```

### Implied returns for the three asset classes (Example 4)

```python
assets = ["Swiss equities", "Global equities", "Swiss bonds"]
vols_3 = np.array([0.16, 0.17, 0.05])
corr_3 = np.array([[1.0, 0.8, 0.1], [0.8, 1.0, 0.0], [0.1, 0.0, 1.0]])
Sigma_3 = np.diag(vols_3) @ corr_3 @ np.diag(vols_3)
w_m3 = np.array([0.25, 0.45, 0.30])
mu_0404 = np.array([0.065, 0.07, 0.02])

var_m3 = w_m3 @ Sigma_3 @ w_m3
delta_3 = 0.04 / var_m3
Pi_3 = delta_3 * Sigma_3 @ w_m3


def tangency(mu, Sigma, rf):
    x = np.linalg.solve(Sigma, mu - rf)
    return x / x.sum()


table = pd.DataFrame({
    "market weight": w_m3,
    "beta": Sigma_3 @ w_m3 / var_m3,
    "implied excess Π": Pi_3,
    "04/04 excess": mu_0404 - rf,
    "tangency from Π": tangency(rf + Pi_3, Sigma_3, rf),
    "tangency from 04/04": tangency(mu_0404, Sigma_3, rf),
}, index=assets)
print(f"sigma_M = {np.sqrt(var_m3):.2%}, delta = {delta_3:.2f}")
table.round(4)
```

### Testing CAPM on simulated data

Use the 30-stock CAPM world (where CAPM is **true**), simulate 10 years of monthly returns,
estimate each stock's beta by regression on the market, and regress average excess returns on
estimated betas. A perfect test would recover intercept 0 and slope 5%.

```python
periods_per_year = 12
n_months = 120

r = rng.multivariate_normal(mu / periods_per_year, Sigma / periods_per_year, size=n_months)
r_m = r @ w_m
excess = r - rf / periods_per_year
excess_m = r_m - rf / periods_per_year

# Time-series regressions: beta_i = Cov(r_i, r_m) / Var(r_m)
beta_hat = np.array([np.cov(excess[:, i], excess_m)[0, 1] for i in range(n)]) / np.var(excess_m, ddof=1)
avg_excess = excess.mean(axis=0) * periods_per_year          # annualized

# Cross-sectional regression: average excess return on estimated beta
slope, intercept = np.polyfit(beta_hat, avg_excess, deg=1)
r2 = np.corrcoef(beta_hat, avg_excess)[0, 1] ** 2
print(f"estimated SML: intercept {intercept:+.2%}, slope {slope:.2%}  (truth: 0%, 5%)")
print(f"R^2 of the cross-section: {r2:.2f}")
print(f"realized market excess return over the sample: {excess_m.mean() * periods_per_year:.2%}")
```

Even in a world where CAPM holds exactly, ten years of data produce a noisy SML. The
realized market return over the sample differs from its expectation, and each stock's average
return carries idiosyncratic noise of the order $\sigma_\varepsilon/\sqrt{10}$. Repeat with
different seeds and the slope can even come out negative.

```python
n_repeats = 500
slopes = np.empty(n_repeats)
for k in range(n_repeats):
    r_k = rng.multivariate_normal(mu / periods_per_year, Sigma / periods_per_year, size=n_months)
    ex_k = r_k - rf / periods_per_year
    ex_m_k = r_k @ w_m - rf / periods_per_year
    b_k = np.array([np.cov(ex_k[:, i], ex_m_k)[0, 1] for i in range(n)]) / np.var(ex_m_k, ddof=1)
    slopes[k] = np.polyfit(b_k, ex_k.mean(axis=0) * periods_per_year, deg=1)[0]

print(f"SML slope over {n_repeats} simulated decades: mean {slopes.mean():.2%}, std {slopes.std():.2%}")
print(f"share of decades with a negative slope: {np.mean(slopes < 0):.0%}")
```

The slopes scatter widely around the truth, and their average sits **below** 5%. Estimated betas
contain noise, and regressing on a noisy regressor biases the slope toward zero
(errors-in-variables). Empirical tests group stocks into portfolios to reduce this noise.

### How long to prove an alpha?

A fund with true alpha 2% per year and idiosyncratic volatility 6%. The t-statistic of the
estimated alpha after $T$ years is about $\alpha\sqrt{T}/\sigma_\varepsilon$.

```python
alpha_true, idio_fund = 0.02, 0.06
for years in [3, 5, 10, 20, 36]:
    print(f"{years:2d} years: expected t-stat {alpha_true * np.sqrt(years) / idio_fund:.2f}")
print(f"years needed for t = 2: {(2 * idio_fund / alpha_true)**2:.0f}")
```

## 8. Solutions / Checks

```python
# 1.
rf_1, prem_1 = 0.01, 0.045
beta_port = 0.6 * 0.8 + 0.4 * 1.5
print(f"1. stock: {rf_1 + 0.8 * prem_1:.2%}, portfolio beta {beta_port:.2f}: {rf_1 + beta_port * prem_1:.2%}")

# 2.
s_vol, rho_m, m_vol = 0.25, 0.5, 0.15
b = rho_m * s_vol / m_vol
sys_vol = b * m_vol
print(f"2. beta {b:.3f}, systematic vol {sys_vol:.1%}, idiosyncratic vol {np.sqrt(s_vol**2 - sys_vol**2):.1%}, "
      f"Sharpe = {rho_m} x market Sharpe")

# 3.
w_alt = np.array([0.40, 0.40, 0.20])
var_alt = w_alt @ Sigma_3 @ w_alt
Pi_alt = 0.04 / var_alt * Sigma_3 @ w_alt
print("3. implied excess returns (40/40/20):", np.round(Pi_alt, 4), " vs (25/45/30):", np.round(Pi_3, 4))
print(f"   bond beta: {(Sigma_3 @ w_alt / var_alt)[2]:.3f} vs {(Sigma_3 @ w_m3 / var_m3)[2]:.3f}")

# 5. numerical check with arbitrary delta
for d in [0.5, 3.0, 10.0]:
    print(f"5. delta = {d:4.1f}: tangency from Π equals w_M: {np.allclose(tangency(rf + d * Sigma_3 @ w_m3, Sigma_3, rf), w_m3)}")
```

**2.** $\beta = 0.5 \cdot 25/15 = 0.833$; systematic volatility $\beta\sigma_M = 12.5\%$, idiosyncratic
$\sqrt{0.0625 - 0.015625} = 21.7\%$. Its Sharpe ratio is half the market's: three quarters of its
variance is unrewarded.

**3.** With less bonds and more equity, the market's variance rises faster than the bond's
covariance with it, so the bond's beta falls from 0.075 to 0.052 and its implied premium from
0.30% to 0.21%. The equity premia fall too (to 4.80% and 5.09%): equities now make up most of
the market, so their betas move toward 1. The market premium stays 4% by construction.

**4.** With $\beta \approx 0$, CAPM predicts an expected return close to $r_f$; with negative beta,
**below** $r_f$. Investors accept a low expected return for an asset that tends to do well when
the market does badly: it pays off in the states where wealth is scarce, like insurance.

**5.** $\Sigma^{-1}\Pi = \delta\,\Sigma^{-1}\Sigma w_M = \delta w_M$, and normalizing to sum 1 gives
$w_M$ (since $\mathbf{1}^\top w_M = 1$ and $\delta > 0$). The implied $\Pi$ already lies in the
direction $\Sigma w_M$, so $\Sigma^{-1}$ maps it back to the market weights instead of amplifying
arbitrary differences in expected returns. In 04/04 the assumed $\mu$ had components outside
that direction, and $\Sigma^{-1}$ turned them into large tilts.

## 9. Conclusion

- Equilibrium argument: everyone holds the tangency portfolio, so it must be the market portfolio.
- SML: $\mu_i - r_f = \beta_i(\mu_M - r_f)$, $\beta_i = \operatorname{Cov}(R_i, R_M)/\sigma_M^2$; it holds for every asset and portfolio.
- $\sigma_i^2 = \beta_i^2\sigma_M^2 + \sigma_\varepsilon^2$: only the systematic part is priced.
- CML (efficient portfolios, slope = market Sharpe) vs SML (all assets, slope = market premium).
- Alpha is the distance above the SML; proving it takes decades of data.
- Reverse optimization $\Pi = \delta\Sigma w_M$ gives equilibrium expected returns whose tangency portfolio is the market.
- Even when CAPM is true, a decade of data gives a noisy SML; real data show a flatter line than the model predicts.
