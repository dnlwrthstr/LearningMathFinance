# Is This Fund's Alpha Real? Testing Outperformance

A fund beat its benchmark by 2% a year for five years. Skill or luck? Hypothesis testing turns the
question into a number, the p-value, and its companion, power, says how likely the test is to detect skill
that is really there. Applied to one fund, the answers are sobering: most track records are too short to
prove anything. Applied to a thousand funds, a new problem appears: some of them will look skilled by chance.

## 1. Introduction

Here **alpha** is the mean **active return** $a_t = r_{\text{fund},t} - r_{\text{bench},t}$ against a benchmark with the same
market exposure (beta 1). Its volatility is the **tracking error** TE, and the ratio of annual alpha to annual TE is the
**information ratio** IR. Funds whose beta differs from 1 need a regression to separate alpha from market
exposure; that is module 07/01. The testing logic is the same.

After this notebook you can:

- set up the null hypothesis "no alpha" and compute the t-statistic and p-value of a track record;
- read a p-value correctly, and say what it does not mean;
- compute the power of the test and the years needed to detect a given information ratio;
- explain why, among many funds, a significant t-statistic is weak evidence, and correct for it;
- recognize regression to the mean and survivorship bias in fund rankings.

## 2. Theory

### The test

With $n$ monthly active returns, sample mean $\bar a$ and sample volatility $\hat s$:

$$
H_0: \alpha = 0 \quad\text{vs}\quad H_1: \alpha > 0, \qquad
t = \frac{\bar a}{\hat s/\sqrt n}
$$

Under $H_0$ and for normal, independent returns, $t$ has a $t_{n-1}$ distribution. The one-sided **p-value** is

$$
p = P(t_{n-1} \ge t_{\text{obs}})
$$

the probability of a track record at least this good **if the manager had no skill**. Reject $H_0$ at level $\alpha_{\text{test}}$ (usually
5%) if $p < \alpha_{\text{test}}$, equivalently $t > 1.645$ for large $n$. (The symbol $\alpha$ does double duty in finance and
statistics; here $\alpha_{\text{test}}$ is the significance level.)

Annualized: $\bar a \approx \alpha$ per year, $\hat s\sqrt{12} \approx \mathrm{TE}$, $n = 12T$, so

$$
t \approx \frac{\alpha}{\mathrm{TE}/\sqrt T} = \mathrm{IR}\cdot\sqrt T
$$

### What a p-value is not

$p$ is $P(\text{data this extreme} \mid \text{no skill})$. It is **not** $P(\text{no skill} \mid \text{data})$. Turning one into the other needs
Bayes' theorem and a prior: how common skill is among the funds being tested.

### Power

If the true information ratio is $\mathrm{IR} > 0$, then $t \approx N(\mathrm{IR}\sqrt T,\ 1)$, and the probability of rejecting $H_0$ is

$$
\text{power} = P(t > z_{1-\alpha_{\text{test}}}) = 1 - \Phi\big(z_{1-\alpha_{\text{test}}} - \mathrm{IR}\sqrt T\big)
$$

Setting power to $1-\beta$ and solving for $T$:

$$
T = \left(\frac{z_{1-\alpha_{\text{test}}} + z_{1-\beta}}{\mathrm{IR}}\right)^2
$$

For a 5% one-sided test and 80% power, $z$-values $1.645 + 0.842 = 2.486$: an IR of 0.5 needs 25 years.

### Many funds: multiple testing

Test $N$ funds with no skill at level 5%: on average $0.05N$ are significant by luck, and the probability that at least one is
significant is $1 - 0.95^N$ (99.4% for $N = 100$). Two ways to respond:

- **Control the chance of any false discovery** (family-wise error): **Bonferroni** tests each fund at $\alpha_{\text{test}}/N$.
- **Raise the bar**: Harvey, Liu and Zhu (2016) argue for $t > 3$ in factor and strategy research.

With a share $\pi$ of skilled funds, Bayes' theorem gives the share of significant funds that are really skilled:

$$
P(\text{skilled} \mid \text{significant}) = \frac{\pi\cdot\text{power}}{\pi\cdot\text{power} + (1-\pi)\,\alpha_{\text{test}}}
$$

### Two biases in fund data

- **Regression to the mean**: funds ranked top on past returns were partly lucky; their next-period returns fall back
  toward their true alpha.
- **Survivorship bias**: funds with poor records close and drop out of databases. The average alpha of the survivors
  overstates the average alpha of all funds that were launched.

## 3. Worked Examples

Annual alpha and tracking error; monthly data.

### Example 1: five years

Alpha 2% per year, TE 4%: $\mathrm{IR} = 0.5$, $t = 0.5\sqrt5 = 1.118$. With $n = 60$, $p = P(t_{59} \ge 1.118) \approx 0.13$. Not significant: a
manager without skill produces a record this good about one time in eight.

### Example 2: twenty years

Same IR, $T = 20$: $t = 0.5\sqrt{20} = 2.236$, $p \approx 0.013$. Significant at 5%, but few managers have 20 years with the same process.

### Example 3: power

IR 0.5 over 5 years: power $= 1 - \Phi(1.645 - 1.118) = 1 - \Phi(0.527) = 0.30$. A genuinely skilled manager fails the test 70% of the time.
For 80% power: $T = (2.486/0.5)^2 = 24.7$ years.

### Example 4: a universe of 1,000 funds

10% of the funds are skilled with IR 0.5, the rest have no skill; 10 years of data each. Power per skilled fund
$= 1 - \Phi(1.645 - 0.5\sqrt{10}) = 1 - \Phi(0.064) = 0.474$.

- Skilled and significant: $100 \cdot 0.474 = 47$.
- Unskilled and significant: $900 \cdot 0.05 = 45$.
- $P(\text{skilled} \mid \text{significant}) = 47.4/(47.4 + 45) = 51\%$.

A significant fund is about a coin flip.

### Example 5: Bonferroni

For 1,000 funds at a family-wise 5% one-sided level, each fund needs $p < 0.00005$, i.e. $t > 3.89$. A skilled fund with IR 0.5 reaches
that with 50% probability only after $T = (3.89/0.5)^2 = 61$ years.

## 4. Finance Application

### Where it matters

- **Manager selection.** Consultants screen thousands of funds. A screen on "significant alpha at 5%" passes many
  lucky funds; the share depends on how rare skill is (Example 4).
- **Strategy research.** Each backtest is a test. Hundreds of tried variants make $t = 2$ meaningless; Harvey,
  Liu and Zhu's $t > 3$ is a rough correction.
- **Performance fees and reviews.** A three-year review with IR 0.5 has power about 0.22 (Python section): a
  "fail to reject" result says little about the manager.
- **Studies of fund performance.** Academic studies find that the average active fund has a negative alpha after fees,
  and that past winners only weakly persist; regression to the mean and survivorship explain part of what looks
  like persistence in raw data.

### The usual misreadings

- "p = 0.03 means a 97% chance the manager is skilled." The p-value conditions on no skill; the probability of skill
  needs a prior (Example 4 and exercise 7).
- "Not significant, so the manager has no skill." With low power, absence of evidence is not evidence of absence.
- "The top-decile funds of the last five years will stay on top." Mostly they regress toward the average.
- "The average fund in the database beat the benchmark." The database lists the survivors.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
months_per_year = 12
z_95 = stats.norm.ppf(0.95)      # one-sided 5% critical value
```

### One fund

Simulated monthly active returns (fund minus benchmark, decimal, CHF) with true alpha 2% per year and TE 4% per year.

```python
def active_returns(alpha_ann, te_ann, years, rng, n_funds=None):
    # monthly active returns; shape (years*12,) or (n_funds, years*12)
    shape = (years * months_per_year,) if n_funds is None else (n_funds, years * months_per_year)
    return alpha_ann / months_per_year + te_ann / np.sqrt(months_per_year) * rng.standard_normal(shape)


def alpha_t_test(a):
    # t-statistic and one-sided p-value of H0: mean active return = 0, along the last axis
    n = a.shape[-1]
    t = a.mean(axis=-1) / (a.std(axis=-1, ddof=1) / np.sqrt(n))
    return t, stats.t.sf(t, df=n - 1)


a = active_returns(0.02, 0.04, 5, rng)
t_obs, p_obs = alpha_t_test(a)
check = stats.ttest_1samp(a, 0.0, alternative="greater")

print(f"annualized alpha estimate: {a.mean() * 12:.2%},  TE estimate: {a.std(ddof=1) * np.sqrt(12):.2%}")
print(f"t = {t_obs:.3f}, p = {p_obs:.3f}   (scipy: t = {check.statistic:.3f}, p = {check.pvalue:.3f})")
print(f"Example 1, true values: t = {0.5 * np.sqrt(5):.3f}, p = {stats.t.sf(0.5 * np.sqrt(5), 59):.3f}")
```

This particular simulated record came out significant ($t = 2.07$, $p = 0.022$) because its realized alpha (2.9%) is above the
true 2% and its realized TE (3.1%) below the true 4%. The expected $t$ for this fund is only 1.12 (Example 1). The
same manager, in another five years, would usually not pass the test.

### p-values under no skill and under skill

10,000 funds without skill and 10,000 with IR 0.5, five years each.

```python
n_sim = 10_000
t0, p0 = alpha_t_test(active_returns(0.0, 0.04, 5, rng, n_sim))
t1, p1 = alpha_t_test(active_returns(0.02, 0.04, 5, rng, n_sim))

print(f"no skill : share with p < 0.05 = {np.mean(p0 < 0.05):.1%}   (5% in expectation)")
print(f"IR 0.5   : share with p < 0.05 = {np.mean(p1 < 0.05):.1%}   (power, formula {1 - stats.norm.cdf(z_95 - 0.5 * np.sqrt(5)):.1%})")
```

### Power and years needed

```python
def power(ir, years, z_crit=z_95):
    return 1 - stats.norm.cdf(z_crit - ir * np.sqrt(years))

def years_for_power(ir, target=0.8, level=0.05):
    return ((stats.norm.ppf(1 - level) + stats.norm.ppf(target)) / ir) ** 2

rows = []
for ir in [0.25, 0.5, 1.0]:
    rows.append({"IR": ir, "power 3y": power(ir, 3), "power 5y": power(ir, 5), "power 10y": power(ir, 10),
                 "years for 80% power": years_for_power(ir)})
pd.DataFrame(rows).set_index("IR").round(2)
```

### A universe of 1,000 funds

10% skilled with IR 0.5 (alpha 2%, TE 4%), 90% without skill; 10 years of monthly data.

```python
n_funds, share_skilled, years_u = 1_000, 0.10, 10
skilled = np.zeros(n_funds, dtype=bool)
skilled[: int(share_skilled * n_funds)] = True
alpha_true = np.where(skilled, 0.02, 0.0)
A = alpha_true[:, None] / 12 + 0.04 / np.sqrt(12) * rng.standard_normal((n_funds, years_u * 12))
t_u, p_u = alpha_t_test(A)


def screen(reject):
    tp, fp = np.sum(reject & skilled), np.sum(reject & ~skilled)
    return {"significant": tp + fp, "skilled": tp, "lucky": fp,
            "share skilled among significant": tp / max(tp + fp, 1), "skilled funds missed": np.sum(~reject & skilled)}

bonf = p_u < 0.05 / n_funds
screens = pd.DataFrame({
    "p < 5%": screen(p_u < 0.05),
    "t > 3": screen(t_u > 3),
    "Bonferroni (p < 0.005%)": screen(bonf),
}).T
screens
```

As in Example 4: the 5% screen passes 103 funds, 57 skilled and 46 lucky, so only 55% of the "significant" funds are
skilled, and 43 of the 100 skilled funds are missed. Raising the bar to $t > 3$ makes the screen cleaner (71% skilled) but
finds only 10 of the 100. Bonferroni finds one.

### Regression to the mean

Rank the funds on their first 5 years and look at the next 5 years, by decile.

```python
first, second = A[:, :60], A[:, 60:]
alpha_1 = first.mean(axis=1) * 12
alpha_2 = second.mean(axis=1) * 12
decile = pd.qcut(alpha_1, 10, labels=False) + 1

persist = pd.DataFrame({"decile (first 5y)": decile, "alpha first 5y": alpha_1,
                        "alpha next 5y": alpha_2, "true alpha": alpha_true, "skilled": skilled}) \
    .groupby("decile (first 5y)").mean()
persist.style.format({"alpha first 5y": "{:.2%}", "alpha next 5y": "{:.2%}", "true alpha": "{:.2%}", "skilled": "{:.0%}"})
```

The top decile of the first five years averaged 3.7% alpha; in the next five years it delivered 1.0%, close to its average
true alpha of 0.8%. It does contain more skilled funds (40% against 10% overall), so ranking is not useless, but most
of the ranking-period spread was luck. The bottom decile went from −2.9% to about zero.

## 7. Exercises

### Basic

1. A fund has alpha 1.5% per year and TE 3% over 8 years of monthly data. Compute $t$ and the one-sided p-value. Is it
   significant at 5%?
2. How many years are needed for 80% power to detect an IR of 0.3 at the 5% level?
3. A fund's alpha has a p-value of 0.03. A colleague says "so there is a 97% chance the manager is skilled". **Interpret** the
   p-value correctly, and say what extra information would be needed to make a statement about the chance of skill.

### Intermediate

4. Simulate 10,000 funds with IR 0.5 over 3 years and compute the share with $p < 0.05$. Compare with `power(0.5, 3)`.
5. Survivorship: simulate 1,000 funds **without** skill over 10 years. Funds whose cumulative active return after 3 years
   is below −5% close and disappear. Compare the average estimated 10-year alpha of the survivors with the average over all
   funds, had none closed.

### Advanced

6. Apply the Benjamini–Hochberg procedure at a false discovery rate of 10% to the p-values of the 1,000-fund universe:
   sort the p-values $p_{(1)} \le \dots \le p_{(N)}$, find the largest $k$ with $p_{(k)} \le 0.10\,k/N$, and declare the $k$ smallest
   significant. How many discoveries, and what share are skilled?
7. Using the formula for $P(\text{skilled} \mid \text{significant})$, compute it for a share of skilled funds $\pi$ = 1%, 5%, 10%, 30%, with
   IR 0.5, 10 years, and thresholds $t > 1.645$ and $t > 3$.

## 8. Solutions / Checks

```python
# 1.
t_1 = 1.5 / 3 * np.sqrt(8)
print(f"1. t = {t_1:.3f}, p = {stats.t.sf(t_1, 8 * 12 - 1):.4f}")

# 2.
print(f"2. years: {years_for_power(0.3):.1f}")

# 4.
_, p4 = alpha_t_test(active_returns(0.02, 0.04, 3, rng, n_sim))
print(f"4. simulated power {np.mean(p4 < 0.05):.1%}, formula {power(0.5, 3):.1%}")
```

```python
# 5.
S = active_returns(0.0, 0.04, 10, rng, 1_000)
cum_3y = S[:, :36].sum(axis=1)
survives = cum_3y >= -0.05
alpha_10 = S.mean(axis=1) * 12
print(f"5. survivors: {survives.mean():.0%} of funds, average alpha {alpha_10[survives].mean():.2%};  "
      f"all funds: {alpha_10.mean():.2%}")

# 6.
order = np.argsort(p_u)
k_vals = np.arange(1, n_funds + 1)
below = p_u[order] <= 0.10 * k_vals / n_funds
k = k_vals[below].max() if below.any() else 0
bh = np.zeros(n_funds, dtype=bool)
bh[order[:k]] = True
print(f"6. Benjamini-Hochberg: {bh.sum()} discoveries, {np.sum(bh & skilled)} skilled "
      f"({np.sum(bh & skilled) / max(bh.sum(), 1):.0%})")

# 7.
for z_crit, lab in [(z_95, "t > 1.645"), (3.0, "t > 3")]:
    pw, level = power(0.5, 10, z_crit), stats.norm.sf(z_crit)
    out = [pi * pw / (pi * pw + (1 - pi) * level) for pi in [0.01, 0.05, 0.10, 0.30]]
    print(f"7. {lab}: power {pw:.2f}, " + ", ".join(f"pi {pi:.0%}: {o:.0%}" for pi, o in zip([0.01, 0.05, 0.10, 0.30], out)))
```

**1.** $\mathrm{IR} = 0.5$, $t = 0.5\sqrt8 = 1.414$, $p \approx 0.08$: not significant at 5%.

**2.** $(2.486/0.3)^2 = 68.7$ years.

**3.** If the manager had no skill, a track record at least this good would occur 3% of the time. That is a statement about
the data given no skill, not about skill given the data. To get the chance of skill you need the share of skilled managers
among those you could have picked (the prior) and the power of the test against the kind of skill you have in mind;
then Bayes' theorem applies (exercise 7). It also matters how many funds were screened to find this one.

**4.** The simulated share is close to the formula value of about 22%: three years detect a genuine IR of 0.5 about one time in five.

**5.** Survivors show a positive average alpha although no fund has skill; across all funds launched the average is
close to zero. The funds that would have pulled the average down are the ones that closed.

**6.** Benjamini–Hochberg finds 12 funds, 9 of them skilled, against one for Bonferroni. It controls the **expected** share of
lucky funds among the discoveries (at most 10%); with only 12 discoveries a single sample can be above it, as here (3 of 12).

**7.** With the usual 5% threshold, a significant fund is likely to be lucky unless skill is common. Raising the bar to $t > 3$
makes a significant fund much more likely to be skilled (87% instead of 51% for $\pi = 10\%$), but it detects far fewer
skilled funds: power 0.08 instead of 0.47 over 10 years.

## 9. Conclusion

- The alpha t-statistic is about $\mathrm{IR}\sqrt T$; a p-value is the chance of a record this good **without** skill.
- Power is $1 - \Phi(1.645 - \mathrm{IR}\sqrt T)$; detecting an IR of 0.5 with 80% power takes about 25 years.
- Among many funds, a 5% test lets through about 5% of the unskilled; with skill rare, many significant funds are lucky.
  Bonferroni, $t > 3$ and Benjamini–Hochberg trade false discoveries against missed skill.
- Past winners regress toward the mean, and databases of survivors overstate average alpha.
