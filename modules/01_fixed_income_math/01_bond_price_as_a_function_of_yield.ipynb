# Bond Price as a Function of Yield

A bond is a promise of fixed payments: coupons every year and the face value at maturity. Its price is the present value of those
payments, discounted at a single rate, the yield. As a function of the yield the price is a polynomial in the discount factor
$v = 1/(1+y)$: decreasing, convex, and steeper for longer bonds. This notebook builds the price function, inverts it to get the yield
from a market price, and explains premium, discount and pull to par from the shape of the curve.

## 1. Introduction

The Swiss Confederation issues bonds that pay an annual coupon and repay CHF 100 per bond at maturity. Once issued, the coupon is fixed,
but the bond's market price moves every day. The price and the yield are two ways of stating the same thing: given one, the other
follows. When market rates rise, the fixed coupons become less attractive and the price falls.

After this notebook you can:

- price a fixed-coupon bond from its yield, and find the yield from its price;
- explain why bonds trade at par, at a premium or at a discount;
- describe the shape of the price–yield curve: decreasing, convex, steeper for long maturities and low coupons;
- show how a bond's price drifts towards par as it ages (pull to par);
- compute the first derivative of price with respect to yield, the starting point of duration (01/02).

## 2. Theory

### The price formula

A bond with face value $F$ (CHF 100), annual coupon rate $c$, $n$ years to maturity and yield $y$ (annual compounding) pays $cF$ at
the end of years $1, \dots, n$ and $F$ at year $n$. Its price is

$$
P(y) = \sum_{t=1}^{n} \frac{cF}{(1+y)^t} + \frac{F}{(1+y)^n} = cF\,a_{n|y} + F v^n ,
\qquad v = \frac{1}{1+y}, \quad a_{n|y} = \frac{1 - v^n}{y} .
$$

The yield to maturity is the rate $y$ that makes this sum equal to the market price: the IRR of buying the bond and holding it to
maturity (00/03).

### Par, premium, discount

Substitute $a_{n|y} = (1 - v^n)/y$:

$$
P(y) = F\Big[\frac{c}{y}(1 - v^n) + v^n\Big] = F\Big[1 + \Big(\frac{c}{y} - 1\Big)(1 - v^n)\Big] .
$$

Since $0 < 1 - v^n < 1$ for $y > 0$:

- $c = y$: the price is exactly $F$ (**par**);
- $c > y$: the price is above $F$ (**premium**): the coupons pay more than the market rate;
- $c < y$: the price is below $F$ (**discount**).

### A polynomial in $v$

In the discount factor the price is a polynomial with positive coefficients,

$$
P(v) = cF(v + v^2 + \dots + v^n) + F v^n, \qquad \frac{dP}{dv} = cF(1 + 2v + \dots + n v^{n-1}) + nF v^{n-1} > 0 .
$$

Since $v = (1+y)^{-1}$ decreases in $y$ with $dv/dy = -v^2$, the chain rule gives

$$
\frac{dP}{dy} = -\frac{1}{1+y}\sum_{t=1}^{n} t \cdot \frac{C_t}{(1+y)^t} < 0 ,
$$

where $C_t$ is the cash flow at $t$. The price falls when the yield rises, and the rate of fall is a cash-flow-weighted sum of payment
times: long bonds and late payments make the curve steep. Each term $C_t(1+y)^{-t}$ is convex in $y$, so the price is convex too: it falls
less and less steeply as the yield rises.

### Zero-coupon bonds

With $c = 0$, $P = F(1+y)^{-n}$ and $dP/dy = -nP/(1+y)$. A zero-coupon bond's relative price sensitivity is its maturity divided by
$1 + y$: the purest form of interest-rate risk.

### Pull to par

At maturity the price is $F$ whatever the yield. If the yield stays constant, a premium bond's price falls towards par and a discount
bond's price rises towards par as time passes. The total return of holding the bond for a year at constant yield is exactly $y$: coupon
plus price change.

## 3. Worked Examples

### Example 1: a ten-year Confederation bond

Coupon 1.5%, 10 years, yield 0.8%. The annuity factor is $a_{10|0.008} = (1 - 1.008^{-10})/0.008 = 9.5737$ and $v^{10} = 0.92341$, so
$P = 1.5 \times 9.5762 + 100 \times 0.92339 = 14.36 + 92.34 = 106.70$. The coupon is above the yield: the bond trades at a premium.

### Example 2: the same bond at 2.5%

$a_{10|0.025} = 8.7521$, $v^{10} = 0.78120$, so $P = 13.13 + 78.12 = 91.25$: a discount. A move of 1.7 percentage points in the yield
moved the price by 15.5 points.

### Example 3: yield from price

A five-year 1% bond trades at 97.50. Try $y = 1.5\%$: $P = 97.61$, slightly too high; try $1.6\%$: $P = 97.14$, too low. Linear
interpolation gives $y \approx 1.5\% + 0.1\% \times (97.61 - 97.50)/(97.61 - 97.14) = 1.523\%$. Newton's method (section 5) does this
systematically.

### Example 4: pull to par

The premium bond of Example 1, one year later, still at a yield of 0.8%: 9 years remain and $P = 106.06$. The price fell by 0.64 while
the holder received a coupon of 1.50, a total of 0.86 on 106.70, or 0.80%: exactly the yield.

## 4. Finance Application

### Where it matters

- **Trading and quoting.** Bonds are quoted by price in Switzerland and Europe, by yield in some other markets; traders convert
  constantly between the two with this formula.
- **Interest-rate risk.** The slope $dP/dy$ is the starting point of duration and of hedging (01/02).
- **Portfolio accounting.** Funds report the yield to maturity of their bond portfolios; mark-to-market valuations reprice each bond
  at current yields.
- **Pension funds and insurers.** Liabilities are fixed future payments too; their present value moves with yields exactly like a bond's.

### The usual misreadings

- "A premium bond is a bad deal." Its higher coupons make up for the price above par; at the same yield it returns the same.
- "The yield is the return I will earn." Only if the bond is held to maturity, does not default, and coupons are reinvested at the yield.
- "Bonds are safe, prices don't move." A 30-year bond loses about a fifth of its value when yields rise by one percentage point.
- "Price and yield move by the same amount." The relationship is curved, and its slope depends on maturity and coupon.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp

pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

### Price function and its derivative

```python
def bond_cash_flows(coupon, years, face=100.0):
    cf = np.full(years, coupon * face)
    cf[-1] += face
    return cf

def bond_price(y, coupon, years, face=100.0):
    t = np.arange(1, years + 1)
    return np.sum(bond_cash_flows(coupon, years, face) / (1 + y) ** t)

def bond_price_slope(y, coupon, years, face=100.0):
    # dP/dy = -(1/(1+y)) * sum t * C_t / (1+y)^t
    t = np.arange(1, years + 1)
    return -np.sum(t * bond_cash_flows(coupon, years, face) / (1 + y) ** (t + 1))

print(f"Example 1: price at 0.8%  = {bond_price(0.008, 0.015, 10):.4f}")
print(f"Example 2: price at 2.5%  = {bond_price(0.025, 0.015, 10):.4f}")
print(f"slope dP/dy at 0.8%       = {bond_price_slope(0.008, 0.015, 10):.2f} CHF per unit of yield "
      f"({bond_price_slope(0.008, 0.015, 10) / 100:.4f} CHF per percentage point)")
```

### The derivative symbolically

SymPy differentiates the polynomial in $v$ and applies the chain rule; the result matches the formula of section 2.

```python
y_sym = sp.symbols("y", positive=True)
c_, F_ = sp.Rational(15, 1000), 100
P_sym = sum(c_ * F_ / (1 + y_sym) ** t for t in range(1, 11)) + F_ / (1 + y_sym) ** 10
dP_sym = sp.diff(P_sym, y_sym)
print("P(0.008)  =", sp.N(P_sym.subs(y_sym, sp.Rational(8, 1000)), 8))
print("P'(0.008) =", sp.N(dP_sym.subs(y_sym, sp.Rational(8, 1000)), 8))
print("P''(0.008) =", sp.N(sp.diff(P_sym, y_sym, 2).subs(y_sym, sp.Rational(8, 1000)), 8), "(positive: convex)")
```

### Yield from price with Newton's method

```python
def bond_yield(price, coupon, years, face=100.0, guess=None, tol=1e-12):
    y = coupon if guess is None else guess          # the coupon rate is a good first guess
    for _ in range(50):
        step = (bond_price(y, coupon, years, face) - price) / bond_price_slope(y, coupon, years, face)
        y -= step
        if abs(step) < tol:
            return y
    raise RuntimeError("no convergence")

y3 = bond_yield(97.50, 0.01, 5)
print(f"Example 3: yield of the 5-year 1% bond at 97.50 = {y3:.6%}")
print(f"check: price at that yield = {bond_price(y3, 0.01, 5):.6f}")
```

### Par, premium and discount

```python
rows = []
for coupon in [0.0, 0.01, 0.02, 0.03]:
    for y in [0.01, 0.02, 0.03]:
        p = bond_price(y, coupon, 10)
        status = "par" if abs(p - 100) < 1e-9 else ("premium" if p > 100 else "discount")
        rows.append({"coupon": coupon, "yield": y, "price": p, "status": status})
pd.DataFrame(rows).pivot(index="coupon", columns="yield", values="price")
```

On the diagonal, where coupon equals yield, every ten-year bond is priced at exactly 100. Above the diagonal (coupon > yield) prices
exceed par; below they are under par. The zero-coupon row is the furthest below par at every yield.

### Sensitivity by maturity and coupon

The relative price change for a one-percentage-point rise in the yield, from 1% to 2%.

```python
rows = []
for years in [2, 5, 10, 20, 30]:
    for coupon in [0.0, 0.01, 0.04]:
        p1, p2 = bond_price(0.01, coupon, years), bond_price(0.02, coupon, years)
        rows.append({"years": years, "coupon": coupon, "price at 1%": p1, "price at 2%": p2,
                     "change (%)": (p2 / p1 - 1) * 100})
sens = pd.DataFrame(rows)
sens.pivot(index="years", columns="coupon", values="change (%)")
```

A two-year bond loses about 2% when the yield rises by a point; a thirty-year zero-coupon bond about 25%. For the same maturity,
lower coupons mean larger losses, because more of the value sits in the distant final payment.

## 6. Visualizations

```python
ys = np.linspace(-0.01, 0.08, 300)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for years in [2, 10, 30]:
    axes[0].plot(ys * 100, [bond_price(y, 0.02, years) for y in ys], label=f"{years} years")
axes[0].axhline(100, color="grey", lw=0.8)
axes[0].axvline(2, color="grey", lw=0.8, ls=":")
axes[0].set_title("Price of a 2% coupon bond against its yield")
axes[0].set_xlabel("yield (% per year)")
axes[0].set_ylabel("price (CHF per CHF 100 face)")
axes[0].legend(title="maturity")

for coupon in [0.0, 0.02, 0.05]:
    axes[1].plot(ys * 100, [bond_price(y, coupon, 10) for y in ys], label=f"coupon {coupon:.0%}")
axes[1].axhline(100, color="grey", lw=0.8)
axes[1].set_title("Ten-year bonds with different coupons")
axes[1].set_xlabel("yield (% per year)")
axes[1].set_ylabel("price (CHF per CHF 100 face)")
axes[1].legend()
for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Left: all three 2% bonds cost 100 at a 2% yield (the curves cross at par). The longer the maturity, the steeper and more curved the
price–yield relation. Right: at the same maturity, higher coupons give higher prices, and each bond crosses 100 where the yield equals
its coupon (0%, 2% and 5%). In percentage terms the zero-coupon bond falls fastest, as the table above showed.

```python
y0 = 0.02
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(ys * 100, [bond_price(y, 0.02, 30) for y in ys], label="price P(y)")
slope = bond_price_slope(y0, 0.02, 30)
ax.plot(ys * 100, 100 + slope * (ys - y0), "--", label=f"tangent at 2%, slope {slope / 100:.1f} CHF per point")
ax.plot(y0 * 100, 100, "ko")
ax.set_ylim(40, 180)
ax.set_title("30-year 2% bond: the price curve lies above its tangent (convexity)")
ax.set_xlabel("yield (% per year)")
ax.set_ylabel("price (CHF per CHF 100 face)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

The tangent at the current yield is the linear approximation used for duration (01/02). Because the price curve is convex, it lies above
the tangent on both sides: the linear approximation overstates losses when yields rise and understates gains when they fall. That gap is
convexity.

```python
years_left = np.arange(10, -1, -1)
fig, ax = plt.subplots(figsize=(10, 5))
for coupon, y, label in [(0.03, 0.01, "premium: 3% coupon at 1% yield"),
                         (0.02, 0.02, "par: 2% coupon at 2% yield"),
                         (0.00, 0.02, "discount: zero coupon at 2% yield")]:
    prices = [bond_price(y, coupon, n) if n > 0 else 100.0 for n in years_left]
    ax.plot(10 - years_left, prices, "o-", label=label)
ax.set_title("Pull to par: prices at a constant yield as the bond ages")
ax.set_xlabel("years since issue (10-year bond)")
ax.set_ylabel("price after the coupon (CHF per CHF 100 face)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

At a constant yield, all three prices converge to 100 at maturity. The premium bond loses value every year and the discount bond gains:
the price change makes up the difference between the coupon and the yield, so every bond returns exactly its yield each year.

## 8. Solutions / Checks

```python
# 1.
print("1.", {f"{y:.1%}": round(bond_price(y, 0.005, 7), 4) for y in [0.0, 0.005, 0.015]})

# 2.
print(f"2. yield {bond_yield(101.20, 0.02, 3):.4%}")

# 3.
p_new = bond_price(0.03, 0.02, 30)
print(f"3. price after the rise {p_new:.2f}, change {p_new - 100:.2f} CHF per 100")

# 4.
p0 = bond_price(0.008, 0.015, 10)
for y_end in [0.008, 0.013]:
    p1 = bond_price(y_end, 0.015, 9)
    print(f"4. yield after one year {y_end:.1%}: return {(p1 + 1.5) / p0 - 1:.4%}")
print(f"   slope approximation of the extra loss: {bond_price_slope(0.008, 0.015, 9) * 0.005 / p0:.4%} of the price")

# 7.
p7 = bond_price(-0.006, 0.005, 10)
print(f"7. price at -0.6%: {p7:.2f}; sum of cash flows: {bond_cash_flows(0.005, 10).sum():.2f}")
```

**3.** The price falls to about 80.4, a mark-to-market loss of almost 20%. If the fund holds to maturity and the Confederation pays, the fund
still receives every coupon and the face value; the loss is the opportunity cost of earning 2% while new bonds pay 3%. It is real
economically, and it shows in the fund's balance sheet if its liabilities are not valued at the same yields.

**4.** At constant yield the return is exactly 0.80%. With the rise to 1.3% after a year the return is about −3.3%. The slope
approximation (using the nine-year bond at 0.8%) predicts a loss of 4.2% of the price on top of the 0.8% carry, about −3.4% in total: a
little worse than the actual result, because convexity cushions the loss.

**5.** The relative sensitivity is $-\frac{1}{P}\frac{dP}{dy} = \frac{1}{1+y}\cdot\frac{\sum_t t\,C_t v^t}{\sum_t C_t v^t}$, the present-value-weighted
average payment time divided by $1 + y$. For the zero it is $10/(1+y)$; for the coupon bond the average time is always below 10 because some
payments come earlier. The two can never be equal.

**6.** $P''(y) = \sum_t t(t+1) C_t (1+y)^{-t-2} > 0$. At $y = 0$ every discount factor is 1; as $y \to \infty$ every discount factor goes to 0.

**7.** The price is about 111.4, more than the 105 of cash flows the bond will ever pay. With a negative yield, holding cash in a bank account or
at the central bank cost more than buying the bond (the SNB charged −0.75% on large sight deposits), so a guaranteed small loss was the best
available safe option.

## 9. Conclusion

- A bond's price is the present value of its coupons and face value at a single yield; the yield is the IRR of buying and holding it.
- Coupon equal to yield means par; coupon above yield means a premium; below means a discount.
- The price is a polynomial in $v = 1/(1+y)$; it falls and is convex in $y$, and $dP/dy = -\frac{1}{1+y}\sum_t t\,C_t v^t$.
- Long maturities and low coupons make the curve steeper; a 30-year zero loses about a quarter of its value per point of yield.
- At a constant yield, prices converge to par and each year's return equals the yield.
