# Duration and Convexity: First and Second Derivatives of the Bond Price

The bond price is a smooth, falling, convex function of the yield (01/01). Its first derivative, scaled by the price, is the
**modified duration**: the percentage price change per unit change in the yield. Its second derivative, scaled the same way, is the
**convexity**. Both have a second reading as averages of payment times, which is why duration is quoted in years. This notebook derives
both, computes them for real-looking Swiss bonds, and uses them to hedge and to immunize a liability.

## 1. Introduction

A bond portfolio manager needs one number that says how much the portfolio loses if yields rise by one percentage point. Duration is
that number. For larger moves, or to compare portfolios with the same duration, the second number, convexity, says how much the first
one is off. Together they are the language of interest-rate risk in banks, insurers, pension funds and central banks.

After this notebook you can:

- compute Macaulay duration, modified duration, DV01 and convexity from cash flows;
- show that they are the scaled first and second derivatives of the price;
- interpret duration as the present-value-weighted average time of the cash flows;
- add durations across a portfolio and hedge one bond with another;
- immunize a liability, and explain why a barbell has more convexity than a bullet with the same duration.

## 2. Theory

### First derivative: duration

With cash flows $C_t$ at times $t = 1, \dots, n$ (years) and yield $y$ (annual compounding),

$$
P(y) = \sum_t C_t (1+y)^{-t}, \qquad P'(y) = -\sum_t t\, C_t (1+y)^{-t-1} .
$$

Define the present-value weights $w_t = C_t(1+y)^{-t}/P$, which are positive and sum to 1. Then

$$
D_{\text{mac}} = \sum_t t\, w_t, \qquad D_{\text{mod}} = -\frac{P'(y)}{P(y)} = \frac{D_{\text{mac}}}{1+y} .
$$

**Macaulay duration** is the average payment time, weighted by present value, in years. **Modified duration** is the relative price
sensitivity: a yield rise of $\Delta y$ changes the price by about $-D_{\text{mod}}\,\Delta y$ in relative terms. **DV01** (dollar value
of a basis point, here CHF value) is $-P'(y) \times 0.0001$: the CHF price change per 0.01 percentage point.

### Second derivative: convexity

$$
P''(y) = \sum_t t(t+1)\, C_t (1+y)^{-t-2}, \qquad \text{Conv} = \frac{P''(y)}{P(y)} = \frac{1}{(1+y)^2}\sum_t t(t+1)\, w_t .
$$

Convexity is positive for any bond with positive cash flows. Combining both (01/03 makes this precise),

$$
\frac{\Delta P}{P} \approx -D_{\text{mod}}\,\Delta y + \tfrac12\,\text{Conv}\,(\Delta y)^2 .
$$

### Special cases

| bond | Macaulay duration | convexity |
|---|---|---|
| zero coupon, maturity $n$ | $n$ | $n(n+1)/(1+y)^2$ |
| perpetuity paying $C$ | $(1+y)/y$ | $2/y^2$ |
| coupon bond, maturity $n$ | $< n$ | $< n(n+1)/(1+y)^2$ |

A coupon bond's duration is below its maturity because some cash arrives earlier. At low yields durations are long: a perpetuity at 1%
has a duration of 101 years.

### Portfolios

If a portfolio holds bonds with values $V_i$, its price is $\sum_i V_i(y)$, so (for a common yield change) its derivatives add:

$$
D_{\text{mod},p} = \sum_i \frac{V_i}{V} D_{\text{mod},i}, \qquad \text{Conv}_p = \sum_i \frac{V_i}{V}\text{Conv}_i, \qquad \text{DV01}_p = \sum_i \text{DV01}_i .
$$

### Hedging and immunization

To hedge a position with DV01 $d_A$ using a bond with DV01 $d_H$, sell $h = d_A/d_H$ units of the hedge: the first derivative of the
combined position is then zero.

To **immunize** a liability $L$ due at time $T$, hold assets with the same present value and the same duration. Then the surplus
$S(y) = A(y) - L(y)$ has $S = 0$ and $S' = 0$ at today's yield, and if the assets have more convexity than the liability, $S'' > 0$:
the surplus has a local minimum at today's yield and is positive after any small parallel shift.

### Barbell and bullet

A **bullet** puts money in one maturity; a **barbell** splits it between a short and a long maturity with the same duration. Since
convexity grows roughly with the square of the payment time, spreading the times out at a fixed average raises the average of $t^2$:
the barbell has more convexity.

## 3. Worked Examples

### Example 1: a ten-year 1.5% bond at 0.8%

The price is 106.70 (01/01) and $P'(0.008) = -993.50$. Modified duration $993.50/106.70 = 9.31$, Macaulay duration
$9.31 \times 1.008 = 9.39$ years (below the 10-year maturity), DV01 $= 0.0994$ CHF per CHF 100 face. A one-point rise in the yield costs
about 9.3% of the price.

### Example 2: a zero and a perpetuity at 2%

A ten-year zero: Macaulay duration 10, modified $10/1.02 = 9.80$, convexity $10 \times 11/1.02^2 = 105.7$. A perpetuity: duration
$1.02/0.02 = 51$ years, modified $50$, convexity $2/0.02^2 = 5{,}000$.

### Example 3: hedging

A trader holds CHF 10 million face of the ten-year bond (DV01 per 100 face 0.0994, so CHF 9,935 per basis point) and hedges with a
two-year 0.5% bond at a yield of 0.6% (DV01 about 0.0199 per 100 face). Hedge: sell $9{,}935/0.0199 \times 100 \approx$ CHF 50 million face
of the two-year bond. The position is protected against small parallel shifts only; a steepening of the curve is not hedged.

### Example 4: immunizing a pension payment

A pension fund owes CHF 10 million in 7 years; at 2% its present value is CHF 8.71 million and its duration is 7. A **bullet** of
seven-year zeros matches it exactly. A **barbell** of 2-year and 20-year zeros with the same value and duration needs weights $w$ and
$1 - w$ with $2w + 20(1 - w) = 7$, so $w = 0.722$. Its convexity is $0.722 \times 2\cdot3/1.02^2 + 0.278 \times 20\cdot21/1.02^2 = 116$,
against $7\cdot8/1.02^2 = 54$ for the liability: after any parallel shift, the barbell is worth more than the liability.

## 4. Finance Application

### Where it matters

- **Risk limits.** Bond funds and bank treasuries set limits in duration or DV01; benchmark-relative managers report duration relative
  to the index.
- **Hedging.** Futures and swaps are sized by DV01 ratios; a mortgage lender hedges fixed-rate loans with swaps of the same DV01.
- **Asset–liability management.** Pension funds and insurers match the duration of their bonds to that of their liabilities; Swiss
  pension funds discount liabilities at the technical rate, which makes their surplus sensitive to rates.
- **Banking regulation.** Interest-rate risk in the banking book (IRRBB) is measured by the change in economic value under rate shocks,
  a duration-and-convexity calculation.

### The usual misreadings

- "Duration is the time until the bond pays back." It is the PV-weighted average payment time, a measure of rate sensitivity.
- "Duration is a fixed property of the bond." It changes with the yield and as time passes.
- "Equal duration means equal risk." Only for small parallel shifts; convexity and curve shape matter for larger and non-parallel moves.
- "Convexity is free." Barbells with more convexity usually yield less than the matching bullet; the market prices it.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

### Duration and convexity from cash flows

```python
def bond_cash_flows(coupon, years, face=100.0):
    cf = np.full(years, coupon * face)
    cf[-1] += face
    return cf

def analytics(cash_flows, y):
    cf = np.asarray(cash_flows, dtype=float)
    t = np.arange(1, len(cf) + 1)
    pv = cf / (1 + y) ** t
    price = pv.sum()
    w = pv / price
    d_mac = np.sum(t * w)
    d_mod = d_mac / (1 + y)
    conv = np.sum(t * (t + 1) * w) / (1 + y) ** 2
    return {"price": price, "Macaulay duration": d_mac, "modified duration": d_mod,
            "DV01 (CHF per 100 face)": d_mod * price * 1e-4, "convexity": conv}

bonds = {
    "2y 0.5% at 0.6%": (bond_cash_flows(0.005, 2), 0.006),
    "10y 1.5% at 0.8%": (bond_cash_flows(0.015, 10), 0.008),
    "10y zero at 2%": (bond_cash_flows(0.0, 10), 0.02),
    "30y 2% at 2%": (bond_cash_flows(0.02, 30), 0.02),
    "30y zero at 2%": (bond_cash_flows(0.0, 30), 0.02),
}
table = pd.DataFrame({name: analytics(cf, y) for name, (cf, y) in bonds.items()}).T
table
```

The ten-year 1.5% bond has a Macaulay duration of 9.39 years and a modified duration of 9.31, as in Example 1. Zero-coupon bonds have
duration equal to maturity. The 30-year zero's convexity is more than twice that of the 30-year coupon bond: its single payment sits at
the far end.

### Checking against numerical derivatives

```python
def price(cf, y):
    t = np.arange(1, len(cf) + 1)
    return np.sum(cf / (1 + y) ** t)

h = 1e-4
for name, (cf, y) in bonds.items():
    p0, p_up, p_dn = price(cf, y), price(cf, y + h), price(cf, y - h)
    d_num = -(p_up - p_dn) / (2 * h) / p0
    c_num = (p_up - 2 * p0 + p_dn) / h**2 / p0
    a = analytics(cf, y)
    print(f"{name:18s} modified duration {a['modified duration']:.6f} vs {d_num:.6f}, "
          f"convexity {a['convexity']:.4f} vs {c_num:.4f}")
```

Central finite differences reproduce the analytic formulas to the printed digits: duration and convexity are the first and second
derivatives of the price, divided by the price.

### Duration of a portfolio, and a DV01 hedge

```python
a10 = analytics(*bonds["10y 1.5% at 0.8%"])
a2 = analytics(*bonds["2y 0.5% at 0.6%"])
face_10 = 10_000_000                                   # CHF face held
dv01_position = a10["DV01 (CHF per 100 face)"] * face_10 / 100
hedge_face = dv01_position / a2["DV01 (CHF per 100 face)"] * 100
print(f"DV01 of the position: {dv01_position:,.0f} CHF per basis point")
print(f"hedge: sell {hedge_face:,.0f} CHF face of the 2-year bond")

# value of position + hedge after parallel shifts of both yields
for shift_bp in [-100, -25, 25, 100]:
    s = shift_bp / 1e4
    pnl_bond = (price(bonds["10y 1.5% at 0.8%"][0], 0.008 + s) - a10["price"]) * face_10 / 100
    pnl_hedge = -(price(bonds["2y 0.5% at 0.6%"][0], 0.006 + s) - a2["price"]) * hedge_face / 100
    print(f"shift {shift_bp:+4d} bp: bond {pnl_bond:>12,.0f}, hedge {pnl_hedge:>12,.0f}, total {pnl_bond + pnl_hedge:>9,.0f} CHF")
```

Unhedged, a one-point move changes the position by about CHF 0.9–1.0 million. Hedged, the residual is a few thousand francs at ±25 bp
and positive for both ±100 bp: the long ten-year bond has more convexity than the short two-year hedge, so the hedged position gains
from large moves in either direction.

### Immunizing a liability: bullet against barbell

```python
y0 = 0.02
L_face, T = 10_000_000, 7
L_pv = L_face / (1 + y0) ** T
w_short = (20 - T) / (20 - 2)                     # 2w + 20(1 - w) = 7
print(f"liability PV {L_pv:,.0f} CHF, barbell weight in 2-year zeros {w_short:.3f}")

face_2 = w_short * L_pv * (1 + y0) ** 2           # face values that cost the right amounts today
face_20 = (1 - w_short) * L_pv * (1 + y0) ** 20
face_7 = L_pv * (1 + y0) ** 7

def zero_pv(face, years, y):
    return face / (1 + y) ** years

conv = lambda n: n * (n + 1) / (1 + y0) ** 2
print(f"convexity: liability {conv(7):.1f}, bullet {conv(7):.1f}, barbell {w_short * conv(2) + (1 - w_short) * conv(20):.1f}")

shifts = np.linspace(-0.02, 0.02, 81)
surplus_bullet = zero_pv(face_7, 7, y0 + shifts) - zero_pv(L_face, 7, y0 + shifts)
surplus_barbell = zero_pv(face_2, 2, y0 + shifts) + zero_pv(face_20, 20, y0 + shifts) - zero_pv(L_face, 7, y0 + shifts)
for s in [-0.02, -0.01, 0.01, 0.02]:
    i = np.argmin(np.abs(shifts - s))
    print(f"shift {s:+.0%}: barbell surplus {surplus_barbell[i]:>10,.0f} CHF")
```

The bullet matches the liability exactly, so its surplus is zero for every shift. The barbell's surplus is zero at today's yield, has
zero slope there (same duration), and is positive for any parallel shift: about CHF 25,000–30,000 for a one-point move and CHF 89,000–135,000 for two points. That
is the convexity difference at work. The catch: it assumes parallel shifts. If short rates fall while long rates rise, the barbell loses.

```python
# A twist: 2-year yield falls by 1 point, 20-year yield rises by 1 point, 7-year unchanged
twist = (zero_pv(face_2, 2, y0 - 0.01) + zero_pv(face_20, 20, y0 + 0.01)) - zero_pv(L_face, 7, y0)
print(f"barbell surplus after a steepening twist: {twist:,.0f} CHF")
```

A one-point steepening around the seven-year point costs the barbell about CHF 300,000, ten times what a parallel shift of the same size
earns. Convexity protects against parallel moves only.

## 6. Visualizations

```python
cf, y = bonds["10y 1.5% at 0.8%"]
t = np.arange(1, len(cf) + 1)
pv = cf / (1 + y) ** t
d_mac = analytics(cf, y)["Macaulay duration"]

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.bar(t, pv, color="C0", alpha=0.8, label="present value of each cash flow")
ax.plot(d_mac, 0, "k^", ms=18, label=f"Macaulay duration = {d_mac:.2f} years")
ax.set_title("Duration as the balance point of discounted cash flows: 10-year 1.5% bond at 0.8%")
ax.set_xlabel("payment time (years)")
ax.set_ylabel("present value (CHF per 100 face)")
ax.legend(loc="upper left")
ax.grid(True, alpha=0.3)
plt.show()
```

The discounted cash flows are masses on a ruler; Macaulay duration is the point where the ruler balances. The final repayment dominates,
so the balance point sits just below ten years.

```python
maturities = np.arange(1, 51)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for coupon in [0.0, 0.01, 0.03, 0.06]:
    d = [analytics(bond_cash_flows(coupon, n), 0.03)["Macaulay duration"] for n in maturities]
    axes[0].plot(maturities, d, label=f"coupon {coupon:.0%}")
axes[0].axhline(1.03 / 0.03, color="grey", ls=":", label="perpetuity (1 + y) / y")
axes[0].set_title("Macaulay duration at a 3% yield")
axes[0].set_xlabel("maturity (years)")
axes[0].set_ylabel("duration (years)")
axes[0].legend()

for years in [5, 10, 30]:
    ys = np.linspace(0.0, 0.08, 100)
    axes[1].plot(ys * 100, [analytics(bond_cash_flows(0.02, years), yy)["modified duration"] for yy in ys], label=f"{years} years")
axes[1].set_title("Modified duration of 2% coupon bonds against yield")
axes[1].set_xlabel("yield (% per year)")
axes[1].set_ylabel("modified duration")
axes[1].legend()
for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Left: zero-coupon durations grow one for one with maturity; coupon bonds' durations grow more slowly and level off; as maturity grows without
bound they approach the perpetuity duration of 34.3 years at 3% (within 50 years only the 1% coupon gets close). High coupons flatten
the curve sooner. Right: duration falls as the yield rises, most for long bonds, because
higher yields shrink the weight of distant payments. This is convexity seen from another angle: the slope of the price curve flattens.

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(shifts * 100, surplus_barbell / 1e3, label="barbell (2y + 20y zeros) - liability")
ax.plot(shifts * 100, surplus_bullet / 1e3, label="bullet (7y zero) - liability")
ax.axhline(0, color="k", lw=0.8)
ax.set_title("Immunized surplus after a parallel shift of all yields")
ax.set_xlabel("parallel yield shift (percentage points)")
ax.set_ylabel("surplus (CHF thousand)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

The barbell's surplus is a parabola touching zero at today's yield: the first derivative is zero (equal durations) and the second positive
(more convexity than the liability).

## 8. Solutions / Checks

```python
# 1.
print("1.", {k: round(v, 4) for k, v in analytics(bond_cash_flows(0.03, 5), 0.02).items()})

# 2.
for dy in [0.005, -0.015]:
    print(f"2. dy = {dy:+.3f}: {(-6.5 * dy + 0.5 * 55 * dy**2):.4%}")

# 4.
for yy in [0.01, 0.02, 0.05]:
    print(f"4. y = {yy:.0%}: duration {analytics(np.full(2000, 1.0), yy)['Macaulay duration']:.3f}, formula {(1 + yy) / yy:.3f}")

# 5.
holdings = {2: 4e6, 10: 3e6, 30: 3e6}
an = {n: analytics(bond_cash_flows(0.02, n), 0.02) for n in holdings}
V = sum(holdings.values())
D_p = sum(holdings[n] / V * an[n]["modified duration"] for n in holdings)
C_p = sum(holdings[n] / V * an[n]["convexity"] for n in holdings)
D_zero = 2 / 1.02
x = V * (D_p - 8) / (8 - D_zero)            # (V D_p + x D_zero) / (V + x) = 8
print(f"5. modified duration {D_p:.3f}, convexity {C_p:.2f}, buy {x:,.0f} CHF of 2-year zeros")
```

**3.** A modified duration near 7 means the fund loses roughly $7 \times 0.75\% \approx 5\%$ if all its yields rise by 0.75 points at once,
partly offset by about one year of yield income. You would also need to know whether "duration" means Macaulay or modified, the fund's
convexity, how the rise is spread across maturities (curve shape), and the spread duration of any corporate bonds.

**5.** The 30-year bond dominates: CHF 3 million of it contributes $0.3 \times 22.4 = 6.7$ of the 10.2 years. Value-weighted averaging gives
$(V D_p + x D_z)/(V + x) = 8$, so about CHF 3.6 million of 2-year zeros (modified duration 1.96) are needed. Selling some of the 30-year
bond would achieve the same with less capital.

**6.** All weights $w_t$ are positive and some weight sits on $t < n$, so $\sum t w_t < n$. Differentiating $D = \sum_t t w_t$ with
$dw_t/dy = -\frac{w_t}{1+y}(t - D)$ gives $dD/dy = -\frac{1}{1+y}\sum_t t\,w_t(t - D) = -\frac{1}{1+y}\operatorname{Var}_w(t) < 0$.

```python
# 7.
w1 = (30 - 7) / (30 - 1)
f1, f30 = w1 * L_pv * (1 + y0), (1 - w1) * L_pv * (1 + y0) ** 30
print(f"7. weight in 1-year zeros {w1:.3f}, convexity {w1 * conv(1) + (1 - w1) * conv(30):.1f}")
for s in [-0.01, 0.01]:
    surplus = zero_pv(f1, 1, y0 + s) + zero_pv(f30, 30, y0 + s) - zero_pv(L_face, 7, y0 + s)
    print(f"   shift {s:+.0%}: surplus {surplus:,.0f} CHF")
```

**7.** The wider barbell has much more convexity and a larger surplus after parallel shifts, but it is far more exposed to non-parallel
moves: a fall in the 1-year rate and a rise in the 30-year rate hurt it badly, and it has to be rebalanced as durations drift apart over time.
The bullet matches the liability's cash flow exactly and needs no rebalancing, which is why actuaries prefer cash-flow matching when the
bonds exist.

## 9. Conclusion

- Modified duration is $-P'/P$, convexity $P''/P$: the scaled first and second derivatives of the price.
- Macaulay duration is the PV-weighted average payment time; for a zero it equals the maturity.
- $\Delta P/P \approx -D_{\text{mod}}\Delta y + \tfrac12\text{Conv}(\Delta y)^2$; DV01 is the CHF price change per basis point.
- Durations and convexities add with value weights; DV01s add directly. Hedge ratios are DV01 ratios.
- Immunization matches PV and duration; extra convexity makes the surplus positive after parallel shifts, but not after twists.
