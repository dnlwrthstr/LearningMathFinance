# Volatility and the Units of Risk

Volatility is the standard deviation of returns. The definition is one line, yet risk numbers are misread all the time because of
units: variance is in "percent squared", a daily volatility is not an annual one, a percentage change in a rate is not a change in
percentage points, and a volatility of returns becomes a volatility in francs only once it is multiplied by a position. This notebook
derives the scaling rules for variance and standard deviation, puts units on every number, and shows what a "three-sigma day" means.

## 1. Introduction

A risk report says: "Portfolio volatility 9.5%, VaR CHF 2.1 million, yesterday's move −2.8 sigma." Each number has hidden units: annual or
daily, percent of what, sigma estimated over which window. Before any model (VaR, optimization, regression) the units have to be right.

After this notebook you can:

- compute variance and standard deviation of returns with the right degrees of freedom and state their units;
- convert a return volatility into a CHF volatility of a position, and apply leverage;
- distinguish percent and percentage points, and basis points;
- standardize a return into a z-score and judge how unusual it is;
- compare standard deviation with mean absolute deviation and downside deviation.

## 2. Theory

### Variance and standard deviation

For returns $r_1, \dots, r_n$ with sample mean $\bar r$:

$$
s^2 = \frac{1}{n - 1}\sum_{i=1}^n (r_i - \bar r)^2, \qquad s = \sqrt{s^2} .
$$

Dividing by $n - 1$ (Bessel's correction) makes $s^2$ an unbiased estimator of $\sigma^2$, because the deviations are measured from the
sample mean, which is closer to the data than the true mean: $E\big[\sum_i (r_i - \bar r)^2\big] = (n - 1)\sigma^2$.

### Units

If returns are in decimals, the variance is in decimals squared; if in percent, in "percent squared" (%²). A daily return volatility of
1.2% has a variance of $1.44\,\%^2 = 0.000144$. Variance is useful for algebra (it adds for independent sums); standard deviation is
useful for interpretation (same units as the returns).

### Linear transformations

For constants $a$ and $b$,

$$
\operatorname{Var}(aX + b) = a^2\operatorname{Var}(X), \qquad \operatorname{sd}(aX + b) = |a|\operatorname{sd}(X) .
$$

Consequences:

- **CHF volatility.** A position of $V$ francs with return volatility $\sigma$ has a P&L volatility of $V\sigma$ francs.
- **Leverage.** Borrowing to hold $L$ times the capital multiplies the return volatility by $L$; the borrowing rate (a constant $b$) shifts
  the mean only.
- **Currency of measurement.** Converting a constant amount does not change percentage volatility; a fluctuating exchange rate does
  (variance of a sum, 03/01).

### Percent, percentage points, basis points

A rate moving from 1.0% to 1.5% has risen by 0.5 **percentage points** (50 **basis points**) or by 50 **percent**. Volatility of yields
is usually quoted in basis points per year; volatility of returns in percent.

### Standardizing

The z-score $z = (r - \mu)/\sigma$ measures a move in units of its own volatility. If returns were normal, $|z| > 2$ would occur on about
4.6% of days and $|z| > 3$ on 0.27%. Real daily returns have fatter tails (02/03), so "three-sigma days" are much more frequent.

### Other dispersion measures

- **Mean absolute deviation** $\frac1n\sum_i |r_i - \bar r|$, in the same units; for a normal distribution it equals $\sqrt{2/\pi}\,\sigma \approx 0.80\,\sigma$.
- **Downside deviation** $\sqrt{\frac1n\sum_i \min(r_i - \tau, 0)^2}$ below a target $\tau$, used in the Sortino ratio; it ignores upside
  dispersion.

### Volatility of returns, not of prices

The standard deviation of a price series measures how far prices wandered over the sample, which grows with the sample length and the
price level. Risk is measured on returns.

## 3. Worked Examples

### Example 1: units of a daily number

Daily returns of a Swiss equity fund have standard deviation 1.2%. Variance $1.44\,\%^2 = 1.44 \times 10^{-4}$. Annualized (252 days,
00/04): $1.2\% \times \sqrt{252} = 19.0\%$, variance $0.0363$ per year.

### Example 2: a position in francs

CHF 5 million in that fund: daily P&L volatility $5{,}000{,}000 \times 0.012 =$ CHF 60,000. With 2× leverage (CHF 5 million own capital,
CHF 5 million borrowed), the position is CHF 10 million and the daily volatility CHF 120,000, or 2.4% of the capital.

### Example 3: a three-sigma day

On a day with a return of −3.9% and a daily volatility of 1.2% (mean about 0), $z = -3.25$. Under normality such a day (or worse) happens
with probability $\Phi(-3.25) = 0.06\%$, about once every 7 years. In real equity data, days like this occur several times a year in
turbulent periods.

### Example 4: percent or points

The SNB policy rate rises from 0.25% to 0.50%: 25 basis points, or a 100% increase. A bond yield volatility of "80 bp a year" means a
one-standard-deviation annual move of 0.8 percentage points in the yield.

## 4. Finance Application

### Where it matters

- **Factsheets.** Funds report annualized volatility of monthly or daily returns, in percent.
- **Risk budgets.** A desk's limit is stated in CHF volatility or VaR; translating it into positions needs $V\sigma$.
- **Leverage.** Margin lending, structured products and leveraged ETFs scale volatility linearly with leverage.
- **Rates.** Yield volatility is in basis points (normal volatility); option markets also quote lognormal (percentage) volatility.
- **Alerts.** Trading and risk systems flag moves above 2 or 3 sigma; with fat tails, these thresholds trigger more often than normal
  probabilities suggest.

### The usual misreadings

- "Variance 4 means twice the risk of variance 1." Risk in return units is the standard deviation: twice, yes, but variance 4 is four times
  the variance.
- "A 16% volatility means I can lose at most 16%." It is one standard deviation; losses of two or three times that are possible.
- "Rates rose 50%." From 1.0% to 1.5% is 50 basis points; say which one you mean.
- "The fund's standard deviation of NAV is CHF 12." Standard deviation of prices is not a risk measure.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

### Ten years of daily returns

A Swiss equity fund: daily simple returns (decimals) with a mean of 0.03% and a volatility of 1.2%, Student-t with 4 degrees of freedom
for realistic tails. 252 trading days per year.

```python
periods_per_year = 252
n_days = 10 * periods_per_year
nu = 4
z = rng.standard_t(nu, n_days) / np.sqrt(nu / (nu - 2))     # unit variance
r = 0.0003 + 0.012 * z                                        # daily simple returns, decimals
prices = 100 * np.cumprod(1 + r)                              # NAV in CHF

var_d = r.var(ddof=1)
sd_d = r.std(ddof=1)
print(f"daily variance   {var_d:.6f} (decimal^2)  = {var_d * 1e4:.3f} %^2")
print(f"daily volatility {sd_d:.4%}")
print(f"annual volatility {sd_d * np.sqrt(periods_per_year):.2%}, annual variance {var_d * periods_per_year:.4f}")
print(f"standard deviation of NAV prices: CHF {prices.std(ddof=1):.2f}  <- not a risk measure")
```

The returns were built with 1.2% daily volatility, and the estimate is close. The standard deviation of the NAV series is a number in
francs that depends on where prices wandered over the ten years; it says nothing about day-to-day risk.

### From return volatility to CHF volatility, and leverage

```python
capital = 5_000_000          # CHF
for leverage in [0.5, 1.0, 2.0, 3.0]:
    position = leverage * capital
    borrowing_cost = (leverage - 1) * 0.02 / periods_per_year if leverage > 1 else 0.0
    r_lev = leverage * r - borrowing_cost                  # return on own capital
    print(f"leverage {leverage:3.1f}: daily vol on capital {r_lev.std(ddof=1):.3%}, "
          f"daily CHF vol {r_lev.std(ddof=1) * capital:>9,.0f}, mean daily return {r_lev.mean():.4%}")
```

Volatility scales exactly with leverage; the financing cost only shifts the mean. At 3× leverage the daily CHF volatility on
CHF 5 million of capital is about CHF 180,000.

### z-scores and how often big days happen

```python
zs = (r - r.mean()) / r.std(ddof=1)
rows = []
for k in [1, 2, 3, 4, 5]:
    observed = np.mean(np.abs(zs) > k)
    normal = 2 * stats.norm.sf(k)
    rows.append({"threshold |z| >": k, "observed share": observed, "normal share": normal,
                 "observed days": int(np.sum(np.abs(zs) > k)), "normal expectation (days)": normal * n_days})
pd.DataFrame(rows).set_index("threshold |z| >")
```

Moves beyond one sigma are actually less frequent than under the normal distribution (the t-distribution is more peaked), but moves
beyond three and four sigma are many times more frequent. A "four-sigma day" should happen about once in 60 years under normality; here
it happened several times in ten.

### Bessel's correction

```python
rows = []
for n in [3, 5, 12, 36, 120]:
    samples = rng.normal(0, 0.04, size=(20_000, n))    # monthly returns, true vol 4%
    v0 = samples.var(axis=1, ddof=0).mean()
    v1 = samples.var(axis=1, ddof=1).mean()
    s1 = samples.std(axis=1, ddof=1).mean()
    rows.append({"n": n, "mean var (ddof=0) / true": v0 / 0.04**2, "mean var (ddof=1) / true": v1 / 0.04**2,
                 "mean sd (ddof=1) / true": s1 / 0.04})
pd.DataFrame(rows).set_index("n")
```

Dividing by $n$ underestimates the variance by the factor $(n-1)/n$; dividing by $n-1$ removes the bias for the variance. The standard
deviation remains slightly biased downwards (the square root is concave, Jensen's inequality), noticeably only for very small samples.
For ten years of monthly data ($n = 120$) the choice hardly matters.

### Other dispersion measures

```python
mad = np.mean(np.abs(r - r.mean()))
downside = np.sqrt(np.mean(np.minimum(r, 0) ** 2))
upside = np.sqrt(np.mean(np.maximum(r, 0) ** 2))
print(f"standard deviation       {sd_d:.4%}")
print(f"mean absolute deviation  {mad:.4%}  (ratio {mad / sd_d:.3f}; normal: {np.sqrt(2 / np.pi):.3f})")
print(f"downside deviation (0)   {downside:.4%}")
print(f"upside deviation (0)     {upside:.4%}")
```

The ratio of mean absolute deviation to standard deviation is below the normal value 0.80: fat tails inflate the standard deviation,
which squares large deviations, more than the absolute deviation. Downside and upside deviations (measured from zero) are close, the upside one
a little larger because the mean return is positive; they differ much more when returns are skewed (02/03).

## 6. Visualizations

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(r * 100, bins=120, density=True, alpha=0.6, color="C0", label="daily returns")
x = np.linspace(r.min(), r.max(), 400)
ax.plot(x * 100, stats.norm.pdf(x, r.mean(), sd_d) / 100, "k", lw=1.5, label="normal, same mean and sd")
for k, shade in [(1, 0.25), (2, 0.12)]:
    ax.axvspan((r.mean() - k * sd_d) * 100, (r.mean() + k * sd_d) * 100, color="C1", alpha=shade)
inside1 = np.mean(np.abs(zs) <= 1)
inside2 = np.mean(np.abs(zs) <= 2)
ax.set_title(f"Daily returns: {inside1:.0%} within ±1 sd (normal 68%), {inside2:.1%} within ±2 sd (normal 95.4%)")
ax.set_xlabel("daily return (%)")
ax.set_ylabel("density (per percentage point)")
ax.set_xlim(-8, 8)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Within one standard deviation lie 76% of the days instead of the normal 68%: the histogram is taller and narrower in the middle.
Within two standard deviations the share is almost exactly the normal 95%. The rest of the difference sits in the far tails, where the
histogram has visible mass beyond ±4% and the normal curve has almost none.

```python
fig, ax = plt.subplots(figsize=(11, 4.5))
days = np.arange(n_days) / periods_per_year
ax.plot(days, zs, lw=0.6)
for k in [3, -3]:
    ax.axhline(k, color="C3", ls="--", lw=1)
big = np.abs(zs) > 3
ax.plot(days[big], zs[big], "o", color="C3", ms=4, label=f"|z| > 3: {big.sum()} days (normal: {2 * stats.norm.sf(3) * n_days:.0f})")
ax.set_title("Daily returns in units of their volatility")
ax.set_xlabel("years")
ax.set_ylabel("z-score")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Three-sigma days are spread through the sample at a rate several times the normal one. (These returns are independent; real returns also
cluster their big days in turbulent periods, module 09.)

```python
lev = np.linspace(0, 4, 100)
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(lev, lev * sd_d * np.sqrt(periods_per_year) * 100, label="annual volatility on capital (%)")
ax.plot(lev, lev * sd_d * capital / 1e3, label="daily volatility (CHF thousand, on CHF 5 m capital)")
ax.set_title("Volatility scales linearly with leverage")
ax.set_xlabel("leverage (position / capital)")
ax.set_ylabel("volatility")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Both lines are straight: $\operatorname{sd}(L\,r) = L\operatorname{sd}(r)$. Variance, by contrast, would grow with the square of the leverage.

## 8. Solutions / Checks

```python
# 1.
m = np.array([0.4, -0.2, 0.6, 0.1, -0.5, 0.3, 0.2, -0.1, 0.4, 0.0, -0.3, 0.5])     # percent
print(f"1. variance {m.var(ddof=1):.4f} %^2, sd {m.std(ddof=1):.4f} %, annualized {m.std(ddof=1) * np.sqrt(12):.4f} %")

# 2.
print(f"2. daily CHF volatility {20e6 * 0.12 / np.sqrt(252):,.0f}")

# 3.
print(f"3. implied daily vol {2.5 / 4:.3f} %, annual {2.5 / 4 * np.sqrt(252):.1f} %; "
      f"normal probability {stats.norm.cdf(-4):.2e}, t(4) probability {stats.t.cdf(-4 * np.sqrt(2), 4):.2e}")
```

**3.** "Four sigma" for a 2.5% fall implies a daily volatility of about 0.63%, i.e. 9.9% a year: a calm-market estimate. Under normality a
fall of four sigma or more has probability $3 \times 10^{-5}$ (once in about 125 years of trading days). With fat tails (a t-distribution with
4 degrees of freedom scaled to the same volatility) the probability is about 75 times higher, and volatility itself changes over time:
measured against a volatility estimated in a turbulent period, the same move might be a two-sigma day.

```python
# 4.
r8 = r / r.std(ddof=1) * 0.08 / np.sqrt(periods_per_year)
r3 = 3 * r8
print(f"4. daily vol: fund {r8.std(ddof=1):.4%}, 3x product {r3.std(ddof=1):.4%}")
cum_fund = np.prod(1 + r8) - 1
cum_3x = np.prod(1 + r3) - 1
print(f"   ten-year cumulative return: fund {cum_fund:.2%}, 3x product {cum_3x:.2%}, 3 x fund {3 * cum_fund:.2%}")

# 5.
for name, sample in [("normal", rng.normal(size=1_000_000))] + [(f"t({k})", rng.standard_t(k, 1_000_000)) for k in [3, 5, 30]]:
    print(f"5. {name:7s}: MAD / sd = {np.mean(np.abs(sample - sample.mean())) / sample.std():.3f}")

# 7.
from math import gamma
n7 = 5
print(f"7. E[s] / sigma for n = 5: {np.sqrt(2 / (n7 - 1)) * gamma(n7 / 2) / gamma((n7 - 1) / 2):.4f}")
```

**4.** Day by day the product has exactly 3 times the volatility. Over ten years, the leveraged product compounds daily: its cumulative return
(64%) is not 3 times the fund's (3 × 25.7% = 77%), because volatility drag grows with the square of the leverage (00/01). Over long horizons, leveraged products
neither deliver 3 times the return nor exactly 3 times the volatility of the cumulative return.

**5.** The normal ratio is 0.798; fatter tails (fewer degrees of freedom) give lower ratios, approaching the normal value as the degrees of
freedom grow.

**6.** Write $\sum_i (X_i - \bar X)^2 = \sum_i (X_i - \mu)^2 - n(\bar X - \mu)^2$. The expectations are $n\sigma^2$ and $n \cdot \sigma^2/n = \sigma^2$, so the
total is $(n-1)\sigma^2$.

**7.** $s = \sqrt{s^2}$ and the square root is strictly concave, so $E[s] < \sqrt{E[s^2]} = \sigma$ unless $s$ is constant. For $n = 5$ the
factor is 0.940: a 6% downward bias.

## 9. Conclusion

- Volatility is the standard deviation of returns; variance is in squared units and is used for algebra, not interpretation.
- $\operatorname{sd}(aX + b) = |a|\operatorname{sd}(X)$: CHF volatility is position times return volatility; leverage scales volatility linearly.
- Percent, percentage points and basis points are different units; say which one you mean.
- z-scores measure moves in units of volatility; with fat tails, "three-sigma days" are far more common than normal theory says.
- Divide by $n - 1$ for an unbiased variance; never compute the volatility of prices.
