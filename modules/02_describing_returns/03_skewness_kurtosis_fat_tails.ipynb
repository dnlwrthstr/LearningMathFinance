# Skewness, Kurtosis, Fat Tails

Mean and volatility describe the centre and the width of a return distribution. They say nothing about its shape: whether losses are
larger than gains (skewness) or whether extreme days are far more common than a normal distribution allows (kurtosis, fat tails). Two
strategies with the same mean and volatility can have very different crash risk. This notebook defines the third and fourth standardized
moments, shows where skewness and fat tails come from in markets, how they fade when returns are aggregated over longer horizons, and
how unreliable their sample estimates are.

## 1. Introduction

Two Swiss hedge funds both report 6% a year at 8% volatility. One buys equities and hedges with options; the other sells out-of-the-money
put options and collects the premium. The second one wins small most months and loses big once in a while: negative skewness and fat
tails. Its volatility understates its risk, and its Sharpe ratio flatters it.

After this notebook you can:

- compute skewness and excess kurtosis and interpret their signs and sizes;
- name market mechanisms that create negative skewness and fat tails (crashes, volatility regimes, option selling);
- show that a mixture of calm and turbulent periods creates kurtosis even when each period is normal;
- explain why skewness and kurtosis shrink with the horizon, and at what rate;
- judge how many observations are needed to estimate them, and why one day can dominate the estimate.

## 2. Theory

### Standardized moments

For a random return $X$ with mean $\mu$ and standard deviation $\sigma$:

$$
\text{skew}(X) = E\Big[\Big(\frac{X - \mu}{\sigma}\Big)^3\Big], \qquad
\text{kurt}(X) = E\Big[\Big(\frac{X - \mu}{\sigma}\Big)^4\Big], \qquad \text{excess kurtosis} = \text{kurt} - 3 .
$$

Both are unit-free. For the normal distribution skewness is 0 and kurtosis 3, so excess kurtosis is 0.

- **Negative skewness**: the left tail is longer; large losses are more likely than large gains of the same size.
- **Positive excess kurtosis** (leptokurtic, "fat-tailed"): more mass in the centre and in the tails, less in the shoulders, compared with
  a normal distribution of the same variance.

### Examples

| distribution | skewness | excess kurtosis |
|---|---|---|
| normal | 0 | 0 |
| Student-$t$ with $\nu$ degrees of freedom | 0 | $6/(\nu - 4)$ for $\nu > 4$ |
| lognormal with log-sd $s$ | $(e^{s^2} + 2)\sqrt{e^{s^2} - 1} > 0$ | large |
| typical daily equity index returns | about $-0.5$ | 5 to 10 |
| typical monthly equity index returns | about $-0.5$ | 1 to 2 |

### Volatility regimes create fat tails

Suppose each day is calm with probability $1 - p$ (volatility $\sigma_c$) or turbulent with probability $p$ ($\sigma_t$), and returns are
normal given the regime with mean 0. Then $E[X^2] = (1-p)\sigma_c^2 + p\sigma_t^2$ and $E[X^4] = 3\big((1-p)\sigma_c^4 + p\sigma_t^4\big)$, so

$$
\text{kurt} = 3\,\frac{E[\sigma^4]}{\big(E[\sigma^2]\big)^2} \ge 3,
$$

by Jensen's inequality, with equality only if the volatility never changes. Changing volatility alone makes returns fat-tailed.

### Aggregation over time

For a sum of $n$ independent, identically distributed returns, the third and fourth cumulants add up:

$$
\text{skew}\Big(\sum_{i=1}^n X_i\Big) = \frac{\text{skew}(X)}{\sqrt n}, \qquad \text{excess kurt}\Big(\sum_{i=1}^n X_i\Big) = \frac{\text{excess kurt}(X)}{n} .
$$

This is the central limit theorem at work: monthly returns are closer to normal than daily ones. Real returns aggregate more slowly
because volatility is persistent (module 09).

### Estimation

The sample versions replace expectations by averages. Under normality their standard errors are about

$$
\text{SE}(\widehat{\text{skew}}) \approx \sqrt{6/n}, \qquad \text{SE}(\widehat{\text{excess kurt}}) \approx \sqrt{24/n} .
$$

With fat tails they are much larger: the fourth power makes a single extreme day dominate the estimate.

## 3. Worked Examples

### Example 1: a three-point distribution

A strategy returns +1% with probability 0.9 and −9% with probability 0.1. Mean 0, variance $0.9 \times 1 + 0.1 \times 81 = 9\,\%^2$, so
$\sigma = 3\%$. Third moment $0.9 \times 1 + 0.1 \times (-729) = -72$, skewness $-72/27 = -2.67$. Fourth moment
$0.9 + 0.1 \times 6561 = 657$, kurtosis $657/81 = 8.1$, excess 5.1.

### Example 2: two regimes

Calm days (90%) with volatility 0.8%, turbulent days (10%) with 2.5%. $E[\sigma^2] = 0.9 \times 0.64 + 0.1 \times 6.25 = 1.201$,
$E[\sigma^4] = 0.9 \times 0.4096 + 0.1 \times 39.06 = 4.275$. Kurtosis $3 \times 4.275/1.201^2 = 8.89$, excess 5.9: realistic for daily equity returns,
with each regime perfectly normal.

### Example 3: aggregation

Daily excess kurtosis 6 and skewness −0.5 for independent days. Over 21 days (a month): excess kurtosis $6/21 = 0.29$, skewness
$-0.5/\sqrt{21} = -0.11$. Real monthly equity data show more than this, because turbulent days come in clusters.

### Example 4: how much data

Ten years of monthly data ($n = 120$): even under normality, the standard error of the skewness is $\sqrt{6/120} = 0.22$ and of the excess
kurtosis $\sqrt{24/120} = 0.45$. A sample skewness of −0.3 is not distinguishable from zero.

## 4. Finance Application

### Where it matters

- **Risk management.** VaR and expected shortfall from a normal distribution underestimate tail losses when kurtosis is high (module 05);
  the Cornish–Fisher expansion adjusts quantiles with skewness and kurtosis.
- **Hedge-fund due diligence.** Option-selling, merger-arbitrage and credit strategies have negative skewness; their Sharpe ratios
  overstate risk-adjusted performance.
- **Product design.** Structured products with capital protection are long a put (positive skew); reverse convertibles are short a put
  (negative skew).
- **Pricing.** Negative skewness of equity returns shows up as the volatility skew in option prices (10/05).

### The usual misreadings

- "Zero skewness and normal volatility, so the risk is normal." Kurtosis can still be large.
- "Monthly data look normal, so tails don't matter." Aggregation hides daily tails, but the crash still happened within the month.
- "The sample kurtosis is 4.2." With a few hundred observations, a single day can change this number by several units.
- "Positive skewness is always good." Lottery-like assets with positive skewness have historically had low average returns: investors pay for it.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
pd.set_option("display.float_format", lambda x: f"{x:,.3f}")

def moments(x):
    x = np.asarray(x)
    z = (x - x.mean()) / x.std()
    return {"mean": x.mean(), "sd": x.std(ddof=1), "skewness": np.mean(z**3), "excess kurtosis": np.mean(z**4) - 3}
```

### Example 1 and Example 2, checked

```python
values, probs = np.array([0.01, -0.09]), np.array([0.9, 0.1])
mu = probs @ values
sd = np.sqrt(probs @ (values - mu) ** 2)
print(f"Example 1: skewness {probs @ ((values - mu) / sd) ** 3:.3f}, excess kurtosis {probs @ ((values - mu) / sd) ** 4 - 3:.3f}")

p_t, s_c, s_t = 0.1, 0.008, 0.025
e2 = (1 - p_t) * s_c**2 + p_t * s_t**2
e4 = (1 - p_t) * s_c**4 + p_t * s_t**4
print(f"Example 2: kurtosis {3 * e4 / e2**2:.3f}, excess {3 * e4 / e2**2 - 3:.3f}")
```

### Simulated daily equity returns with regimes and crashes

Forty years of daily returns (252 days a year). Volatility switches between a calm (0.8%) and a turbulent (2.5%) regime with persistence
(regimes last weeks, not days); on about one day in 500 a crash adds a jump of −5% on average. Returns are decimals.

```python
periods_per_year = 252
n_days = 40 * periods_per_year
stay_calm, stay_turbulent = 0.995, 0.96          # daily probabilities of staying in the regime
regime = np.empty(n_days, dtype=int)
regime[0] = 0
u = rng.uniform(size=n_days)
for t in range(1, n_days):
    if regime[t - 1] == 0:
        regime[t] = 0 if u[t] < stay_calm else 1
    else:
        regime[t] = 1 if u[t] < stay_turbulent else 0
vol = np.where(regime == 0, 0.008, 0.025)
crash = rng.uniform(size=n_days) < 1 / 500
jumps = np.where(crash, rng.normal(-0.05, 0.02, n_days), 0.0)
daily = 0.0004 + vol * rng.standard_normal(n_days) + jumps

print(f"share of turbulent days: {regime.mean():.1%}, crash days: {crash.sum()}")
pd.DataFrame({"daily": moments(daily)}).T
```

Each day is normal given its regime, yet the daily returns have an excess kurtosis of about 6 (from the regimes and the crashes) and a
negative skewness of about −0.2 (from the 22 crash days). Kurtosis is the much larger effect; that is typical of real index data too.

### Aggregation: weekly, monthly, quarterly, annual

```python
def aggregate(x, h):
    n = len(x) // h * h
    return x[:n].reshape(-1, h).sum(axis=1)       # sums of daily returns (close to log-return aggregation)

shuffled = rng.permutation(daily)                  # same days, clustering destroyed
rows = []
for name, h in [("daily", 1), ("weekly", 5), ("monthly", 21), ("quarterly", 63), ("annual", 252)]:
    m_real, m_iid = moments(aggregate(daily, h)), moments(aggregate(shuffled, h))
    rows.append({"horizon": name, "n": len(daily) // h,
                 "skew": m_real["skewness"], "skew if iid": m_iid["skewness"],
                 "skew theory iid": moments(daily)["skewness"] / np.sqrt(h),
                 "ex. kurt": m_real["excess kurtosis"], "ex. kurt if iid": m_iid["excess kurtosis"],
                 "ex. kurt theory iid": moments(daily)["excess kurtosis"] / h})
pd.DataFrame(rows).set_index("horizon")
```

Shuffling the days keeps the same daily distribution but destroys the regime persistence; then the excess kurtosis follows the $1/h$
rule closely at weekly and monthly horizons (4.46 against 1.25 weekly, 2.94 against 0.29 monthly for the real and shuffled series). With
persistent regimes, monthly returns keep ten times more excess kurtosis: a turbulent month is a month of turbulent days. The skewness
estimates are small and noisy at every horizon beyond daily; with 480 monthly observations their standard error is about
$\sqrt{6/480} = 0.11$, and the quarterly and annual estimates rest on 160 and 40 observations.

### A put-selling strategy

```python
def bs_put(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + sigma**2 / 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    return K * np.exp(-r * T) * stats.norm.cdf(-d2) - S * stats.norm.cdf(-d1)

monthly_index = aggregate(daily, 21)                   # monthly log-like index returns
K_rel, T, rf, implied = 0.95, 1 / 12, 0.005, 0.20       # 5% out-of-the-money, one month
premium = bs_put(1.0, K_rel, T, rf, implied)
payoff = np.maximum(K_rel - np.exp(monthly_index), 0)
put_seller = premium - payoff                           # per CHF 1 of notional, collateral earns rf (ignored)

print(f"premium per CHF 1 notional: {premium:.4%} per month")
pd.DataFrame({"index (monthly)": moments(monthly_index), "put seller (monthly)": moments(put_seller)}).T
```

```python
print(f"months with a gain: {np.mean(put_seller > 0):.1%}")
print(f"worst month: {put_seller.min():.2%}, best month: {put_seller.max():.2%}")
print(f"annualized Sharpe (rf ignored): index {monthly_index.mean() / monthly_index.std() * np.sqrt(12):.2f}, "
      f"put seller {put_seller.mean() / put_seller.std() * np.sqrt(12):.2f}")
```

The put seller gains a small, fixed amount in most months and occasionally loses many times that amount. Its skewness is strongly
negative and its kurtosis very large. Its Sharpe ratio can look attractive because the volatility, which squares deviations equally on
both sides, does not see the asymmetry.

### How reliable are the sample estimates?

```python
worst = np.argmin(daily)
without = np.delete(daily, worst)
m_all, m_wo = moments(daily), moments(without)
print(f"all {n_days} days:          skewness {m_all['skewness']:.3f}, excess kurtosis {m_all['excess kurtosis']:.2f}")
print(f"without the worst day:   skewness {m_wo['skewness']:.3f}, excess kurtosis {m_wo['excess kurtosis']:.2f}")

# ten-year windows
windows = daily.reshape(4, -1)
for k, w in enumerate(windows):
    m = moments(w)
    print(f"decade {k + 1}: skewness {m['skewness']:6.2f}, excess kurtosis {m['excess kurtosis']:6.2f}")
```

Removing a single day out of more than 10,000 changes the skewness and kurtosis noticeably, and the four decades of the same process
give quite different estimates. Fat-tailed data make higher moments hard to pin down: they are driven by the few days that matter most.

## 6. Visualizations

```python
x = np.linspace(-8, 8, 800)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
dists = {
    "normal": stats.norm.pdf(x),
    "t(5), unit variance": stats.t.pdf(x * np.sqrt(5 / 3), 5) * np.sqrt(5 / 3),
    "two-regime mixture": (0.9 * stats.norm.pdf(x, 0, 0.8 / np.sqrt(1.201)) + 0.1 * stats.norm.pdf(x, 0, 2.5 / np.sqrt(1.201))),
}
for name, pdf in dists.items():
    axes[0].plot(x, pdf, label=name)
    axes[1].semilogy(x, pdf, label=name)
axes[0].set_xlim(-4, 4)
axes[0].set_title("Densities with variance 1")
axes[1].set_ylim(1e-7, 1)
axes[1].set_title("Same densities, log scale: the tails")
for ax in axes:
    ax.set_xlabel("return in standard deviations")
    ax.set_ylabel("density")
    ax.legend()
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

All three densities have variance 1. On a linear scale the fat-tailed ones are taller and narrower in the middle. On a log scale the
difference is in the tails: the normal density falls like a parabola (it is $e^{-x^2/2}$), the others much more slowly. At five standard
deviations they are thousands of times more likely than the normal says.

```python
hs = np.array([1, 2, 5, 10, 21, 42, 63, 126])
k_real = [moments(aggregate(daily, h))["excess kurtosis"] for h in hs]
k_iid = [moments(aggregate(shuffled, h))["excess kurtosis"] for h in hs]
fig, ax = plt.subplots(figsize=(9, 5))
ax.semilogx(hs, k_real, "o-", label="with volatility regimes (persistent)")
ax.semilogx(hs, k_iid, "s-", label="same days shuffled (iid)")
ax.semilogx(hs, moments(daily)["excess kurtosis"] / hs, "k--", label="1/h rule")
ax.axhline(0, color="grey", lw=0.8)
ax.set_title("Excess kurtosis of h-day returns")
ax.set_xlabel("horizon h (trading days, log scale)")
ax.set_ylabel("excess kurtosis")
ax.legend()
ax.grid(True, alpha=0.3, which="both")
plt.show()
```

For independent days the excess kurtosis follows the $1/h$ line until sampling noise takes over (beyond about a month, where fewer than
500 observations remain). With persistent volatility regimes it falls much more slowly, still near 3 at a month: fat tails survive
aggregation when volatility clusters. The last points, with 160 and 80 observations, are dominated by noise.

```python
fig, ax = plt.subplots(figsize=(10, 5))
bins = np.linspace(put_seller.min(), put_seller.max(), 80)
ax.hist(put_seller * 100, bins=bins * 100, alpha=0.7, label="put seller, monthly")
q_normal = put_seller.mean() + stats.norm.ppf(0.01) * put_seller.std()
ax.axvline(q_normal * 100, color="k", ls="--", label=f"normal 1% quantile, same mean and sd ({q_normal:.1%})")
ax.axvline(np.quantile(put_seller, 0.01) * 100, color="C3", label=f"actual 1% quantile ({np.quantile(put_seller, 0.01):.1%})")
ax.set_yscale("log")
ax.set_title("Monthly returns of selling 5% out-of-the-money puts")
ax.set_xlabel("monthly return per CHF 1 notional (%)")
ax.set_ylabel("number of months (log scale)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Almost all months sit in one bar at the premium (note the log scale); a long tail of losses extends far to the left. A normal
distribution with the same mean and volatility would put the worst month in a hundred at the dashed line; the actual 1% quantile, −9.6%,
is almost three times as far out.

## 8. Solutions / Checks

```python
# 1.
print("1.", {k: round(v, 3) for k, v in moments(np.array([2, 1, 3, 2, -12, 2, 1, 3, 2, 1, 2, 3]) / 100).items()})

# 2.
print(f"2. t(6): {6 / (6 - 4):.1f}, t(4.5): {6 / (4.5 - 4):.1f}")

# 4. solve 3 E[s^4] / E[s^2]^2 = 13 for s_t
from scipy.optimize import brentq
def excess_kurt(s_turb, p=0.05, s_calm=0.007):
    e2 = (1 - p) * s_calm**2 + p * s_turb**2
    e4 = (1 - p) * s_calm**4 + p * s_turb**4
    return 3 * e4 / e2**2 - 3
s_star = brentq(lambda s: excess_kurt(s) - 10, 0.007, 0.2)
print(f"4. turbulent volatility {s_star:.4%} (ratio to calm {s_star / 0.007:.2f})")

# 5.
n5, reps = 120, 2000
k_norm = [moments(rng.normal(size=n5))["excess kurtosis"] for _ in range(reps)]
k_t4 = [moments(rng.standard_t(4, n5) / np.sqrt(2))["excess kurtosis"] for _ in range(reps)]
print(f"5. sd of sample excess kurtosis: normal {np.std(k_norm):.2f}, t(4) {np.std(k_t4):.2f}, formula {np.sqrt(24 / n5):.2f}")
print(f"   mean of sample excess kurtosis: normal {np.mean(k_norm):.2f}, t(4) {np.mean(k_t4):.2f} (true value: infinite)")
```

**3.** The high Sharpe ratio comes with the return profile of an option seller: steady gains, rare large losses. Over five years (60 months) a
skewness of −2.1 and excess kurtosis of 9 can rest on one or two bad months; the volatility in the Sharpe ratio counts those months
the same as good ones and understates crash risk. Ask what the fund lost in the worst month, in 2008, in March 2020; compare drawdowns and
expected shortfall rather than Sharpe ratios.

**5.** Under normality the spread matches the formula (slightly smaller because the sample kurtosis is bounded for small $n$). For $t(4)$, whose
true excess kurtosis is infinite, the sample values are far more dispersed and on average far below the truth: no finite sample can reveal an
infinite fourth moment.

**6.** For independent $X, Y$ with means 0, $E[(X + Y)^3] = E[X^3] + 3E[X^2]E[Y] + 3E[X]E[Y^2] + E[Y^3] = E[X^3] + E[Y^3]$. For $n$ i.i.d. terms the third
central moment is $n\mu_3$ and the standard deviation $\sqrt n \sigma$, so the skewness is $n\mu_3/(n^{3/2}\sigma^3) = \text{skew}/\sqrt n$.

**7.** Given $\sigma$, $E[X^2 \mid \sigma] = \sigma^2$ and $E[X^4 \mid \sigma] = 3\sigma^4$. By the law of total expectation $E[X^2] = E[\sigma^2]$ and
$E[X^4] = 3E[\sigma^4]$. Jensen's inequality for the convex function $u \mapsto u^2$ gives $E[\sigma^4] \ge (E[\sigma^2])^2$.

## 9. Conclusion

- Skewness and kurtosis are the standardized third and fourth moments; for the normal they are 0 and 3.
- Daily equity returns are negatively skewed and fat-tailed; crashes create skewness, changing volatility creates kurtosis.
- For independent returns, skewness falls like $1/\sqrt h$ and excess kurtosis like $1/h$ with the horizon; persistent volatility slows this.
- Option-selling strategies have strong negative skewness that volatility and the Sharpe ratio do not show.
- Sample estimates are noisy and can hinge on a single day; with fat tails, the true kurtosis may not even be finite.
