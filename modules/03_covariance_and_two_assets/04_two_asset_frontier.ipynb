# The Two-Asset Frontier

Mix two assets in every proportion and plot risk against return. The curve you get, and how
correlation bends it, is the whole of portfolio theory in miniature.

## 1. Introduction

01 showed that a 50/50 mix of two assets can be less risky than either. This notebook asks the
full question: for **every** weight $w$ in asset A (and $1-w$ in asset B), what are the
portfolio's expected return and volatility?

Plotted in the (volatility, return) plane, the answer is a curve, the **two-asset frontier**.
After this notebook you can:

- compute any point on it, and the minimum-variance portfolio;
- explain why the curve bends left, and why correlation controls how far;
- tell the **efficient** part of the curve from the part no rational investor should hold.

Module 04 extends the same ideas to many assets with matrices.

## 2. Theory

### Return and variance as functions of the weight

Weights $w$ in A and $1-w$ in B (fully invested; $w < 0$ or $w > 1$ means a short position).

$$
\mu_p(w) = w\,\mu_A + (1-w)\,\mu_B
$$

$$
\sigma_p^2(w) = w^2\sigma_A^2 + (1-w)^2\sigma_B^2 + 2w(1-w)\,\rho\,\sigma_A\sigma_B
$$

Return is **linear** in $w$, variance is **quadratic**. That mismatch is why the frontier curves.

### The two extreme correlations

**$\rho = 1$**: the variance is a perfect square,

$$
\sigma_p^2 = \big(w\sigma_A + (1-w)\sigma_B\big)^2
\quad\Longrightarrow\quad
\sigma_p = w\sigma_A + (1-w)\sigma_B \quad (0 \le w \le 1)
$$

Volatility is linear in $w$ as well: the frontier is the **straight line** between the two
assets. No diversification.

**$\rho = -1$**: $\sigma_p = |w\sigma_A - (1-w)\sigma_B|$, which is zero at

$$
w_0 = \frac{\sigma_B}{\sigma_A + \sigma_B}
$$

Two straight lines meeting on the return axis: a riskless portfolio built from two risky assets.

For $-1 < \rho < 1$ the curve lies **between** these: always to the left of the straight line,
because $\sigma_p \le w\sigma_A + (1-w)\sigma_B$ with equality only at $\rho = 1$.

### The minimum-variance portfolio

Set the derivative of $\sigma_p^2(w)$ to zero:

$$
\frac{d\sigma_p^2}{dw} = 2w\sigma_A^2 - 2(1-w)\sigma_B^2 + 2(1-2w)\rho\sigma_A\sigma_B = 0
$$

$$
w_{\text{mv}} = \frac{\sigma_B^2 - \rho\,\sigma_A\sigma_B}{\sigma_A^2 + \sigma_B^2 - 2\rho\,\sigma_A\sigma_B}
$$

The denominator is $\operatorname{Var}(A - B) > 0$ (unless $A$ and $B$ are perfectly
correlated with equal volatility), so this is a minimum.

- $w_{\text{mv}} > 0$ exactly when $\rho < \sigma_B/\sigma_A$ (B the lower-volatility asset).
  Above that correlation, the least risky portfolio **shorts** the riskier asset.
- Compare 03: the hedge ratio minimized $\operatorname{Var}(S - hF)$ over $h$; here we minimize
  over $w$ with the weights tied to sum to 1. Same calculus.

### The shape: a hyperbola

Solve $\mu_p$ for $w$, $w = (\mu_p - \mu_B)/(\mu_A - \mu_B)$, and substitute: $\sigma_p^2$ becomes
a quadratic in $\mu_p$. A curve with $\sigma^2 = a\mu^2 + b\mu + c$ is a hyperbola in the
$(\sigma, \mu)$ plane, with its leftmost point at the minimum-variance portfolio.

### Efficient and inefficient

Each volatility above $\sigma_{\text{mv}}$ is reached by two portfolios: one on the upper branch,
one on the lower branch with **lower** return. The upper branch (from the minimum-variance
point upward) is the **efficient frontier**; nobody should hold a portfolio on the lower branch.

## 3. Worked Examples

Annual parameters, CHF investor, returns in %:

| | expected return $\mu$ | volatility $\sigma$ |
|---|---|---|
| A: Swiss equities | 7% | 16% |
| B: Swiss bonds | 2% | 5% |

Correlation $\rho = 0.1$.

### Example 1: the 60/40 portfolio

$$
\mu_p = 0.6 \cdot 7 + 0.4 \cdot 2 = 5.0\%
$$

$$
\sigma_p^2 = 0.36 \cdot 0.0256 + 0.16 \cdot 0.0025 + 2 \cdot 0.6 \cdot 0.4 \cdot 0.1 \cdot 0.16 \cdot 0.05
= 0.009216 + 0.0004 + 0.000384 = 0.0100
$$

$\sigma_p = 10.0\%$. The straight line ($\rho = 1$) would give $0.6 \cdot 16 + 0.4 \cdot 5 = 11.6\%$.

### Example 2: the minimum-variance portfolio

$$
w_{\text{mv}} = \frac{0.0025 - 0.1 \cdot 0.16 \cdot 0.05}{0.0256 + 0.0025 - 2 \cdot 0.1 \cdot 0.16 \cdot 0.05}
= \frac{0.0017}{0.0265} = 6.4\% \text{ in equities}
$$

This portfolio has $\mu = 2.3\%$ and $\sigma = 4.9\%$: **more** return and **less** risk than
100% bonds. A pure bond portfolio sits on the inefficient branch.

### Example 3: perfect negative correlation

If $\rho$ were $-1$: $w_0 = 5/(16+5) = 23.8\%$ in equities gives zero volatility and a return of
$0.238 \cdot 7 + 0.762 \cdot 2 = 3.19\%$. A riskless 3.19% while the risk-free rate is lower
would be an arbitrage, which is one reason $\rho = -1$ between risky assets does not occur
in practice.

## 4. Finance Application

### The stock–bond allocation

The two-asset frontier is the textbook picture of the most important decision in a balanced
portfolio: how much in equities, how much in bonds. Every point between 0/100 and 100/0 is a
standard strategy (conservative, 40/60, balanced 60/40, growth).

### "All bonds" is not the safest portfolio

With low stock–bond correlation, adding a small equity share **lowers** risk and raises
return (Example 2). The safest portfolio is the minimum-variance one, not the lowest-volatility
asset alone.

### Correlation regimes move the frontier

From roughly 2000 to 2020 the stock–bond correlation was mostly negative in major markets,
which made balanced portfolios unusually efficient. In 2022 both fell together with inflation
and the correlation turned positive. A positive $\rho$ straightens the frontier: the same 60/40
mix carries more risk (the Visualizations show how much).

### The frontier is only as good as its inputs

The curve needs $\mu_A$, $\mu_B$, $\sigma_A$, $\sigma_B$, $\rho$. Volatilities and correlation can
be estimated reasonably well; expected returns cannot (module 06). The sample frontier in the
Python section shows how much it moves from one 10-year history to the next.

### The usual misreading

"The efficient frontier tells me which portfolio to hold." It only tells you which portfolios
**not** to hold (the lower branch). Choosing a point on the upper branch needs a risk
preference, or a risk-free asset (the tangency portfolio and Sharpe ratio in module 04).

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp

rng = np.random.default_rng(42)
periods_per_year = 12   # monthly data in the simulation below
```

### The minimum-variance weight with SymPy

```python
w_sym, rho_sym = sp.symbols("w rho", real=True)
sa_sym, sb_sym = sp.symbols("sigma_A sigma_B", positive=True)

var_p = w_sym**2 * sa_sym**2 + (1 - w_sym)**2 * sb_sym**2 + 2 * w_sym * (1 - w_sym) * rho_sym * sa_sym * sb_sym
w_mv_sym = sp.solve(sp.diff(var_p, w_sym), w_sym)[0]

print("w_mv           =", sp.simplify(w_mv_sym))
print("d2/dw2 var_p   =", sp.factor(sp.diff(var_p, w_sym, 2)), "  (= 2 Var(A - B))")
print("rho = 1:  var_p =", sp.factor(var_p.subs(rho_sym, 1)))
```

### Portfolio statistics for any weight

```python
def portfolio_stats(w, mu_a, mu_b, sigma_a, sigma_b, rho):
    # Expected return and volatility of w in A, (1 - w) in B. Works with arrays of w.
    w = np.asarray(w, dtype=float)
    mu_p = w * mu_a + (1 - w) * mu_b
    var_p = w**2 * sigma_a**2 + (1 - w)**2 * sigma_b**2 + 2 * w * (1 - w) * rho * sigma_a * sigma_b
    return mu_p, np.sqrt(var_p)


def min_variance_weight(sigma_a, sigma_b, rho):
    return (sigma_b**2 - rho * sigma_a * sigma_b) / (sigma_a**2 + sigma_b**2 - 2 * rho * sigma_a * sigma_b)


mu_a, sigma_a = 0.07, 0.16   # Swiss equities, annual
mu_b, sigma_b = 0.02, 0.05   # Swiss bonds, annual
rho = 0.1

rows = []
for label, w in [("100% bonds", 0.0), ("min variance", min_variance_weight(sigma_a, sigma_b, rho)),
                 ("40/60", 0.4), ("60/40", 0.6), ("100% equities", 1.0)]:
    mu_p, sigma_p = portfolio_stats(w, mu_a, mu_b, sigma_a, sigma_b, rho)
    rows.append({"portfolio": label, "equity weight": w, "expected return": float(mu_p), "volatility": float(sigma_p)})

table = pd.DataFrame(rows).set_index("portfolio")
table.style.format("{:.2%}")
```

### Which correlation forces a short position?

$w_{\text{mv}} > 0$ requires $\rho < \sigma_B/\sigma_A = 5/16 = 0.3125$.

```python
for corr in [-0.5, 0.0, 0.3, 0.3125, 0.5, 0.9]:
    print(f"rho = {corr:+.4f}:  w_mv (equities) = {min_variance_weight(sigma_a, sigma_b, corr):+.2%}")
```

### A frontier estimated from data

Simulate 10 years of monthly returns with the true parameters, estimate means, volatilities and
correlation from the sample, and draw the frontier those estimates imply. Repeat for several
independent histories.

```python
def simulate_monthly(n_months, rng):
    mu_m = np.array([mu_a, mu_b]) / periods_per_year
    sd_m = np.array([sigma_a, sigma_b]) / np.sqrt(periods_per_year)
    cov_m = np.array([
        [sd_m[0]**2, rho * sd_m[0] * sd_m[1]],
        [rho * sd_m[0] * sd_m[1], sd_m[1]**2],
    ])
    return rng.multivariate_normal(mu_m, cov_m, size=n_months)


def annualized_estimates(r):
    mu_hat = r.mean(axis=0) * periods_per_year
    sd_hat = r.std(axis=0, ddof=1) * np.sqrt(periods_per_year)
    rho_hat = np.corrcoef(r[:, 0], r[:, 1])[0, 1]
    return mu_hat, sd_hat, rho_hat


n_histories = 8
estimates = [annualized_estimates(simulate_monthly(120, rng)) for _ in range(n_histories)]

pd.DataFrame(
    [{"mu_A": m[0], "mu_B": m[1], "sigma_A": s[0], "sigma_B": s[1], "rho": c} for m, s, c in estimates]
).round(3)
```

Volatilities land within one or two percentage points of the truth; the equity mean misses by
several percentage points in either direction. The frontier inherits that error.

## 6. Visualizations

```python
w_long = np.linspace(0, 1, 201)        # no short selling
w_short = np.linspace(-0.3, 1.3, 321)  # with short positions

fig, ax = plt.subplots(figsize=(9, 6))
for corr in [-1.0, -0.5, 0.1, 0.5, 1.0]:
    mu_p, sigma_p = portfolio_stats(w_long, mu_a, mu_b, sigma_a, sigma_b, corr)
    ax.plot(sigma_p, mu_p, label=f"ρ = {corr:+.1f}")
    w_mv = np.clip(min_variance_weight(sigma_a, sigma_b, corr), 0, 1)
    mu_mv, sigma_mv = portfolio_stats(w_mv, mu_a, mu_b, sigma_a, sigma_b, corr)
    ax.scatter(sigma_mv, mu_mv, s=20, color=ax.lines[-1].get_color(), zorder=3)

ax.scatter([sigma_a, sigma_b], [mu_a, mu_b], color="black", zorder=4)
ax.annotate("equities", (sigma_a, mu_a), textcoords="offset points", xytext=(6, -4))
ax.annotate("bonds", (sigma_b, mu_b), textcoords="offset points", xytext=(6, -10))
ax.set_title("Two-asset frontiers: lower correlation bends the curve further left")
ax.set_xlabel("annual volatility")
ax.set_ylabel("annual expected return")
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.set_xlim(left=0)
ax.grid(True, alpha=0.3)
ax.legend(title="dots: min-variance (long only)")
plt.show()
```

```python
mu_p, sigma_p = portfolio_stats(w_short, mu_a, mu_b, sigma_a, sigma_b, rho)
w_mv = min_variance_weight(sigma_a, sigma_b, rho)
efficient = w_short >= w_mv
long_only = (w_short >= 0) & (w_short <= 1)

fig, ax = plt.subplots(figsize=(9, 6))
# Full curve faded (includes short positions), long-only part drawn on top
ax.plot(sigma_p[efficient], mu_p[efficient], color="tab:green", alpha=0.3)
ax.plot(sigma_p[~efficient], mu_p[~efficient], color="tab:red", linestyle="--", alpha=0.3)
ax.plot(sigma_p[efficient & long_only], mu_p[efficient & long_only], color="tab:green", label="efficient branch")
ax.plot(sigma_p[~efficient & long_only], mu_p[~efficient & long_only], color="tab:red", linestyle="--", label="inefficient branch")

label_offsets = {"100% bonds": (6, -12), "min variance": (6, 2), "60/40": (6, -4), "100% equities": (6, -4)}
for label, offset in label_offsets.items():
    row = table.loc[label]
    ax.scatter(row["volatility"], row["expected return"], color="black", zorder=3)
    ax.annotate(label, (row["volatility"], row["expected return"]), textcoords="offset points", xytext=offset)

ax.set_title(f"Efficient and inefficient branches (ρ = {rho}); faded parts need short positions")
ax.set_xlabel("annual volatility")
ax.set_ylabel("annual expected return")
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.1%}"))
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 6))

true_mu, true_sigma = portfolio_stats(w_long, mu_a, mu_b, sigma_a, sigma_b, rho)
ax.plot(true_sigma, true_mu, color="black", linewidth=2.5, label="true frontier")

for i, (m, s, c) in enumerate(estimates):
    est_mu, est_sigma = portfolio_stats(w_long, m[0], m[1], s[0], s[1], c)
    ax.plot(est_sigma, est_mu, color="tab:blue", alpha=0.5, label="estimated from 10 years" if i == 0 else None)

ax.set_title("Frontiers estimated from independent 10-year monthly histories")
ax.set_xlabel("annual volatility")
ax.set_ylabel("annual expected return")
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

## 8. Solutions / Checks

```python
# 1.
mu_p, sigma_p = portfolio_stats(0.3, mu_a, mu_b, sigma_a, sigma_b, rho)
print(f"1. 30/70: expected return {float(mu_p):.2%}, volatility {float(sigma_p):.2%}")

# 2.
for corr in [-0.3, 0.5]:
    w_mv = min_variance_weight(sigma_a, sigma_b, corr)
    _, s_mv = portfolio_stats(w_mv, mu_a, mu_b, sigma_a, sigma_b, corr)
    print(f"2. rho = {corr:+.1f}: w_mv = {w_mv:+.2%}, min-variance volatility = {float(s_mv):.2%}")

# 3. 60/40 volatility at both correlations, then solve sigma_p(w) = 10% at rho = 0.5 on a grid
for corr in [-0.3, 0.5]:
    print(f"3. 60/40 volatility at rho = {corr:+.1f}: {float(portfolio_stats(0.6, mu_a, mu_b, sigma_a, sigma_b, corr)[1]):.2%}")
w_grid = np.linspace(0, 1, 100_001)
_, s_grid = portfolio_stats(w_grid, mu_a, mu_b, sigma_a, sigma_b, 0.5)
w_target = w_grid[(s_grid <= 0.10)].max()
print(f"   equity weight for 10% volatility at rho = +0.5: {w_target:.1%}")
```

**1.** $\mu_p = 0.3 \cdot 7 + 0.7 \cdot 2 = 3.5\%$;
$\sigma_p^2 = 0.09 \cdot 0.0256 + 0.49 \cdot 0.0025 + 2 \cdot 0.3 \cdot 0.7 \cdot 0.1 \cdot 0.008 = 0.003865$,
$\sigma_p \approx 6.22\%$.

**4.** At $\rho = 0.1$, 100% bonds lies on the **inefficient** branch: the minimum-variance
portfolio (about 6% equities) has lower volatility **and** higher expected return. Even the most
risk-averse investor should hold some equities. 100% bonds would be the minimum-variance
portfolio only if $\rho \ge \sigma_B/\sigma_A = 0.3125$ (and short selling is not allowed).

**5.** For $0 \le w \le 1$ both $w$ and $1-w$ are non-negative, so

$$
\big(w\sigma_A + (1-w)\sigma_B\big)^2 - \sigma_p^2 = 2w(1-w)\,\sigma_A\sigma_B\,(1 - \rho) \ge 0
$$

with equality (for $0 < w < 1$) only at $\rho = 1$. Taking square roots of non-negative numbers
keeps the order. A long-only portfolio is never riskier than the weighted average of its
holdings' volatilities, and strictly less risky unless all are perfectly correlated: that gap is
the diversification benefit.

## 9. Conclusion

- $\mu_p$ is linear in the weight, $\sigma_p^2$ is quadratic: the frontier is a hyperbola in the (volatility, return) plane.
- $\rho = 1$ gives the straight line (no diversification), $\rho = -1$ a riskless mix; real correlations bend the curve in between.
- Minimum-variance weight: $w_{\text{mv}} = (\sigma_B^2 - \rho\sigma_A\sigma_B)/(\sigma_A^2 + \sigma_B^2 - 2\rho\sigma_A\sigma_B)$; it needs a short position once $\rho > \sigma_B/\sigma_A$.
- Only the upper branch is efficient. A single low-risk asset can be inefficient (100% bonds).
- The frontier depends on estimated inputs, and expected returns are estimated badly.
