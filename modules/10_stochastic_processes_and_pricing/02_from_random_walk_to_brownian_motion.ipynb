# From Random Walk to Brownian Motion

The binomial tree of 10/01 moves the stock up or down in discrete steps. Let the steps become more frequent and smaller, in just the right
proportion, and the tree turns into a continuous-time process: Brownian motion. It is the building block of continuous-time finance. This
notebook constructs it as the limit of a scaled random walk, derives its main properties (independent normal increments, covariance $\min(s,t)$,
self-similarity, quadratic variation), and uses the reflection principle to compute the probability of touching a barrier.

## 1. Introduction

Option pricing, interest-rate models and risk models in continuous time all start from one object: a process $W_t$ whose increments over any
interval are normal with variance equal to the length of the interval. Its paths are continuous but so rough that ordinary calculus fails on
them, which is why 10/04 needs Itô's lemma. Seeing where the process comes from makes its strange properties less mysterious.

After this notebook you can:

- build Brownian motion as the limit of a scaled random walk and explain why the step size must be $\sqrt{\Delta t}$;
- state the defining properties and derive $\operatorname{Cov}(W_s, W_t) = \min(s, t)$;
- explain self-similarity, non-differentiability and quadratic variation;
- use the martingale property and the reflection principle;
- compute barrier-hitting probabilities and see the bias of discrete monitoring.

## 2. Theory

### The scaled random walk

Divide $[0, T]$ into $n$ steps of length $\Delta t = T/n$. Let $\xi_1, \xi_2, \dots$ be independent with $P(\xi_i = 1) = P(\xi_i = -1) = \frac12$, and

$$
W^{(n)}_t = \sqrt{\Delta t}\sum_{i=1}^{\lfloor t/\Delta t\rfloor}\xi_i
$$

Then $E[W^{(n)}_t] = 0$ and $\operatorname{Var}(W^{(n)}_t) = \lfloor t/\Delta t\rfloor\,\Delta t \approx t$, independent of $n$. The step size $\sqrt{\Delta t}$ is the only choice that keeps the variance finite
and non-zero: steps of size $\Delta t$ would make the variance vanish, steps of size 1 would make it explode. By the central limit theorem,
$W^{(n)}_t \to N(0, t)$ as $n \to \infty$, and the whole path converges to Brownian motion (Donsker's theorem).

### Definition

A **standard Brownian motion** $W_t$, $t \ge 0$, satisfies:

1. $W_0 = 0$;
2. independent increments: $W_{t_2} - W_{t_1}$, $W_{t_3} - W_{t_2}$, … are independent for $t_1 < t_2 < t_3 < \dots$;
3. normal increments: $W_t - W_s \sim N(0, t - s)$ for $s < t$;
4. continuous paths.

### Covariance

For $s < t$: $W_t = W_s + (W_t - W_s)$ with the second term independent of $W_s$, so

$$
\operatorname{Cov}(W_s, W_t) = \operatorname{Var}(W_s) = s = \min(s, t), \qquad \operatorname{Corr}(W_s, W_t) = \sqrt{s/t}
$$

### Self-similarity and roughness

For $c > 0$, $\tilde W_t = W_{ct}/\sqrt c$ is again a standard Brownian motion: zooming in by a factor $c$ in time and $\sqrt c$ in space gives a path that looks
statistically the same. The slope over an interval of length $h$, $(W_{t+h} - W_t)/h$, has standard deviation $1/\sqrt h \to \infty$ as $h \to 0$: the path has no derivative
anywhere.

### Quadratic variation

For a partition of $[0, T]$ into $n$ steps,

$$
\sum_{i=1}^n (W_{t_i} - W_{t_{i-1}})^2 \;\to\; T, \qquad \sum_{i=1}^n |W_{t_i} - W_{t_{i-1}}| \;\to\; \infty
$$

The first sum has mean $T$ and variance $2T^2/n \to 0$: in the limit it is not random. The second grows like $\sqrt n$. For a smooth function both limits would
be 0 and finite respectively. "$(dW)^2 = dt$" is the shorthand behind Itô's lemma (10/04). In finance, the sum of squared high-frequency returns
(**realized variance**) estimates $\sigma^2T$ for the same reason.

### Martingales

$E[W_t \mid \text{information at } s] = W_s$ for $s < t$: the best forecast of a future value is the current value. Also $W_t^2 - t$ is a martingale, while $W_t^2$ is not.

### Brownian motion with drift

$X_t = \mu t + \sigma W_t$ has $X_t - X_s \sim N(\mu(t-s), \sigma^2(t-s))$. It is the continuous version of the random walk with drift of 09/01, and the model for log prices in 10/03.

### Reflection principle

For $a > 0$, the first time $W$ reaches $a$, reflect the rest of the path. The reflected path ends above $a$ exactly when the original ends below.
Hence every path ending above $a$ has hit $a$, and half of the paths that hit $a$ end above it:

$$
P\Big(\max_{0 \le s \le t} W_s \ge a\Big) = 2P(W_t \ge a) = 2\Big(1 - \Phi\big(a/\sqrt t\big)\Big)
$$

With drift, the corresponding results give the barrier formula $e^{-2mx/\sigma^2}$ used for drawdowns in 05/03.

### Discrete monitoring

A barrier checked only at daily closes is hit less often than one monitored continuously: the path can cross and come back between observations.
A common correction (Broadie, Glasserman and Kou 1997) shifts the barrier by $0.5826\,\sigma\sqrt{\Delta t}$ away from the start.

## 3. Worked Examples

### Example 1: scaling the steps

One year in 100 steps: $\Delta t = 0.01$, step size $\pm 0.1$. After 100 steps the variance is $100\cdot 0.01 = 1$. With 10,000 steps: step size $\pm 0.01$, variance still 1.

### Example 2: covariance

$\operatorname{Cov}(W_{0.25}, W_1) = 0.25$, $\operatorname{Corr} = \sqrt{0.25} = 0.5$. Knowing the value after three months explains a quarter of the variance of the value after a year.

### Example 3: reflection

$P(\max_{[0,1]} W \ge 1) = 2(1 - \Phi(1)) = 2\cdot 0.1587 = 0.317$, while $P(W_1 \ge 1) = 0.159$.

### Example 4: a stock touching −20%

Log price as Brownian motion with volatility 20% and no drift. A 20% fall is $\ln 0.8 = -0.223$ in log terms. Within a year:
$P = 2\Phi(-0.223/0.20) = 2\Phi(-1.116) = 0.264$. The probability of **ending** the year 20% down is only $\Phi(-1.116) = 0.132$.

### Example 5: realized variance

Daily log returns over one year with 20% volatility: $\sum r_t^2 \approx \sigma^2T = 0.04$. The sum of absolute returns is about $252\cdot 0.0126\cdot\sqrt{2/\pi} = 2.53$, and doubles if the sampling
frequency quadruples.

## 4. Finance Application

### Where it matters

- **Continuous-time models.** Stock prices (10/03), interest rates and credit spreads are modelled as functions of Brownian motions.
- **Barrier options and stop-losses.** Knock-in and knock-out options, and the probability of hitting a stop-loss or a margin call, use the reflection
  principle; the discrete-monitoring correction matters for daily-monitored barriers.
- **Realized volatility.** Quadratic variation justifies estimating variance from the sum of squared intraday returns.
- **Drawdowns.** The probability that a strategy falls a given amount below its peak (05/03) follows from the same barrier results.

### The usual misreadings

- "Brownian motion means returns are normal." It means log returns over any interval are normal; real returns have fatter tails and jumps
  (exercise 3).
- "A smooth-looking price chart is differentiable." Every chart is drawn from discrete data; the underlying model has no derivative anywhere.
- "The probability of hitting a level equals the probability of ending beyond it." For a driftless Brownian motion it is twice as large.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
```

### Random walks with smaller and smaller steps

```python
def scaled_random_walk(n_steps, T, n_paths, rng_):
    dt = T / n_steps
    steps = rng_.choice([-1.0, 1.0], size=(n_paths, n_steps)) * np.sqrt(dt)
    return np.concatenate([np.zeros((n_paths, 1)), np.cumsum(steps, axis=1)], axis=1)

rows = []
for n_steps in [4, 16, 64, 256, 1024]:
    W1 = scaled_random_walk(n_steps, 1.0, 20_000, rng)[:, -1]
    rows.append({"steps": n_steps, "mean W_1": W1.mean(), "var W_1": W1.var(), "kurtosis": stats.kurtosis(W1, fisher=False),
                 "P(W_1 > 1.645)": np.mean(W1 > 1.645), "distinct values": np.unique(W1.round(10)).size})
pd.DataFrame(rows).set_index("steps").round(4)
```

For every number of steps the variance of $W^{(n)}_1$ is 1. With 4 steps the distribution has only five possible values and kurtosis 2.5; by 64 steps the
kurtosis is close to the normal 3 and the tail probability $P(W_1 > 1.645)$ close to 5%.

### Properties of Brownian motion

Simulate Brownian motion on a fine grid (normal increments) and check covariance, martingale behaviour and the scaling property.

```python
def brownian_paths(n_steps, T, n_paths, rng_):
    dt = T / n_steps
    dW = np.sqrt(dt) * rng_.standard_normal((n_paths, n_steps))
    return np.concatenate([np.zeros((n_paths, 1)), np.cumsum(dW, axis=1)], axis=1)

n_steps, T = 1_000, 1.0
Wp = brownian_paths(n_steps, T, 20_000, rng)
t_grid = np.linspace(0, T, n_steps + 1)
i25, i50, i100 = 250, 500, 1000
print(f"Cov(W_0.25, W_1) = {np.cov(Wp[:, i25], Wp[:, i100])[0, 1]:.4f} (theory 0.25), "
      f"Corr = {np.corrcoef(Wp[:, i25], Wp[:, i100])[0, 1]:.3f} (theory 0.5)")
print(f"E[W_t^2 - t] at t = 0.5, 1: {np.mean(Wp[:, i50]**2) - 0.5:.4f}, {np.mean(Wp[:, i100]**2) - 1.0:.4f};  "
      f"E[W_t^2]: {np.mean(Wp[:, i50]**2):.3f}, {np.mean(Wp[:, i100]**2):.3f}")
up = Wp[:, i50] > 0.5
print(f"martingale: E[W_1 | W_0.5 > 0.5] = {Wp[up, i100].mean():.4f}, E[W_0.5 | W_0.5 > 0.5] = {Wp[up, i50].mean():.4f}")
for h in [0.1, 0.01, 0.001]:
    k = int(h * n_steps)
    slopes = (Wp[:, k] - Wp[:, 0]) / h
    print(f"slope over h = {h:5.3f}: std {slopes.std():7.2f} (theory 1/sqrt(h) = {1 / np.sqrt(h):6.2f})")
```

The simulated covariance (0.252) and correlation (0.502) match $\min(s, t)$ and $\sqrt{s/t}$. $E[W_t^2]$ grows with $t$ while $E[W_t^2 - t]$ stays at zero. Given a high value at
$t = 0.5$, the expected value at $t = 1$ is the same as at $t = 0.5$ (0.919 against 0.918): the martingale property. The standard deviation of the slope grows
like $1/\sqrt h$ as the interval shrinks: there is no limiting derivative.

### Quadratic vs total variation

```python
fine = brownian_paths(2**16, 1.0, 1, rng)[0]
qv_rows = []
for n_sub in [2**k for k in range(4, 17, 2)]:
    sub = fine[:: 2**16 // n_sub]
    inc = np.diff(sub)
    qv_rows.append({"steps": n_sub, "quadratic variation": np.sum(inc**2), "total variation": np.sum(np.abs(inc)),
                    "theory SD of QV": np.sqrt(2 / n_sub)})
qv = pd.DataFrame(qv_rows).set_index("steps")
qv.round(4)
```

As the same path is sampled more finely, the sum of squared increments settles at 1 (1.003–1.010 from 16,384 steps on), while the sum of absolute increments
doubles every time the number of steps quadruples, from 3.4 to 205.

### Reflection principle and discrete monitoring

```python
Wm = brownian_paths(2_000, 1.0, 20_000, rng)
running_max = Wm.max(axis=1)
for a in [0.5, 1.0, 1.5]:
    sim = np.mean(running_max >= a)
    print(f"a = {a}: P(max >= a) simulated {sim:.4f}, reflection 2(1 - Phi(a)) = {2 * stats.norm.sf(a):.4f}, "
          f"P(W_1 >= a) = {stats.norm.sf(a):.4f}")

sigma, barrier = 0.20, np.log(0.8)
LP = sigma * Wm                                         # log price, 2,000 steps per year
cont = 2 * stats.norm.cdf(barrier / sigma)
daily = np.mean((LP[:, ::2_000 // 250][:, 1:]).min(axis=1) <= barrier)
monthly = np.mean((LP[:, ::2_000 // 12][:, 1:]).min(axis=1) <= barrier)
fine_hit = np.mean(LP.min(axis=1) <= barrier)
corr_daily = 2 * stats.norm.cdf((barrier - 0.5826 * sigma * np.sqrt(1 / 250)) / sigma)
print(f"\nstock touches -20% within a year: continuous formula {cont:.4f}, simulated on 2,000 steps {fine_hit:.4f}")
print(f"monitored daily {daily:.4f} (corrected formula {corr_daily:.4f}), monitored monthly {monthly:.4f}")
```

The simulated probabilities of reaching $a$ match the reflection formula and are twice the probability of ending above $a$ (slightly below the
formula because the 2,000-step grid misses some crossings). For a stock with 20% volatility, the probability of touching −20% within a year is 26.5% with
continuous monitoring, 24.8% when checked at daily closes (the corrected formula gives 24.9%), and 20.0% when checked monthly.

## 8. Solutions / Checks

```python
# 1. and 2.
print(f"1. Var(W3 - W1) = 2, Cov(W2, W5) = 2, Corr = {np.sqrt(2 / 5):.3f}")
print(f"2. P = 2(1 - Phi(2/2)) = {2 * stats.norm.sf(1):.4f}")

# 3.
p_month = 2 * stats.norm.cdf(np.log(0.8) / (0.20 * np.sqrt(1 / 12)))
print(f"3. P(touch -20% within a month) = {p_month:.2e}, about once in {1 / p_month / 12:,.0f} years")

# 4.
W4 = brownian_paths(2_000, 2.0, 20_000, rng)
for t4 in [0.25, 0.5, 1.0, 2.0]:
    w = W4[:, int(t4 / 2.0 * 2_000)]
    print(f"4. t = {t4}: E[W^2] {np.mean(w**2):.3f}, E[W^2 - t] {np.mean(w**2 - t4):+.3f}, E[W^3 - 3tW] {np.mean(w**3 - 3 * t4 * w):+.3f}")
```

```python
# 5.
mu5, s5, level, years5, steps_per_year = 0.06, 0.15, -0.3, 50, 100
dt5 = 1 / steps_per_year
hit = np.zeros(4_000, dtype=bool)
X = np.zeros(4_000)
for _ in range(years5 * steps_per_year):
    X += mu5 * dt5 + s5 * np.sqrt(dt5) * rng.standard_normal(4_000)
    hit |= X <= level
print(f"5. simulated {hit.mean():.3f} (discrete steps, finite horizon), formula {np.exp(-2 * mu5 * 0.3 / s5**2):.3f}")

# 6.
for n6 in [10, 1_000]:
    inc = np.sqrt(1 / n6) * rng.standard_normal((10_000, n6))
    qv6 = (inc**2).sum(axis=1)
    print(f"6. n = {n6}: mean {qv6.mean():.4f}, variance {qv6.var():.5f} (theory {2 / n6:.5f})")

# 7.
levels = 10
W7 = np.array([0.0, rng.standard_normal()])
for lev in range(levels):
    h = 1 / 2 ** (lev + 1)                                   # half of the current interval length
    mids = (W7[:-1] + W7[1:]) / 2 + np.sqrt(h / 2) * rng.standard_normal(W7.size - 1)
    new = np.empty(2 * W7.size - 1)
    new[0::2], new[1::2] = W7, mids
    W7 = new
inc7 = np.diff(W7)
print(f"7. grid points {W7.size}, increment variance {inc7.var():.6f} (theory {2**-levels:.6f}), "
      f"lag-1 autocorrelation {np.corrcoef(inc7[1:], inc7[:-1])[0, 1]:.3f}")
```

**1.** $\operatorname{Var}(W_3 - W_1) = 2$; $\operatorname{Cov}(W_2, W_5) = \min(2, 5) = 2$; $\operatorname{Corr} = \sqrt{2/5} = 0.632$.

**2.** $2(1 - \Phi(2/\sqrt 4)) = 2(1 - \Phi(1)) = 0.317$.

**3.** Under the model, $P = 2\Phi(\ln 0.8/(0.2\sqrt{1/12})) = 2\Phi(-3.87) \approx 1.1\cdot 10^{-4}$, once in several hundred years. Real stocks fall 20% within a month far more often
(several times per decade for single stocks, and in most market crises for indices). The gap is the fat tails and jumps of real returns, and
volatility that rises in crises (09/02): a Brownian model with constant volatility understates such events badly.

**4.** $E[W_t^2] = t$ grows; $E[W_t^2 - t]$ and $E[W_t^3 - 3tW_t]$ stay at zero: both are martingales.

**5.** The simulated probability, 0.194, is close to the formula $e^{-2\cdot 0.06\cdot 0.3/0.0225} = 0.202$; it is a little lower because the barrier is checked only at discrete steps
and the horizon is finite.

**6.** Each $(\Delta W_i)^2 = \Delta t\,Z_i^2$ with $Z_i$ standard normal has mean $\Delta t$ and variance $2\Delta t^2$; summing $n$ independent terms gives mean $n\Delta t = T$ and variance $2n\Delta t^2 = 2T^2/n$.

**7.** The midpoint construction reproduces Brownian motion: increments on the final grid have variance $2^{-10}$ and no autocorrelation. (Given the
endpoints, the midpoint of a Brownian path is normal with the average as mean and a quarter of the interval as variance: a Brownian bridge.)

## 9. Conclusion

- A random walk with steps $\pm\sqrt{\Delta t}$ keeps variance $t$ at every scale and converges to Brownian motion.
- Brownian motion has independent increments $W_t - W_s \sim N(0, t-s)$, covariance $\min(s, t)$ and continuous, nowhere-differentiable paths.
- It is self-similar; its quadratic variation over $[0, T]$ is exactly $T$, the basis of realized variance and of Itô's lemma.
- $W_t$ and $W_t^2 - t$ are martingales.
- The reflection principle gives $P(\max_{[0,t]} W \ge a) = 2P(W_t \ge a)$: hitting a level is twice as likely as ending beyond it; discrete monitoring lowers it.
