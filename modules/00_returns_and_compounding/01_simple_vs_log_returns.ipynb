# Simple vs Log Returns: Aggregation over Time and across Assets

A return can be measured in two ways. The simple return $r = P_1/P_0 - 1$ is what an investor earns. The log return $\ln(P_1/P_0)$ is what
a mathematician prefers, because log returns add up over time. Simple returns add up across assets in a portfolio; log returns do not.
Mixing the two produces errors that are small for a day and large for a year. This notebook derives both definitions, shows which one
aggregates where, and measures the size of the error when the wrong one is used.

## 0. Math link

- [Derivative of logarithmic functions](https://github.com/dnlwrthstr/LearningCalculus/blob/main/hypersklill/02_derivaties/05_derivative_of_logarithmic_functions.ipynb):
  $\frac{d}{dx}\ln x = 1/x$, so a log return is the sum (integral) of tiny relative price changes $dP/P$.
- [Derivative of exponential functions](https://github.com/dnlwrthstr/LearningCalculus/blob/main/hypersklill/02_derivaties/06_derivative_of_exponential_functions.ipynb):
  $e^x$ undoes $\ln$, which turns a sum of log returns back into a growth factor.

## 1. Introduction

A Swiss stock falls 50% and then rises 50%. The two simple returns average to zero, yet the investor has lost a quarter of the money. The two
log returns, $\ln 0.5 = -0.693$ and $\ln 1.5 = 0.405$, add to $\ln 0.75 = -0.288$, the true total. Log returns get time right. But a portfolio
that holds 60% in one stock and 40% in another earns $0.6\,r_1 + 0.4\,r_2$ in simple returns, and no such formula holds for log returns.
Simple returns get the cross-section right.

After this notebook you can:

- compute simple and log returns from prices and convert between them;
- aggregate returns over time (products of growth factors, or sums of logs);
- aggregate across assets (weighted sums of simple returns);
- size the error of the approximation $\ln(1+r) \approx r$ and of averaging the wrong kind of return;
- explain why the arithmetic mean return overstates the growth of wealth by about $\sigma^2/2$.

## 2. Theory

### Two definitions

For prices $P_{t-1}$ and $P_t$ (with a dividend $D_t$ added to $P_t$ if paid):

$$
r_t = \frac{P_t}{P_{t-1}} - 1, \qquad \ell_t = \ln\frac{P_t}{P_{t-1}} = \ln(1 + r_t), \qquad r_t = e^{\ell_t} - 1 .
$$

The simple return is bounded below by $-1$ (a total loss); the log return can be any real number, which is one reason it suits models
with normal distributions.

### Why the logarithm: summing relative changes

Split the period into many small steps. Each step changes the price by $dP$, a relative change of $dP/P$. Adding them up,

$$
\int_{P_{t-1}}^{P_t} \frac{dP}{P} = \ln P_t - \ln P_{t-1} = \ell_t ,
$$

because $\frac{d}{dP}\ln P = 1/P$. The log return is the sum of infinitely many tiny simple returns, i.e. the continuously compounded rate
(continuous compounding gets its own notebook, 00/02).

### Aggregation over time

Wealth multiplies by $1 + r_t$ each period, so over $T$ periods

$$
1 + R_{0,T} = \prod_{t=1}^{T} (1 + r_t), \qquad \ell_{0,T} = \sum_{t=1}^{T} \ell_t .
$$

Log returns **add** over time. Simple returns do not: $\sum_t r_t$ is not the total return.

### Aggregation across assets

A portfolio with weights $w_i$ (fractions of value at the start of the period, $\sum_i w_i = 1$) is worth $\sum_i w_i (1 + r_i)$ per franc
invested, so

$$
r_p = \sum_i w_i r_i, \qquad \ell_p = \ln\Big(\sum_i w_i e^{\ell_i}\Big) \neq \sum_i w_i \ell_i .
$$

Simple returns **add** across assets. Because $\ln$ is concave, Jensen's inequality gives $\sum_i w_i \ell_i \le \ell_p$: the weighted log
return understates the portfolio's log return.

### Size of the gap

The Taylor expansion $\ln(1 + r) = r - \frac{r^2}{2} + \frac{r^3}{3} - \dots$ gives

$$
\ell \approx r - \tfrac12 r^2 .
$$

For a daily return of 1% the difference is 0.005 percentage points; for an annual return of 30% it is about 4 points. Averaging over
periods, $E[\ell] \approx E[r] - \tfrac12 E[r^2] \approx E[r] - \tfrac12 \sigma^2$ when the mean is small compared with the volatility.

### Arithmetic and geometric means

The geometric mean $G = \big(\prod_t (1 + r_t)\big)^{1/T} - 1$ is the constant return that produces the same final wealth; $\ln(1+G)$ is the
average log return. The arithmetic mean $A = \frac1T \sum_t r_t$ is always at least $G$ (AM–GM inequality), and

$$
G \approx A - \tfrac12 \sigma^2 .
$$

This gap is called **volatility drag**: two funds with the same arithmetic mean grow at different rates if their volatilities differ.

## 3. Worked Examples

### Example 1: down 50%, up 50%

Prices 100, 50, 75 (CHF). Simple returns $-0.50$ and $+0.50$, arithmetic mean $0$. Total return $0.5 \times 1.5 - 1 = -0.25$.
Log returns $-0.693$ and $+0.405$, sum $-0.288$, and $e^{-0.288} - 1 = -0.25$. The geometric mean is $\sqrt{0.75} - 1 = -13.4\%$ per period.

### Example 2: a two-asset portfolio

CHF 600 in a stock that gains 10% and CHF 400 in one that loses 20%. End value $660 + 320 = 980$, so $r_p = -2\%$, and indeed $0.6 \times 0.10
+ 0.4 \times (-0.20) = -0.02$. The log returns are $\ln 1.1 = 0.0953$ and $\ln 0.8 = -0.2231$; their weighted sum is $-0.0321$, while the
true portfolio log return is $\ln 0.98 = -0.0202$. Using the weighted log return overstates the loss by 1.2 percentage points.

### Example 3: how good is $\ell \approx r$?

| simple return $r$ | log return $\ell$ | $r - \ell$ |
|---|---|---|
| 0.01 | 0.00995 | 0.00005 |
| 0.10 | 0.0953 | 0.0047 |
| 0.30 | 0.2624 | 0.0376 |
| −0.30 | −0.3567 | 0.0567 |

The gap grows with the square of the return and is larger for losses than for gains of the same size.

### Example 4: a fund report

A fund returned +25%, −20%, +25%, −20% in four years. The arithmetic mean is 2.5% a year; the investor's money is exactly where it started
($1.25^2 \times 0.8^2 = 1$), so the geometric mean is 0%. With a volatility of about 22.5% per year, $\tfrac12\sigma^2 = 2.5\%$, which
explains the whole gap.

## 4. Finance Application

### Where it matters

- **Performance reporting.** Multi-year performance is the product of growth factors, and annualized returns are geometric means (GIPS
  standards require time-weighted, geometrically linked returns).
- **Portfolio accounting.** Daily portfolio returns are weighted sums of simple asset returns; risk systems that aggregate positions use
  simple (or P&L) numbers across assets.
- **Modelling.** Statistical models of single assets (random walks, GARCH, Black–Scholes) use log returns because they add over time and are
  unbounded.
- **Forecasting wealth.** Expected wealth grows at the arithmetic mean, median wealth at the geometric mean; long-horizon planning has to
  say which.

### The usual misreadings

- "The average return was 2.5%, so the fund made money." The arithmetic mean says nothing about growth when volatility is high.
- "Log returns and simple returns are the same." Only for small returns; for annual equity returns the difference is several percentage points.
- "The portfolio log return is the weighted log return." Across assets, only simple returns are additive.
- "A 50% loss and a 50% gain cancel." It takes a 100% gain to recover from a 50% loss.

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
```

### Simulated prices for two Swiss stocks

Ten years of daily prices (2015–2024, weekdays only, about 261 a year) for two stocks with annual volatilities of 18% and 30%, correlation 0.5, and an
expected log return of 6% a year. Prices start at CHF 100.

```python
dates = pd.bdate_range("2015-01-01", "2024-12-31")   # weekdays; holidays ignored
years = 10
n_days = len(dates) - 1
periods_per_year = n_days / years                    # about 261 weekdays a year

vol = np.array([0.18, 0.30])          # annual volatilities
mu_log = np.array([0.06, 0.06])       # annual expected log returns
corr = np.array([[1.0, 0.5], [0.5, 1.0]])
cov_daily = np.outer(vol, vol) * corr / periods_per_year

log_r = rng.multivariate_normal(mu_log / periods_per_year, cov_daily, size=n_days)
prices = 100 * np.exp(np.vstack([np.zeros(2), np.cumsum(log_r, axis=0)]))
prices = pd.DataFrame(prices, index=dates, columns=["Stock A", "Stock B"])

r = prices.pct_change().dropna()            # simple returns, decimals, daily
log_r = np.log(prices).diff().dropna()      # log returns, decimals, daily
print(prices.iloc[[0, -1]].round(2))
print("\nmax |log_r - log(1 + r)|:", float(np.abs(log_r - np.log1p(r)).max().max()))
```

### Time aggregation: products of simple returns, sums of log returns

```python
total_from_prices = prices.iloc[-1] / prices.iloc[0] - 1
total_from_simple = (1 + r).prod() - 1
total_from_logs = np.exp(log_r.sum()) - 1
sum_of_simple = r.sum()

pd.DataFrame({
    "from prices": total_from_prices,
    "product of (1 + r)": total_from_simple,
    "exp(sum of log r) - 1": total_from_logs,
    "sum of r (wrong)": sum_of_simple,
})
```

It was an unlucky decade: both stocks ended below CHF 100 despite an expected log return of 6% a year (how noisy a ten-year mean is
comes in module 06). The first three columns agree: they are the same number computed three ways. Summing the simple returns gets even
the sign wrong for Stock B, +10% instead of −29%, because each daily return is a little larger than its log return, by about $\tfrac12 r^2$, and the
2,600 small differences add up to roughly $\tfrac12 \sigma^2 \times 10$ years.

```python
gap = sum_of_simple - log_r.sum()
print("sum of r - sum of log r:", gap.round(4).to_dict())
print("0.5 * annual variance * years:", (0.5 * vol**2 * years).round(4))
```

### Aggregation across assets: a 60/40 portfolio

A portfolio rebalanced to 60/40 at the start of each period. Its simple return is the weighted sum of simple returns; its log return is
$\ln(1 + r_p)$. We compare with the (wrong) weighted log return at daily, monthly and annual frequencies.

```python
w = np.array([0.6, 0.4])

def portfolio_gap(freq):
    # resample prices to the given frequency, rebalance to w at the start of each period
    p = pd.concat([prices.iloc[[0]], prices.resample(freq).last()]) if freq != "D" else prices
    r_f = p.pct_change().dropna()
    lr_f = np.log(p).diff().dropna()
    r_p = r_f @ w
    true_log = np.log1p(r_p)
    approx_log = lr_f @ w
    err = true_log - approx_log
    return {"periods": len(r_p), "mean error": err.mean(), "max error": err.max(),
            "mean |r_p|": r_p.abs().mean()}

pd.DataFrame({name: portfolio_gap(f) for name, f in [("daily", "D"), ("monthly", "ME"), ("annual", "YE")]}).T
```

The error is always positive (Jensen's inequality) and grows with the length of the period: negligible for daily data, visible for
monthly data and material for annual returns. It is roughly $\tfrac12 \sum_i w_i (\ell_i - \bar\ell)^2$, half the weighted spread of
the asset returns, so it is larger when the assets move differently.

### Arithmetic vs geometric mean

```python
annual = prices.resample("YE").last()
annual = pd.concat([prices.iloc[[0]], annual])      # include the start price
r_annual = annual.pct_change().dropna()

A = r_annual.mean()
G = (1 + r_annual).prod() ** (1 / len(r_annual)) - 1
sigma = r_annual.std(ddof=0)
pd.DataFrame({"arithmetic A": A, "geometric G": G, "A - G": A - G, "sigma^2 / 2": sigma**2 / 2})
```

For both stocks $A - G$ is close to $\sigma^2/2$ computed from the same ten annual returns. Stock B, with the higher volatility, has the
larger drag. With only ten observations the means themselves are very noisy (module 06), but the gap between them is a near-deterministic
function of the spread.

## 6. Visualizations

```python
x = np.linspace(-0.6, 0.6, 400)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].plot(x, x, label="simple return r")
axes[0].plot(x, np.log1p(x), label="log return ln(1 + r)")
axes[0].plot(x, x - x**2 / 2, "--", label="r - r²/2")
axes[0].set_title("Log return against simple return")
axes[0].set_xlabel("simple return (decimal)")
axes[0].set_ylabel("return (decimal)")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(x * 100, (x - np.log1p(x)) * 100)
axes[1].set_title("Gap r - ln(1 + r)")
axes[1].set_xlabel("simple return (%)")
axes[1].set_ylabel("gap (percentage points)")
axes[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

The log return lies below the simple return everywhere except at zero, and the gap grows quadratically. For losses near −60% it is
over 30 percentage points; for gains and losses below 5% it is under 0.15 points.

```python
wealth_true = prices["Stock B"] / 100
wealth_sum_simple = 1 + r["Stock B"].cumsum()
wealth_sum_log = np.exp(log_r["Stock B"].cumsum())

plt.figure(figsize=(10, 5))
plt.plot(wealth_true, lw=3, alpha=0.4, label="actual wealth (P_t / P_0)")
plt.plot(wealth_sum_log, lw=1, label="exp(cumulative sum of log returns)")
plt.plot(wealth_sum_simple, lw=1, label="1 + cumulative sum of simple returns (wrong)")
plt.title("Stock B: growth of CHF 1, three ways")
plt.xlabel("date")
plt.ylabel("wealth (CHF per CHF 1 invested)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

The cumulative log return reproduces wealth exactly. The cumulative simple return drifts above it, and the gap widens steadily: summing
simple returns ignores compounding and the volatility drag.

```python
vols = np.linspace(0, 0.5, 6)
A_true = 0.07                                   # arithmetic mean of annual simple returns
n_paths, n_years = 4000, 30
fig, ax = plt.subplots(figsize=(10, 5))
for s in vols:
    # lognormal annual growth factors with mean 1 + A_true and volatility s
    s_log = np.sqrt(np.log(1 + s**2 / (1 + A_true)**2))
    m_log = np.log(1 + A_true) - s_log**2 / 2
    growth = np.exp(rng.normal(m_log, s_log, size=(n_paths, n_years)))
    wealth = np.cumprod(growth, axis=1)
    ax.plot(range(1, n_years + 1), np.median(wealth, axis=0), label=f"vol {s:.0%}")
ax.set_yscale("log")
ax.set_yticks([0.5, 1, 2, 4, 8])
ax.set_yticklabels(["0.5", "1", "2", "4", "8"])
ax.set_title("Median wealth with the same 7% arithmetic mean return")
ax.set_xlabel("years")
ax.set_ylabel("median wealth (CHF per CHF 1, log scale)")
ax.legend(title="annual volatility")
ax.grid(True, alpha=0.3, which="both")
plt.show()
```

All six investments have the same expected annual return of 7%. With zero volatility CHF 1 grows to about CHF 7.6 in 30 years; with
40% volatility the median investor barely moves, and with 50% ends near CHF 0.35. Volatility drag is invisible in the arithmetic mean and dominant in the outcome.

```python
spreads = np.linspace(0, 0.4, 9)
errs = []
for d in spreads:
    l1, l2 = 0.05 + d / 2, 0.05 - d / 2          # log returns of the two assets
    true_log = np.log(0.6 * np.exp(l1) + 0.4 * np.exp(l2))
    errs.append(true_log - (0.6 * l1 + 0.4 * l2))
errs = np.array(errs)

plt.figure(figsize=(8, 5))
plt.plot(spreads * 100, errs * 100, "o-", label="exact error")
plt.plot(spreads * 100, 0.5 * 0.6 * 0.4 * spreads**2 * 100, "--", label="½ · w₁w₂ · (ℓ₁ − ℓ₂)²")
plt.title("Error of the weighted log return, 60/40 portfolio")
plt.xlabel("difference between the two assets' log returns (percentage points)")
plt.ylabel("error (percentage points)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

The error of averaging log returns across assets depends on how far apart the asset returns are. For two assets it is close to
$\tfrac12 w_1 w_2 (\ell_1 - \ell_2)^2$, the weighted variance of the two log returns. A 40-point spread, common between stocks in a year,
costs almost 2 percentage points.

## 8. Solutions / Checks

```python
# 1.
r1 = (86 + 2) / 80 - 1
print(f"1. simple {r1:.4f}, log {np.log1p(r1):.4f}")

# 2.
m = np.array([0.04, -0.03, 0.02])
print(f"2. product {np.prod(1 + m) - 1:.6f}, via logs {np.exp(np.log1p(m).sum()) - 1:.6f}")

# 3.
A3, s3 = 0.09, 0.20
print(f"3. approximate geometric mean {A3 - s3**2 / 2:.3f}")
```

**3.** $G \approx 9\% - \tfrac12 \times 0.20^2 = 7\%$. The 9% describes the average of yearly returns and the growth of *expected* wealth, which is
pulled up by a few very good paths. A typical (median) investor compounds at about 7% a year.

```python
# 4.
monthly = prices.resample("ME").last()
monthly = pd.concat([prices.iloc[[0]], monthly])
r_m = monthly.pct_change().dropna()
rebalanced = (1 + r_m @ np.array([0.5, 0.5])).prod() - 1
buy_hold = (0.5 * monthly.iloc[-1] / monthly.iloc[0]).sum() - 1
print(f"4. rebalanced monthly {rebalanced:.4f}, buy-and-hold {buy_hold:.4f}")

# 5.
rB, lB = r["Stock B"], log_r["Stock B"]
lhs = rB.mean() * periods_per_year - lB.mean() * periods_per_year
print(f"5. annualized mean simple - mean log {lhs:.4f}, half variance {0.5 * rB.var() * periods_per_year:.4f}")
```

**4.** Both are correct ten-year returns of different strategies; they differ because the rebalanced portfolio sells the winner and buys the loser
each month. Neither is computed by summing returns.

**5.** The two numbers agree to within rounding: summing daily simple returns overstates growth by half the variance per year.

**6.** $\ln$ is concave, so Jensen's inequality gives $\sum_i w_i \ln x_i \le \ln \sum_i w_i x_i$ with $x_i = e^{\ell_i}$. For the size, expand
$e^{\ell_i} = e^{\bar\ell}\big(1 + d_i + \tfrac12 d_i^2 + \dots\big)$ with $d_i = \ell_i - \bar\ell$. Since $\sum_i w_i d_i = 0$,
$\ln \sum_i w_i e^{\ell_i} = \bar\ell + \ln\big(1 + \tfrac12 \sum_i w_i d_i^2 + \dots\big) \approx \bar\ell + \tfrac12 \sum_i w_i d_i^2$.

**7.** $E[e^{X}] = e^{m + s^2/2}$ for $X \sim N(m, s^2)$ (the moment generating function of the normal), so $E[r] = e^{m + s^2/2} - 1$. The
median of $e^X$ is $e^m$ because $e^x$ is increasing. After $T$ years the log growth is $N(Tm, Ts^2)$, so mean over median wealth is
$e^{T s^2/2}$; with $s = 0.2$ and $T = 30$ this is $e^{0.6} = 1.82$.

## 9. Conclusion

- Simple returns $r = P_1/P_0 - 1$ are what investors earn; log returns $\ell = \ln(1 + r)$ are sums of tiny relative changes.
- Over time, multiply growth factors or add log returns; never add simple returns.
- Across assets, take the weighted sum of simple returns; the weighted log return understates the portfolio log return.
- $\ell \approx r - r^2/2$: the two agree for daily data and differ by percentage points for annual data.
- The geometric mean is about $\sigma^2/2$ below the arithmetic mean; volatility drag decides long-run median wealth.
