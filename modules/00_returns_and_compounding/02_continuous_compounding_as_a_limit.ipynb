# Continuous Compounding as a Limit

A bank that pays 3% a year once a year pays less than one that pays 1.5% twice a year, which pays less than one that pays 0.25% every
month. Compounding more often helps, but by less and less: the growth factor converges to $e^{0.03}$. That limit, continuous
compounding, is the convention behind log returns, discount factors $e^{-rt}$ and almost every pricing formula. This notebook derives
the limit, measures how fast it is reached, and converts rates between compounding conventions.

- [Limits of sequences](https://github.com/dnlwrthstr/LearningCalculus/blob/main/hypersklill/01_limits/00_limits_of_sequences.ipynb):
  the growth factor $(1 + r/m)^m$ is a sequence in $m$ whose limit is $e^r$.
- [Meaning of the exponential e](https://github.com/dnlwrthstr/LearningCalculus/blob/main/coursera/00_derivatives/09_meaning_of_the_exponential_e.ipynb)
  and [derivative of exponential functions](https://github.com/dnlwrthstr/LearningCalculus/blob/main/hypersklill/02_derivaties/06_derivative_of_exponential_functions.ipynb):
  $e^{rt}$ is the function whose growth rate is proportional to its value, $\frac{d}{dt}e^{rt} = r e^{rt}$.
- Log returns from [00/01](./01_simple_vs_log_returns.ipynb): a continuously compounded rate is a log return.

## 1. Introduction

Interest rates are quoted with a compounding convention: annual for Swiss government bonds, semi-annual for many US bonds, monthly for
some loans, simple (no compounding) for money-market deposits, and continuous in derivatives pricing. The same number means different
things under different conventions, so a rate is incomplete without its convention. Continuous compounding is the common denominator:
every other convention converts to it with a logarithm.

After this notebook you can:

- compute growth with $m$ compounding periods per year and its limit $e^{rt}$;
- prove the limit with logarithms and estimate the error for finite $m$;
- convert between annual, $m$-times-a-year, effective annual and continuous rates;
- discount with $e^{-rt}$ and explain the link to log returns;
- compute doubling times and check the rule of 72.

## 2. Theory

### Compounding $m$ times a year

A nominal annual rate $r$ compounded $m$ times a year pays $r/m$ per sub-period. After one year CHF 1 grows to

$$
a_m = \Big(1 + \frac{r}{m}\Big)^{m},
$$

and after $t$ years to $(1 + r/m)^{mt}$.

### The limit

Take logarithms: $\ln a_m = m \ln(1 + r/m)$. With $x = r/m \to 0$,

$$
\ln a_m = r \cdot \frac{\ln(1 + x)}{x} \longrightarrow r \cdot 1 = r ,
$$

because $\lim_{x\to 0} \ln(1+x)/x = \frac{d}{dx}\ln(1+x)\big|_{x=0} = 1$. Since $e^x$ is continuous, $a_m \to e^r$:

$$
\lim_{m \to \infty} \Big(1 + \frac{r}{m}\Big)^{m} = e^{r}, \qquad \text{growth over } t \text{ years: } e^{rt} .
$$

### How fast

From $\ln(1+x) = x - x^2/2 + O(x^3)$, $\ln a_m = r - \frac{r^2}{2m} + O(m^{-2})$, so

$$
e^{r} - a_m \approx e^{r}\,\frac{r^2}{2m} .
$$

The error falls like $1/m$: monthly compounding is about 12 times closer to continuous than annual compounding. The sequence also
increases with $m$ for $r > 0$, so more frequent compounding always pays more.

### The differential-equation view

Continuous compounding means the balance $B(t)$ grows at the rate $r$ times itself:

$$
\frac{dB}{dt} = r B \quad\Rightarrow\quad B(t) = B(0)\, e^{rt},
$$

because $\frac{d}{dt} e^{rt} = r e^{rt}$. Discounting runs this backwards: CHF 1 at time $t$ is worth $e^{-rt}$ today.

### Converting between conventions

Two rates are equivalent if they produce the same growth over a year. With $r_c$ continuous, $r_m$ compounded $m$ times a year and
$r_{\text{eff}}$ the effective annual rate (annual compounding):

$$
1 + r_{\text{eff}} = \Big(1 + \frac{r_m}{m}\Big)^m = e^{r_c}, \qquad r_c = m\ln\Big(1 + \frac{r_m}{m}\Big), \qquad r_m = m\big(e^{r_c/m} - 1\big).
$$

The continuous rate is the log of the annual growth factor, i.e. the annual log return.

### Doubling time

Solving $e^{r_c T} = 2$ gives $T = \ln 2 / r_c \approx 0.693/r_c$. With annual compounding, $T = \ln 2 / \ln(1 + r)$; the **rule of 72**,
$T \approx 72 / (100 r)$, is tuned for rates around 8%.

## 3. Worked Examples

### Example 1: CHF 10,000 at 3% for one year

| compounding | growth factor | end value (CHF) |
|---|---|---|
| annual, $m = 1$ | 1.030000 | 10,300.00 |
| semi-annual, $m = 2$ | 1.030225 | 10,302.25 |
| monthly, $m = 12$ | 1.030416 | 10,304.16 |
| daily, $m = 365$ | 1.030453 | 10,304.53 |
| continuous | $e^{0.03} = 1.030455$ | 10,304.55 |

Annual compounding is CHF 4.55 short of the limit; the error formula predicts $10{,}000 \times e^{0.03} \times 0.03^2/2 = 4.64$.
Monthly is CHF 0.39 short, about 12 times less.

### Example 2: a mortgage quote

A loan quoted at 2.4% with monthly compounding has an effective annual rate of $(1 + 0.024/12)^{12} - 1 = 2.427\%$ and a continuous
rate of $12 \ln(1.002) = 2.398\%$.

### Example 3: a negative rate

Swiss sight deposits paid −0.75% from 2015 to 2022. Continuously compounded for seven years, CHF 1 million becomes
$10^6 \times e^{-0.0525} = 948{,}854$; with annual compounding at −0.75% it becomes $10^6 \times 0.9925^7 = 948{,}667$. The formulas work for
negative rates; with $r < 0$, more frequent compounding means a slightly smaller loss.

### Example 4: doubling time

At 6% continuous, $T = 0.693/0.06 = 11.6$ years. At 6% annual, $T = \ln 2/\ln 1.06 = 11.9$ years; the rule of 72 says 12.

## 4. Finance Application

### Where it matters

- **Derivatives pricing.** Black–Scholes, binomial trees ($R = e^{r\Delta t}$) and Monte Carlo use continuously compounded rates
  because they add over time and make discount factors simple exponentials.
- **Yield curves.** Zero rates are stored as continuous rates so that discount factors are $e^{-r(t)t}$ and forward rates are
  differences of $r(t)t$ (module 01).
- **Comparing quotes.** Loans, deposits and bonds quote rates under different conventions; converting all of them to effective annual or
  continuous rates makes them comparable. Consumer-credit regulation (the effective annual rate in Swiss and EU law) exists for this reason.
- **Returns.** A continuously compounded rate of return is a log return (00/01), so everything said there about time aggregation applies.

### The usual misreadings

- "3% is 3%." Without the compounding convention the number is incomplete; at higher rates the differences become large.
- "Continuous compounding pays much more." For rates of a few percent the difference from monthly compounding is a few hundredths of a percent.
- "The rule of 72 is exact." It is an approximation for annual compounding near 8%; at 1% it overstates the doubling time by about 3%
  (72 vs 69.7 years).

## 5. Python Examples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp

pd.set_option("display.float_format", lambda x: f"{x:,.6f}")
```

### The limit, symbolically

```python
r_sym, m_sym = sp.symbols("r m", positive=True)
print("limit (1 + r/m)^m        =", sp.limit((1 + r_sym / m_sym) ** m_sym, m_sym, sp.oo))
print("series of m*log(1 + r/m) =", sp.series(m_sym * sp.log(1 + r_sym / m_sym), m_sym, sp.oo, 3))
```

SymPy confirms $\lim (1 + r/m)^m = e^r$ and the expansion $m\ln(1 + r/m) = r - \frac{r^2}{2m} + \frac{r^3}{3m^2} - \dots$ used for the
error estimate.

### Example 1 in code, with the error estimate

```python
principal = 10_000      # CHF
rate = 0.03             # nominal annual rate, decimal

rows = {}
for name, m in [("annual", 1), ("semi-annual", 2), ("quarterly", 4), ("monthly", 12), ("daily", 365)]:
    factor = (1 + rate / m) ** m
    rows[name] = {"m": m, "growth factor": factor, "end value CHF": principal * factor,
                  "shortfall CHF": principal * (np.exp(rate) - factor),
                  "estimate CHF": principal * np.exp(rate) * rate**2 / (2 * m)}
rows["continuous"] = {"m": np.inf, "growth factor": np.exp(rate), "end value CHF": principal * np.exp(rate),
                      "shortfall CHF": 0.0, "estimate CHF": 0.0}
pd.DataFrame(rows).T
```

The estimate $e^r r^2/(2m)$ matches the actual shortfall to within a few percent at every frequency, and both fall in proportion to $1/m$.

### Converting between conventions

```python
def to_continuous(rate, m):
    # nominal rate compounded m times a year -> continuous rate
    return m * np.log1p(rate / m)

def from_continuous(r_c, m):
    # continuous rate -> nominal rate compounded m times a year
    return m * np.expm1(r_c / m)

def effective_annual(rate, m):
    return (1 + rate / m) ** m - 1

quotes = {"mortgage 2.4% monthly": (0.024, 12), "bond 3% semi-annual": (0.03, 2),
          "deposit 1.5% annual": (0.015, 1), "credit card 12% monthly": (0.12, 12)}
pd.DataFrame({name: {"effective annual": effective_annual(q, m), "continuous": to_continuous(q, m),
                     "back to quote": from_continuous(to_continuous(q, m), m)}
              for name, (q, m) in quotes.items()}).T
```

The round trip reproduces each quote. For low rates the conventions differ by a few basis points; for the 12% credit card the effective
annual rate is 12.68%, more than half a percentage point above the quote.

### Large rates make the convention matter

```python
for rate in [0.01, 0.05, 0.10, 0.25, 1.00]:
    annual = 1 + rate
    monthly = (1 + rate / 12) ** 12
    cont = np.exp(rate)
    print(f"rate {rate:5.0%}: annual {annual:.4f}, monthly {monthly:.4f}, continuous {cont:.4f}, "
          f"continuous - annual = {cont - annual:.4f}")
```

At 1% the conventions are indistinguishable; at 25% continuous compounding adds 3.4 percentage points; at 100% the factor is
$e = 2.718$ instead of 2. The last line is Jacob Bernoulli's original question that led to the number $e$.

### Doubling times

```python
rates = np.array([0.01, 0.02, 0.04, 0.06, 0.08, 0.10, 0.15, 0.20])
pd.DataFrame({
    "rate": rates,
    "exact, annual compounding": np.log(2) / np.log1p(rates),
    "rule of 72": 72 / (100 * rates),
    "exact, continuous": np.log(2) / rates,
    "rule of 69.3": 69.3 / (100 * rates),
}).set_index("rate").round(2)
```

The rule of 72 is almost exact near 8% and a little too high at low rates, where the exact answer for annual compounding approaches
$69.3/(100r)$ plus about a third of a year. For continuous compounding the rule of 69.3 is exact by construction.

## 6. Visualizations

```python
rate, years = 0.25, 3     # a high rate makes the steps visible
t = np.linspace(0, years, 2000)
plt.figure(figsize=(10, 5))
for m in [1, 4, 12]:
    steps = np.floor(t * m) / m
    plt.step(t, (1 + rate / m) ** (steps * m), where="post", label=f"m = {m}")
plt.plot(t, np.exp(rate * t), "k", lw=2, label="continuous e^(rt)")
plt.title("Growth of CHF 1 at a nominal 25% a year")
plt.xlabel("years")
plt.ylabel("value (CHF)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

With annual compounding the balance jumps once a year; with quarterly and monthly compounding the jumps get smaller and more
frequent, and the step functions approach the smooth exponential from below.

```python
ms = np.logspace(0, 6, 60)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for rate in [0.03, 0.10, 0.30]:
    a_m = (1 + rate / ms) ** ms
    axes[0].semilogx(ms, a_m / np.exp(rate), label=f"r = {rate:.0%}")
    axes[1].loglog(ms, np.exp(rate) - a_m, label=f"r = {rate:.0%}")
    axes[1].loglog(ms, np.exp(rate) * rate**2 / (2 * ms), "k:", lw=1)
axes[0].set_title("(1 + r/m)^m relative to e^r")
axes[0].set_xlabel("compounding periods per year m")
axes[0].set_ylabel("ratio")
axes[1].set_title("Shortfall e^r - (1 + r/m)^m (dotted: e^r r²/2m)")
axes[1].set_xlabel("compounding periods per year m")
axes[1].set_ylabel("shortfall per CHF 1")
for ax in axes:
    ax.legend()
    ax.grid(True, alpha=0.3, which="both")
plt.tight_layout()
plt.show()
```

On the left the ratio climbs to 1 from below for every rate. On the right the shortfall is a straight line of slope −1 on log–log axes:
it falls in proportion to $1/m$, as the error formula (dotted) predicts. Higher rates start further away, in proportion to $r^2$.

```python
rates = np.linspace(0.005, 0.25, 200)
exact_annual = np.log(2) / np.log1p(rates)
plt.figure(figsize=(9, 5))
plt.plot(rates * 100, (72 / (100 * rates) / exact_annual - 1) * 100, label="rule of 72")
plt.plot(rates * 100, (69.3 / (100 * rates) / exact_annual - 1) * 100, "--", label="rule of 69.3")
plt.axhline(0, color="k", lw=0.8)
plt.title("Error of doubling-time rules against exact annual compounding")
plt.xlabel("rate (% per year)")
plt.ylabel("relative error (%)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"rule of 72 is exact at r = {rates[np.argmin(np.abs(72 / (100 * rates) / exact_annual - 1))]:.2%}")
print(f"largest error of the rule of 72 between 1% and 20%: "
      f"{np.max(np.abs(72 / (100 * rates) / exact_annual - 1)[(rates >= 0.01) & (rates <= 0.20)]):.1%}")
```

The rule of 72 is exact near 7.9% and stays within about 5% of the true doubling time for rates between 1% and 20%. The rule of
69.3 is exact only for continuous compounding; applied to annual compounding it underestimates the doubling time, more so at high rates.

## 8. Solutions / Checks

```python
# 1.
P, rate, T = 50_000, 0.02, 10
print(f"1. annual {P * (1 + rate)**T:,.2f}, quarterly {P * (1 + rate / 4)**(4 * T):,.2f}, "
      f"continuous {P * np.exp(rate * T):,.2f}")

# 2.
r_c = 0.04
print(f"2. effective annual {np.expm1(r_c):.5f}, semi-annual {from_continuous(r_c, 2):.5f}")

# 3.
bank1 = effective_annual(0.02, 12)
bank2 = 0.0202
print(f"3. bank 1 effective {bank1:.5f}, bank 2 {bank2:.5f}, difference on CHF 100,000: "
      f"{100_000 * (bank2 - bank1):.2f} CHF")
```

**3.** Monthly compounding at 2.00% gives an effective 2.0184%, so the bank with 2.02% annual pays CHF 1.6 more on CHF 100,000. The
difference is real but tiny: at low rates, the compounding convention is worth a fraction of a basis point to a few basis points, and the
quoted number matters far more.

```python
# 4.
for rate in [0.05, -0.05]:
    seq = np.array([(1 + rate / m) ** m for m in [1, 2, 4, 12, 52, 365]])
    print(f"4. r = {rate:+.2f}: {np.round(seq, 6)}, e^r = {np.exp(rate):.6f}, increasing: {np.all(np.diff(seq) > 0)}")

# 5.
r_eff = (1000 / 870) ** (1 / 7) - 1
r_c = np.log(1000 / 870) / 7
print(f"5. annual {r_eff:.5f}, continuous {r_c:.5f}, ln(1 + r_eff) = {np.log1p(r_eff):.5f}")

# 7.
rate = 0.3
for m in [1, 2, 12]:
    exact = np.exp(rate) - (1 + rate / m) ** m
    first = np.exp(rate) * rate**2 / (2 * m)
    second = np.exp(rate) * (rate**2 / (2 * m) - rate**3 / (3 * m**2) - rate**4 / (8 * m**2))
    print(f"7. m = {m:2d}: exact {exact:.6f}, first order {first:.6f}, second order {second:.6f}")
```

**4.** For $r = -5\%$ the sequence also increases with $m$, towards $e^{-0.05} = 0.951229$ from below: more frequent compounding of a negative
rate charges each sub-period's loss on a balance that has already shrunk, so the total loss is slightly smaller.

**5.** Both yields describe the same price: $r_c = \ln(1 + r_{\text{eff}})$ to all printed digits.

**6.** With $x = r/m > 0$, $\ln(1 + x) < x$ gives $m\ln(1 + r/m) < r$, and exponentiating (an increasing function) gives $(1 + r/m)^m < e^r$.

**7.** $\ln a_m = r - \frac{r^2}{2m} + \frac{r^3}{3m^2} + O(m^{-3})$, so $a_m = e^r e^{-\varepsilon}$ with $\varepsilon = \frac{r^2}{2m} - \frac{r^3}{3m^2}$, and
$e^r - a_m = e^r(\varepsilon - \varepsilon^2/2 + \dots) = e^r\big(\frac{r^2}{2m} - \frac{r^3}{3m^2} - \frac{r^4}{8m^2}\big) + O(m^{-3})$.
The printed numbers show the second-order formula closer to the exact shortfall than the first-order one.

## 9. Conclusion

- $(1 + r/m)^m$ increases in $m$ and converges to $e^r$; the proof is one line with logarithms.
- The shortfall of $m$-period compounding is about $e^r r^2/(2m)$: small at low rates, and falling like $1/m$.
- Continuous compounding solves $dB/dt = rB$; discount factors are $e^{-rt}$, and the continuous rate is the annual log return.
- Convert conventions through the annual growth factor: $1 + r_{\text{eff}} = (1 + r_m/m)^m = e^{r_c}$.
- Doubling time is $\ln 2/r_c$; the rule of 72 is a good approximation for annual compounding near 8%.
